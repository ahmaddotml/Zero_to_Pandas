# Exception Handling

**Exception handling** allows a Python program to deal with errors that occur while the program is running.

Normally, when Python encounters an error, it stops the program and displays an error message.

For example:

```python id="q8m4x2"
number = int(input("Enter a number: "))

print(number)
```

If the user enters:

```text id="k3v7p1"
hello
```

Python cannot convert `"hello"` into an integer, so it raises a `ValueError`.

Without exception handling, the program may stop unexpectedly.

Exception handling allows us to **detect errors, handle them properly, and allow the program to continue when appropriate**.

Python provides four important keywords for exception handling:

* `try`
* `except`
* `else`
* `finally`

---

# 1. `try`

The `try` block contains code that **might cause an exception**.

### Syntax

```python id="m5r2v8"
try:
    # code that might cause an error
```

The code inside `try` is executed normally.

If an exception occurs, Python looks for a matching `except` block.

### Example

```python id="x7c1n4"
try:
    number = int(input("Enter a number: "))
    print(number)
except:
    print("Something went wrong.")
```

If the user enters:

```text id="a9q3m6"
25
```

Output:

```text id="z2k8p5"
25
```

If the user enters:

```text id="w4v1c7"
hello
```

Output:

```text id="p6n3x9"
Something went wrong.
```

The `try` block allows Python to attempt the operation safely.

---

# 2. `except`

The `except` block is used to **handle an exception**.

### Basic Syntax

```python id="r8m2q5"
try:
    # risky code
except:
    # code to handle the error
```

For example:

```python id="c4x7n1"
try:
    number = int("hello")
except:
    print("Invalid number.")
```

Output:

```text id="v9k3m6"
Invalid number.
```

When the conversion fails, Python jumps from the `try` block to the `except` block.

---

## Handling a Specific Exception

It is usually better to catch a **specific exception** instead of using a bare `except`.

For example, converting invalid text to an integer raises `ValueError`:

```python id="m6p2w8"
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Please enter a valid number.")
```

If the user enters:

```text id="q3x7v1"
hello
```

Output:

```text id="t8n4c5"
Please enter a valid number.
```

This is better than:

```python id="j5r9m2"
try:
    number = int(input("Enter a number: "))
except:
    print("Something went wrong.")
```

because the program clearly identifies the type of error it expects to handle.

---

# Common Python Exceptions

Python has many built-in exception types.

Some commonly encountered exceptions include:

| Exception             | Meaning                                            |
| --------------------- | -------------------------------------------------- |
| `ValueError`          | A value has an inappropriate or invalid format     |
| `TypeError`           | An operation is performed on an inappropriate type |
| `ZeroDivisionError`   | A number is divided by zero                        |
| `IndexError`          | A sequence index does not exist                    |
| `KeyError`            | A dictionary key does not exist                    |
| `NameError`           | A variable or name is not defined                  |
| `FileNotFoundError`   | A requested file does not exist                    |
| `AttributeError`      | An object does not have the requested attribute    |
| `ImportError`         | An import cannot be performed                      |
| `ModuleNotFoundError` | The requested module cannot be found               |

---

## Handling `ZeroDivisionError`

Dividing a number by zero causes a `ZeroDivisionError`.

```python id="f7k3p9"
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("You cannot divide by zero.")
```

Output:

```text id="b2v8m4"
You cannot divide by zero.
```

---

## Handling `IndexError`

An `IndexError` occurs when you try to access an index that does not exist.

```python id="n4x7q2"
numbers = [10, 20, 30]

try:
    print(numbers[5])
except IndexError:
    print("Index does not exist.")
```

Output:

```text id="c8m1r6"
Index does not exist.
```

---

## Handling `KeyError`

A `KeyError` can occur when trying to access a dictionary key that doesn't exist.

```python id="w5p9k3"
user = {
    "name": "Ali",
    "age": 20
}

try:
    print(user["email"])
except KeyError:
    print("Email key does not exist.")
```

Output:

```text id="y2v6n8"
Email key does not exist.
```

---

# 3. Multiple `except` Blocks

You can use multiple `except` blocks to handle different types of exceptions separately.

### Example

```python id="q6m2x8"
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print(result)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Number cannot be zero.")
```

If the user enters:

```text id="a4p7v9"
hello
```

Output:

```text id="j8n3c5"
Please enter a valid number.
```

If the user enters:

```text id="r5x1m6"
0
```

Output:

```text id="k2v8q4"
Number cannot be zero.
```

If the user enters:

```text id="t7c3p9"
20
```

Output:

```text id="m1q6w8"
5.0
```

Python checks the `except` blocks and uses the first one that matches the exception.

---

# 4. Catching Multiple Exceptions

If multiple exception types should be handled in the same way, you can place them in a tuple.

```python id="v8m4x2"
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except (ValueError, ZeroDivisionError):
    print("Invalid input.")
```

This handles both:

* `ValueError`
* `ZeroDivisionError`

with the same error message.

---

# 5. Accessing the Exception Object

You can store the exception object using `as`.

### Syntax

```python id="c5q9m1"
try:
    # code
except ExceptionType as error:
    # use error
```

### Example

```python id="x7v2p8"
try:
    number = int("hello")
except ValueError as error:
    print("Error:", error)
```

Example output:

```text id="n4k6w3"
Error: invalid literal for int() with base 10: 'hello'
```

The variable `error` contains information about the exception.

This can be useful for debugging or displaying a more specific error message.

---

# 6. `else`

The `else` block is executed **only when no exception occurs in the `try` block**.

### Syntax

```python id="r3m8q5"
try:
    # code that might cause an exception
except:
    # code if an exception occurs
else:
    # code if no exception occurs
```

### Example

```python id="p6x1v9"
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid number.")
else:
    print("You entered:", number)
```

If the user enters:

```text id="k4n7c2"
25
```

Output:

```text id="z8m3q6"
You entered: 25
```

If the user enters:

```text id="w1v5p9"
hello
```

Output:

```text id="q3x7r4"
Invalid number.
```

The `else` block only executes when the `try` block completes successfully.

---

## Why Use `else`?

You can put successful-operation code inside `else`.

For example:

```python id="m8q2c5"
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("The number is:", number)
    print("Processing complete.")
```

This makes the structure clear:

```text id="j6v9x3"
try    → attempt the operation
except → handle an error
else   → execute if there was no error
```

---

# 7. `finally`

The `finally` block contains code that should run **regardless of whether an exception occurs or not**.

### Syntax

```python id="t2m7p4"
try:
    # code
except:
    # error handling
finally:
    # code that always runs
```

### Example

```python id="x9c4v6"
try:
    number = int(input("Enter a number: "))
    print(number)
except ValueError:
    print("Invalid number.")
finally:
    print("Program finished.")
```

If the user enters:

```text id="k8q2m5"
25
```

Output:

```text id="v3n7p1"
25
Program finished.
```

If the user enters:

```text id="r4x9c2"
hello
```

Output:

```text id="m6w1q8"
Invalid number.
Program finished.
```

The `finally` block runs in both cases.

---

# 8. Why Use `finally`?

`finally` is especially useful for **cleanup operations**.

For example, if you open a file or database connection, you may want to make sure it is closed after you're finished.

```python id="p5m8x2"
file = None

try:
    file = open("data.txt", "r")
    content = file.read()
except FileNotFoundError:
    print("File not found.")
finally:
    if file is not None:
        file.close()

print("Finished.")
```

The `finally` block gives you a place for cleanup code that should happen whether the operation succeeds or fails.

For file handling, Python's `with` statement is usually preferred because it handles closing the file automatically, but `finally` is still an important general exception-handling concept.

---

# 9. Using `try`, `except`, `else`, and `finally` Together

All four keywords can be used in one structure.

### Syntax

```python id="w7c2n9"
try:
    # code that might raise an exception

except SomeException:
    # handle the exception

else:
    # runs when no exception occurs

finally:
    # always runs
```

### Example

```python id="a3m8v5"
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Number cannot be zero.")

else:
    print("Result:", result)

finally:
    print("Operation finished.")
```

If the user enters:

```text id="c6q1x8"
20
```

Output:

```text id="r4m9p2"
Result: 5.0
Operation finished.
```

If the user enters:

```text id="v8k3n5"
0
```

Output:

```text id="j2w7c4"
Number cannot be zero.
Operation finished.
```

If the user enters:

```text id="q5x9m1"
hello
```

Output:

```text id="p7c2v6"
Please enter a valid number.
Operation finished.
```

The flow can be visualized as:

```text id="f4n8q2"
try
 │
 ├── Exception occurs ──→ except
 │
 └── No exception ──────→ else
                             
finally
 │
 └── Runs after the above
```

# 10. Nested Exception Handling

Exception handling can also be nested.

For example:

```python id="m2x7v9"
try:
    number = int(input("Enter a number: "))

    try:
        result = 100 / number
        print(result)

    except ZeroDivisionError:
        print("Cannot divide by zero.")

except ValueError:
    print("Invalid number.")
```

Here, one `try` block exists inside another.

Nested exception handling can be useful when different parts of a program require different error-handling logic. However, it should not be overused because excessive nesting can make code difficult to understand.

---

# 11. Raising Exceptions with `raise`

Python also allows you to intentionally raise an exception using the `raise` statement.

For example:

```python id="v9c4m2"
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
```

Output:

```text id="x6q1n8"
ValueError: Age cannot be negative.
```

This is useful when you want to enforce rules in your program.

For example, a function can validate its input:

```python id="p3m7w5"
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")

    return age
```

Then:

```python id="k8v2x4"
age = set_age(-5)
```

raises a `ValueError`.

---

# 12. Custom Exceptions

Python allows you to create your own exception classes.

Custom exceptions are useful in larger applications when you want to represent specific application-level errors.

### Example

```python id="r5n9c3"
class InsufficientBalanceError(Exception):
    pass
```

You can then use it:

```python id="x2m6v8"
balance = 100
withdraw = 150

if withdraw > balance:
    raise InsufficientBalanceError("Not enough balance.")
```

The custom exception inherits from Python's built-in `Exception` class.

For simple programs, built-in exceptions such as `ValueError`, `TypeError`, and `FileNotFoundError` are usually sufficient.

---

# 13. Exception Handling with Functions

Exception handling can be placed inside functions.

```python id="q7m3x1"
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero."
```

Now:

```python id="v4c8p2"
print(divide(10, 2))
print(divide(10, 0))
```

Output:

```text id="w9n5k6"
5.0
Cannot divide by zero.
```

You can also allow the exception to propagate to the code that calls the function:

```python id="m1x7q4"
def divide(a, b):
    return a / b

try:
    result = divide(10, 0)
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

This approach can be useful when the caller is responsible for deciding how the error should be handled.

---

# 14. Common Mistakes

## Using a Bare `except`

You can write:

```python id="p8c2m5"
try:
    number = int(input("Enter a number: "))
except:
    print("Error.")
```

Although this works, it is usually better to catch specific exceptions:

```python id="x6v9r3"
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Please enter a valid number.")
```

Specific exception handling makes the program easier to understand and prevents you from accidentally hiding unexpected programming errors.

---

## Putting Too Much Code in `try`

Avoid putting unrelated operations inside one large `try` block.

Instead of:

```python id="m4q8v1"
try:
    # lots of unrelated code
    ...
except:
    print("Something went wrong.")
```

keep the `try` block focused on the operation that might raise the exception:

```python id="r7x2c6"
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid number.")
```

This makes it easier to determine which operation caused the problem.

---

## Using `finally` for Normal Program Logic

`finally` is mainly intended for cleanup or actions that must happen regardless of success or failure.

For example:

```python id="n5k9p3"
try:
    file = open("data.txt")
except FileNotFoundError:
    print("File not found.")
finally:
    print("Finished.")
```

Use `else` for successful-operation logic and `finally` for code that should always execute.

### Example

```python id="n6x2p9"
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("You entered:", number)
finally:
    print("Program finished.")
```

> **Key idea:** Exception handling allows your Python programs to **handle runtime errors gracefully instead of crashing unexpectedly**. Use `try` for code that might fail, `except` to handle specific errors, `else` for code that should run when no error occurs, and `finally` for cleanup or code that must run regardless of the result.
