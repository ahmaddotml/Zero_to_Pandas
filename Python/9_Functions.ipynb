# Functions

**Functions** are reusable blocks of code that perform a specific task. They allow you to organize your program into smaller, manageable parts and avoid repeating the same code multiple times.

For example, instead of writing the same calculation several times:

```python
number1 = 10
number2 = 20
print(number1 + number2)

number1 = 5
number2 = 15
print(number1 + number2)
```

You can create a function:

```python
def add_numbers(a, b):
    print(a + b)

add_numbers(10, 20)
add_numbers(5, 15)
```

Output:

```text
30
20
```

Functions are useful because they make code:

* **Reusable** — the same function can be called multiple times.
* **Organized** — large programs can be divided into smaller parts.
* **Readable** — function names can describe what the code does.
* **Maintainable** — changes can be made in one place instead of many places.
* **Testable** — individual functions can be tested separately.

---

# 1. Defining Functions

A function is defined using the `def` keyword.

### Syntax

```python
def function_name():
    # code to execute
```

The `def` keyword tells Python that you are creating a function.

### Example

```python
def greet():
    print("Hello, Python!")
```

The function has now been **defined**, but the code inside it has not executed yet.

To execute the function, you need to **call** it:

```python
greet()
```

Output:

```text
Hello, Python!
```

---

## Function Definition and Function Call

There are two important concepts:

### Function Definition

Creating the function:

```python
def greet():
    print("Hello!")
```

### Function Call

Running the function:

```python
greet()
```

A function can be called multiple times:

```python
def greet():
    print("Hello!")

greet()
greet()
greet()
```

Output:

```text
Hello!
Hello!
Hello!
```

The same block of code is reused three times.

---

## Functions with Multiple Statements

A function can contain multiple statements.

```python
def introduce():
    name = "Ali"
    age = 20

    print("Name:", name)
    print("Age:", age)
```

Calling the function:

```python
introduce()
```

Output:

```text
Name: Ali
Age: 20
```

All statements with the same indentation belong to the function.

---

# 2. Parameters

**Parameters** are variables written inside the parentheses when defining a function.

They allow a function to receive data from outside.

### Example

```python
def greet(name):
    print("Hello,", name)
```

Here, `name` is a **parameter**.

When we call the function:

```python
greet("Ali")
```

Output:

```text
Hello, Ali
```

The value `"Ali"` is passed to the `name` parameter.

---

## Multiple Parameters

A function can have multiple parameters.

```python
def add_numbers(a, b):
    print(a + b)
```

Calling the function:

```python
add_numbers(10, 20)
```

Output:

```text
30
```

Here:

```text
a = 10
b = 20
```

You can call the same function with different values:

```python
add_numbers(5, 7)
add_numbers(100, 200)
add_numbers(2, 8)
```

Output:

```text
12
300
10
```

---

## Parameter Names

Parameter names should describe the data that the function expects.

For example, instead of:

```python
def calculate(x, y):
    print(x + y)
```

You could write:

```python
def calculate(price, tax):
    print(price + tax)
```

The second version is easier to understand because the parameter names describe what the values represent.

---

# 3. Arguments

An **argument** is the actual value passed to a function when the function is called.

For example:

```python
def greet(name):
    print("Hello,", name)

greet("Ali")
```

Here:

* `name` is the **parameter**.
* `"Ali"` is the **argument**.

A simple way to remember the difference is:

```text
Parameter → variable in the function definition
Argument  → actual value passed to the function
```

---

## Positional Arguments

Arguments can be passed based on their position.

```python
def introduce(name, age):
    print("Name:", name)
    print("Age:", age)

introduce("Ali", 20)
```

Output:

```text
Name: Ali
Age: 20
```

Python assigns the values according to their position:

```text
name = "Ali"
age = 20
```

The order matters when using positional arguments.

For example:

```python
introduce(20, "Ali")
```

would assign:

```text
name = 20
age = "Ali"
```

which is probably not what the function expects.

---

## Keyword Arguments

You can also specify the parameter name when passing an argument.

```python
def introduce(name, age):
    print("Name:", name)
    print("Age:", age)

introduce(age=20, name="Ali")
```

Output:

```text
Name: Ali
Age: 20
```

The order does not matter when using keyword arguments because Python knows which value belongs to which parameter.

---

## Mixing Positional and Keyword Arguments

You can combine positional and keyword arguments.

```python
def introduce(name, age, city):
    print(name, age, city)

introduce("Ali", age=20, city="Peshawar")
```

Output:

```text
Ali 20 Peshawar
```

However, positional arguments must come before keyword arguments.

Correct:

```python
introduce("Ali", age=20, city="Peshawar")
```

Incorrect:

```python
introduce(name="Ali", 20, city="Peshawar")
```

# 4. Return Values

A function can send a result back to the code that called it using the `return` statement.

### Example

```python
def add_numbers(a, b):
    return a + b
```

Now you can store the result:

```python
result = add_numbers(10, 20)

print(result)
```

Output:

```text
30
```

The function calculates the value and returns it to the caller.

---

## `return` vs `print`

It is important to understand the difference between `return` and `print`.

### Using `print()`

```python
def add_numbers(a, b):
    print(a + b)

result = add_numbers(10, 20)
```

The function prints the result, but it does not return the result.

### Using `return`

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)

print(result)
```

Now the result can be stored and used later.

For example:

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)

double = result * 2

print(double)
```

Output:

```text
60
```

This is why `return` is usually preferred when a function calculates a value that another part of the program needs.

---

## Returning Different Data Types

A function can return many types of values.

### Returning a String

```python
def get_name():
    return "Ali"

name = get_name()
print(name)
```

Output:

```text
Ali
```

### Returning a Boolean

```python
def is_adult(age):
    return age >= 18

print(is_adult(20))
```

Output:

```text
True
```

### Returning a List

```python
def get_numbers():
    return [1, 2, 3, 4, 5]

numbers = get_numbers()

print(numbers)
```

Output:

```text
[1, 2, 3, 4, 5]
```

---

## Returning Multiple Values

Python allows a function to return multiple values.

```python
def get_user():
    name = "Ali"
    age = 20

    return name, age
```

You can store the returned values separately:

```python
name, age = get_user()

print(name)
print(age)
```

Output:

```text
Ali
20
```

Python internally returns these values as a tuple.

---

## `return` Ends a Function

When Python reaches a `return` statement, the function immediately stops executing.

```python
def check_number(number):
    if number > 0:
        return "Positive"

    print("This will not run for positive numbers.")
```

For example:

```python
result = check_number(10)
print(result)
```

Output:

```text
Positive
```

Once `return "Positive"` executes, Python leaves the function.

---

## Function Without `return`

A function does not have to return a value.

```python
def greet():
    print("Hello!")

greet()
```

Such a function performs an action but does not explicitly return a result.

If a function reaches the end without a `return` statement, Python automatically returns `None`.

```python
def greet():
    print("Hello!")

result = greet()

print(result)
```

Output:

```text
Hello!
None
```

# 5. Default Arguments

A **default argument** is a parameter that already has a default value.

If the caller does not provide a value for that parameter, Python uses the default value.

### Syntax

```python
def function_name(parameter=default_value):
    # code
```

### Example

```python
def greet(name="Guest"):
    print("Hello,", name)
```

Calling the function without an argument:

```python
greet()
```

Output:

```text
Hello, Guest
```

Calling it with an argument:

```python
greet("Ali")
```

Output:

```text
Hello, Ali
```

When `"Ali"` is provided, it replaces the default value `"Guest"`.

---

## Multiple Default Arguments

A function can have multiple parameters with default values.

```python
def introduce(name="Guest", age=18):
    print("Name:", name)
    print("Age:", age)
```

Calling:

```python
introduce()
```

Output:

```text
Name: Guest
Age: 18
```

Calling:

```python
introduce("Ali", 20)
```

Output:

```text
Name: Ali
Age: 20
```

You can also provide only some values using keyword arguments:

```python
introduce(age=25)
```

Output:

```text
Name: Guest
Age: 25
```

---

## Required Parameters and Default Parameters

Parameters without default values are generally required.

```python
def greet(name, message="Hello"):
    print(message, name)
```

Here:

* `name` is a required parameter.
* `message` has a default value.

You must provide `name`:

```python
greet("Ali")
```

Output:

```text
Hello Ali
```

You can also replace the default:

```python
greet("Ali", "Welcome")
```

Output:

```text
Welcome Ali
```

---

## Important Rule for Default Parameters

Parameters with default values must come **after** parameters without default values.

Correct:

```python
def greet(name, message="Hello"):
    print(message, name)
```

Incorrect:

```python
def greet(message="Hello", name):
    print(message, name)
```

Python will raise a syntax error because a required parameter cannot follow a default parameter.

# 6. Scope

**Scope** determines where a variable can be accessed in a Python program.

A variable created inside a function usually has a different scope from a variable created outside the function.

The two important scopes for beginners are:

* **Local scope**
* **Global scope**

---

## Local Scope

A variable created inside a function is called a **local variable**.

It can normally only be accessed inside that function.

```python
def greet():
    name = "Ali"
    print(name)

greet()
```

Output:

```text
Ali
```

The variable `name` belongs to the function.

Trying to access it outside:

```python
def greet():
    name = "Ali"

greet()

print(name)
```

will result in a `NameError` because `name` does not exist in the global scope.

---

## Global Scope

A variable created outside a function belongs to the **global scope**.

```python
name = "Ali"

def greet():
    print(name)

greet()
```

Output:

```text
Ali
```

The function can read the global variable because it is available outside the function.

---

## Local and Global Variables with the Same Name

A local variable can have the same name as a global variable.

```python
name = "Ali"

def greet():
    name = "Ahmed"
    print(name)

greet()

print(name)
```

Output:

```text
Ahmed
Ali
```

Inside the function, Python uses the local variable:

```text
name = "Ahmed"
```

Outside the function, Python uses the global variable:

```text
name = "Ali"
```

They are separate variables.

---

# Modifying a Global Variable

By default, assigning a value to a variable inside a function creates a local variable.

For example:

```python
count = 10

def update_count():
    count = 20
    print(count)

update_count()

print(count)
```

Output:

```text
20
10
```

The assignment inside the function does not change the global `count`.

If you intentionally need to modify the global variable, Python provides the `global` keyword.

```python
count = 10

def update_count():
    global count
    count = 20

update_count()

print(count)
```

Output:

```text
20
```

However, using many global variables can make programs harder to understand and maintain. In most situations, it is better to pass values into functions and return results.

# Practical Example

Let's create a function that calculates the total price of a product.

```python
def calculate_total(price, quantity, tax=0.05):
    subtotal = price * quantity
    total = subtotal + (subtotal * tax)

    return total
```

Now call the function:

```python
total = calculate_total(100, 3)

print(total)
```

Output:

```text
315.0
```

Here:

* `price` is a parameter.
* `quantity` is a parameter.
* `tax` is a default parameter.
* `100`, `3` are arguments.
* `subtotal` is a local variable.
* `return total` sends the calculated value back to the caller.

You can also provide a different tax rate:

```python
total = calculate_total(100, 3, 0.10)

print(total)
```

Output:

```text
330.0
```

# Practical Example with Multiple Functions

Functions become especially useful when a program is divided into several smaller tasks.

```python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    return a / b


print(add(10, 5))
print(subtract(10, 5))
print(multiply(10, 5))
print(divide(10, 5))
```

Output:

```text
15
5
50
2.0
```

Each function performs one specific task.

This approach makes the program easier to understand and maintain.

# Common Mistakes

## Forgetting to Call the Function

Defining a function does not automatically execute it.

```python
def greet():
    print("Hello!")
```

Nothing is printed until you call it:

```python
greet()
```

---

## Forgetting `return`

Incorrect:

```python
def add(a, b):
    a + b
```

The expression is calculated but its result is not returned.

Correct:

```python
def add(a, b):
    return a + b
```

---

## Using the Wrong Number of Arguments

If a function requires two parameters:

```python
def add(a, b):
    return a + b
```

You need to provide two arguments:

```python
add(10, 20)
```

Calling:

```python
add(10)
```

will cause a `TypeError` because the required `b` argument is missing.

---

## Incorrect Default Parameter Order

Incorrect:

```python
def greet(message="Hello", name):
    print(message, name)
```

Correct:

```python
def greet(name, message="Hello"):
    print(message, name)
```

Required parameters should come before parameters with default values.
