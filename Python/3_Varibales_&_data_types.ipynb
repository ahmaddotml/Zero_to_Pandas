# Variables & Data Types

In Python, **variables** are used to store data in memory. Every value in Python has a **data type**, which tells us what kind of data it represents.

---

## 1. Variables

A variable is a name that refers to a value.

```python
name = "Ali"
age = 20
height = 5.8
is_student = True
```

Here:

* `name` stores a string.
* `age` stores an integer.
* `height` stores a float.
* `is_student` stores a Boolean value.

Python does not require us to declare the data type explicitly.

```python
x = 10
x = "Hello"
```

A variable can even be assigned a different type of value later.

---

## 2. `int`

`int` represents **whole numbers** without a decimal point.

```python
age = 20
temperature = -5
count = 100
```

You can perform mathematical operations with integers:

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a // b)
```

---

## 3. `float`

`float` represents **numbers with decimal points**.

```python
price = 99.99
height = 5.8
temperature = -2.5
```

Example:

```python
x = 10.5
y = 2.5

print(x + y)
print(x * y)
```

---

## 4. `str`

`str` represents **text**, also called a string.

Strings are written inside single or double quotes.

```python
name = "Ali"
message = 'Hello World'
```

You can combine strings using `+`:

```python
first_name = "Ali"
last_name = "Khan"

full_name = first_name + " " + last_name
print(full_name)
```

You can also use **f-strings** to insert variables into text:

```python
name = "Ali"
age = 20

print(f"My name is {name} and I am {age} years old.")
```

---

## 5. `bool`

`bool` represents a logical value. It has only two possible values:

* `True`
* `False`

```python
is_student = True
is_raining = False
```

Boolean values are commonly used with conditions:

```python
age = 20

is_adult = age >= 18
print(is_adult)
```

Output:

```text
True
```

---

## 6. `None`

`None` represents the **absence of a value**.

It is different from `0`, `False`, or an empty string.

```python
result = None

print(result)
```

`None` is often used when a variable does not have a value yet:

```python
username = None

# Later, we can assign a value
username = "Ali"
```

The type of `None` is `NoneType`.

---

## 7. Checking Types with `type()`

The `type()` function tells us the data type of a value or variable.

```python
age = 20
price = 10.5
name = "Ali"
is_student = True
result = None

print(type(age))
print(type(price))
print(type(name))
print(type(is_student))
print(type(result))
```

Output:

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
<class 'NoneType'>
```

---

## 8. Type Conversion

**Type conversion** means changing a value from one data type to another.

### `int()`

Converts a value to an integer.

```python
x = int(10.8)
print(x)
```

Output:

```text
10
```

You can also convert a string containing a whole number:

```python
age = int("20")
print(age)
```

---

### `float()`

Converts a value to a floating-point number.

```python
x = float(10)
print(x)
```

Output:

```text
10.0
```

You can also convert a numeric string:

```python
price = float("99.99")
print(price)
```

---

### `str()`

Converts a value to a string.

```python
age = 20
message = "I am " + str(age) + " years old."

print(message)
```

---

### `bool()`

Converts a value to a Boolean.

```python
print(bool(1))
print(bool(0))
print(bool("Hello"))
print(bool(""))
```

Output:

```text
True
False
True
False
```

---

## Quick Summary

| Data Type  | Description       | Example   |
| ---------- | ----------------- | --------- |
| `int`      | Whole numbers     | `10`      |
| `float`    | Decimal numbers   | `10.5`    |
| `str`      | Text              | `"Hello"` |
| `bool`     | `True` or `False` | `True`    |
| `NoneType` | No value          | `None`    |

### Key Functions

```python
type(value)   # Check the type

int(value)    # Convert to integer
float(value)  # Convert to float
str(value)    # Convert to string
bool(value)   # Convert to Boolean
```

> **Key idea:** Python variables store values, and every value has a data type. Understanding data types and type conversion is essential for writing correct Python programs.
