# Strings

A **string** is a sequence of characters used to represent text in Python.

Strings can be created using single quotes `'...'`, double quotes `"..."`, or triple quotes `'''...'''` / `"""..."""`.

```python
name = "Python"
message = 'Hello World'
```

---

## 1. String Indexing

**Indexing** is used to access individual characters in a string.

Python uses **zero-based indexing**, which means the first character has index `0`.

```python
word = "Python"

print(word[0])
print(word[1])
print(word[2])
```

Output:

```text
P
y
t
```

### Negative Indexing

Python also supports negative indexing.

The last character has index `-1`, the second-last has `-2`, and so on.

```python
word = "Python"

print(word[-1])
print(word[-2])
print(word[-3])
```

Output:

```text
n
o
h
```

### Indexing Example

For the string:

```python
word = "Python"
```

| Character      |  `P` |  `y` |  `t` |  `h` |  `o` |  `n` |
| -------------- | ---: | ---: | ---: | ---: | ---: | ---: |
| Positive index |  `0` |  `1` |  `2` |  `3` |  `4` |  `5` |
| Negative index | `-6` | `-5` | `-4` | `-3` | `-2` | `-1` |

## 2. String Slicing

**Slicing** is used to extract a portion of a string.

The basic syntax is:

```python
string[start:stop]
```

The `start` index is included, but the `stop` index is excluded.

```python
word = "Python"

print(word[0:3])
```

Output:

```text
Pyt
```

### Omitting Start or Stop

You can leave out the start or stop index.

```python
word = "Python"

print(word[:3])
print(word[2:])
print(word[:])
```

Output:

```text
Pyt
thon
Python
```

### Slicing with a Step

You can provide a third value:

```python
string[start:stop:step]
```

Example:

```python
word = "Python"

print(word[0:6:2])
```

Output:

```text
Pto
```

### Reverse a String

A step of `-1` reverses the string.

```python
word = "Python"

print(word[::-1])
```

Output:

```text
nohtyP
```

## 3. Common String Methods

Python provides many built-in **string methods** for working with text.

### `upper()`

Converts all characters to uppercase.

```python
text = "hello world"

print(text.upper())
```

Output:

```text
HELLO WORLD
```

### `lower()`

Converts all characters to lowercase.

```python
text = "HELLO WORLD"

print(text.lower())
```

Output:

```text
hello world
```

### `title()`

Capitalizes the first letter of each word.

```python
text = "hello world"

print(text.title())
```

Output:

```text
Hello World
```

### `capitalize()`

Capitalizes the first character of the string.

```python
text = "hello world"

print(text.capitalize())
```

Output:

```text
Hello world
```

### `strip()`

Removes whitespace from the beginning and end of a string.

```python
text = "   Hello World   "

print(text.strip())
```

Output:

```text
Hello World
```

### `replace()`

Replaces one piece of text with another.

```python
text = "I like Java"

print(text.replace("Java", "Python"))
```

Output:

```text
I like Python
```

### `split()`

Splits a string into a list.

```python
text = "Python is easy"

words = text.split()

print(words)
```

Output:

```text
['Python', 'is', 'easy']
```

You can also specify a separator:

```python
data = "apple,banana,mango"

print(data.split(","))
```

### `join()`

Joins multiple strings together.

```python
words = ["Python", "is", "fun"]

sentence = " ".join(words)

print(sentence)
```

Output:

```text
Python is fun
```

### `find()`

Returns the index of the first occurrence of a substring.

```python
text = "Hello Python"

print(text.find("Python"))
```

Output:

```text
6
```

If the text is not found, `find()` returns `-1`.

### `count()`

Counts how many times a substring appears.

```python
text = "banana"

print(text.count("a"))
```

Output:

```text
3
```

### `startswith()` and `endswith()`

Check whether a string starts or ends with specific text.

```python
filename = "python_notes.txt"

print(filename.startswith("python"))
print(filename.endswith(".txt"))
```

Output:

```text
True
True
```

## 4. f-Strings

**f-strings** provide an easy way to insert variables and expressions inside strings.

Place `f` before the opening quote and put variables inside `{}`.

```python
name = "Ali"
age = 20

print(f"My name is {name} and I am {age} years old.")
```

Output:

```text
My name is Ali and I am 20 years old.
```

### Expressions in f-Strings

You can also put expressions inside `{}`.

```python
a = 10
b = 5

print(f"The sum is {a + b}")
```

Output:

```text
The sum is 15
```

### Formatting Numbers

f-strings can also format numbers.

```python
price = 99.9876

print(f"Price: ${price:.2f}")
```

Output:

```text
Price: $99.99
```

Here, `.2f` means to display the number with **2 digits after the decimal point**.

| Concept     | Purpose                       | Example                  |
| ----------- | ----------------------------- | ------------------------ |
| Indexing    | Access one character          | `text[0]`                |
| Slicing     | Extract part of a string      | `text[1:4]`              |
| `upper()`   | Convert to uppercase          | `text.upper()`           |
| `lower()`   | Convert to lowercase          | `text.lower()`           |
| `strip()`   | Remove surrounding whitespace | `text.strip()`           |

| `find()`    | Find a substring              | `text.find("Python")`    |

> **Key idea:** Strings are sequences of characters. Python provides indexing and slicing to access parts of strings, many useful methods to manipulate text, and f-strings to create formatted text easily.
