# Conditional Statements

**Conditional statements** allow a Python program to make decisions based on whether a condition is `True` or `False`.

For example:

```python id="q1v9s3"
age = 20

if age >= 18:
    print("You are an adult.")
```

Python uses **indentation** to define the code that belongs to a condition.

---

## 1. `if`

The `if` statement executes a block of code **only when a condition is `True`**.

### Syntax

```python id="k5r2mn"
if condition:
    # code to execute
```

### Example

```python id="0c7g1d"
age = 20

if age >= 18:
    print("You are an adult.")
```

Output:

```text id="j1n8qx"
You are an adult.
```

If the condition is `False`, the code inside the `if` block is skipped.

```python id="z2f5wk"
age = 15

if age >= 18:
    print("You are an adult.")
```

There is no output because `age >= 18` is `False`.

---

## 2. `else`

The `else` statement provides an alternative block of code.

It executes when the `if` condition is `False`.

### Syntax

```python id="5p6q7r"
if condition:
    # if condition is True
else:
    # if condition is False
```

### Example

```python id="y8w2kd"
age = 15

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

Output:

```text id="4r1j8m"
You are a minor.
```

---

## 3. `elif`

`elif` means **"else if"**.

It allows you to check multiple conditions.

### Syntax

```python id="0r8t2v"
if condition1:
    # code
elif condition2:
    # code
else:
    # code
```

Python checks the conditions **from top to bottom** and executes the first condition that is `True`.

### Example

```python id="q5m7xa"
marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
```

Output:

```text id="2w8y4p"
Grade B
```

### Important

You can use:

* One `if`
* Zero or more `elif`
* Zero or one `else`

Example:

```python id="n4k6zp"
temperature = 30

if temperature > 35:
    print("Very hot")
elif temperature > 25:
    print("Warm")
elif temperature > 15:
    print("Cool")
else:
    print("Cold")
```

---

# 4. Nested Conditions

A **nested condition** means putting one conditional statement inside another conditional statement.

### Example

```python id="7s4v9c"
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("You can enter.")
    else:
        print("You need an ID.")
else:
    print("You are not old enough.")
```

Output:

```text id="z5w2na"
You can enter.
```

The second `if` is inside the first `if`, so it is called a **nested `if` statement**.

---

## Nested `if` with Multiple Conditions

You can have multiple levels of nested conditions.

```python id="p7k3mx"
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login successful.")
    else:
        print("Incorrect password.")
else:
    print("Unknown user.")
```

Output:

```text id="2x9m4v"
Login successful.
```

---

# Combining Conditions

Instead of nesting conditions, you can often combine them using logical operators such as `and`, `or`, and `not`.

For example, this nested condition:

```python id="c8v1qn"
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("You can enter.")
```

Can be written more simply as:

```python id="f2m6rx"
if age >= 18 and has_id:
    print("You can enter.")
```

Both approaches can work, but combining conditions can make simple logic easier to read.

# Conditional Expressions

For simple conditions, Python also supports a **conditional expression**, sometimes called a ternary expression.

### Syntax

```python id="3j7xqk"
value_if_true if condition else value_if_false
```

### Example

```python id="v5n2cb"
age = 20

status = "Adult" if age >= 18 else "Minor"

print(status)
```

Output:

```text id="8r3q1m"
Adult
```

---

# Truthy and Falsy Values

Python conditions do not always have to be explicit comparisons.

Some values are treated as `True` or `False` in a condition.

### Falsy values include:

```python id="5w8n3p"
False
None
0
0.0
""
[]
()
{}
set()
```

For example:

```python id="m2k7vx"
name = ""

if name:
    print("Name is provided.")
else:
    print("Name is empty.")
```

Output:

```text id="b9q4ds"
Name is empty.
```

A non-empty string is considered truthy:

```python id="n6p3yc"
name = "Ali"

if name:
    print("Name is provided.")
```

Output:

```text id="a1v7zk"
Name is provided.
```

# Common Comparison Conditions

Conditional statements are often used with comparison operators:

```python id="w4r8mt"
x = 10

if x == 10:
    print("Equal")

if x != 5:
    print("Not equal")

if x > 5:
    print("Greater than 5")

if x < 20:
    print("Less than 20")

if x >= 10:
    print("Greater than or equal to 10")

if x <= 10:
    print("Less than or equal to 10")
```

---

# Practical Example

Let's create a simple age classification program:

```python id="q7m2kx"
age = 25

if age < 0:
    print("Invalid age")
elif age < 13:
    print("Child")
elif age < 18:
    print("Teenager")
elif age < 60:
    print("Adult")
else:
    print("Senior")
```

Output:

```text id="t5c9vz"
Adult
```

---

# Quick Summary

| Statement   | Purpose                                                      |
| ----------- | ------------------------------------------------------------ |
| `if`        | Executes code when a condition is `True`                     |
| `elif`      | Checks another condition if previous conditions were `False` |
| `else`      | Executes when all previous conditions are `False`            |
| Nested `if` | Places one condition inside another                          |

### Basic Structure

```python id="8v4m1x"
if condition:
    # code

elif another_condition:
    # code

else:
    # code
```

> **Key idea:** Conditional statements allow your Python program to **make decisions**. Use `if` for the first condition, `elif` for additional conditions, `else` for the fallback case, and nested conditions when one decision depends on another.
