# Modules & Packages

**Modules and packages** allow Python programs to organize code into separate, reusable components.

As programs become larger, putting everything into a single Python file can make the code difficult to understand and maintain. Modules and packages solve this problem by allowing related code to be separated into different files and folders.

A **module** is usually a Python file (`.py`) containing functions, classes, variables, or other Python code.

A **package** is a directory containing related Python modules. Modern Python packages commonly include an `__init__.py` file, although Python also supports namespace packages without one.

For example, a project could have this structure:

```text id="q7x4m2"
my_project/
│
├── main.py
│
└── utilities/
    ├── __init__.py
    ├── math_tools.py
    └── string_tools.py
```

Here:

* `main.py` is the main program.
* `math_tools.py` is a module.
* `string_tools.py` is another module.
* `utilities` is a package.
* `__init__.py` marks the package and can also contain package initialization code.

Modules and packages make it easier to:

* Reuse code.
* Organize large projects.
* Separate different responsibilities.
* Avoid repeating code.
* Use Python's standard library.
* Use third-party libraries created by other developers.

---

# 1. Modules

A **module** is a Python file containing code that can be imported and reused in another Python file.

For example, suppose you create a file called:

```text id="f3q5v7"
math_tools.py
```

Inside the file:

```python id="z1s4ac"
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b
```

You can use these functions from another file.

For example, `main.py`:

```python id="2n7j5k"
import math_tools

print(math_tools.add(10, 5))
print(math_tools.subtract(10, 5))
```

Output:

```text id="x8v2q1"
15
5
```

The `math_tools.py` file is a **module**.

---

# 2. `import`

The `import` statement is used to import a module into your Python program.

### Syntax

```python id="c5x8m2"
import module_name
```

### Example

Python provides a built-in module called `math`.

```python id="g4p9w1"
import math

print(math.sqrt(25))
```

Output:

```text id="n6d2r8"
5.0
```

The module is imported using:

```python id="0a7k3c"
import math
```

Then the function is accessed using the module name:

```python id="5w9m1b"
math.sqrt(25)
```

The dot `.` is used to access something inside the module.

---

## Importing Multiple Modules

You can import multiple modules.

```python id="j2v6p8"
import math
import random
import datetime
```

You can then use them separately:

```python id="s8c3y5"
import math
import random

print(math.sqrt(16))
print(random.randint(1, 10))
```

---

## Importing a Module with an Alias

You can give an imported module a shorter or alternative name using `as`.

### Syntax

```python id="r7x4m9"
import module_name as alias
```

### Example

```python id="d2k8p1"
import math as m

print(m.sqrt(25))
```

Output:

```text id="v5n9q3"
5.0
```

Here:

```text id="3k7s2a"
math → module name
m    → alias
```

Aliases can make frequently used modules easier to reference.

A common example is:

```python id="6p4m8v"
import datetime as dt
```

Then:

```python id="q1r7w5"
today = dt.date.today()

print(today)
```

---

# 3. `from ... import`

Instead of importing the entire module, you can import a specific function, class, or variable from a module.

### Syntax

```python id="k9c3x6"
from module_name import name
```

### Example

```python id="p7m2v4"
from math import sqrt

print(sqrt(25))
```

Output:

```text id="w4n8s1"
5.0
```

You don't need to write:

```python
math.sqrt(25)
```

because `sqrt` was imported directly.

---

## Importing Multiple Items

You can import multiple items from the same module.

```python id="h6q2z8"
from math import sqrt, pi

print(sqrt(25))
print(pi)
```

Output:

```text id="m3x7c5"
5.0
3.141592653589793
```

---

## Using `as` with `from ... import`

You can also give an imported item an alias.

```python id="y8p4k1"
from math import sqrt as square_root

print(square_root(36))
```

Output:

```text id="t5r2n7"
6.0
```

Here, `sqrt` is imported as `square_root`.

---

# 4. `import` vs `from ... import`

Both approaches can be used to access functionality from a module.

### Using `import`

```python id="f1m8q4"
import math

print(math.sqrt(25))
```

### Using `from ... import`

```python id="w6c2y9"
from math import sqrt

print(sqrt(25))
```

The first approach keeps the module name visible:

```python id="3v7n1k"
math.sqrt()
```

The second imports the specific item directly:

```python id="8q4m6p"
sqrt()
```

Using `import module` can sometimes make code clearer because it shows exactly where a function or class comes from.

---

# 5. Importing Everything with `*`

Python also allows this syntax:

```python id="x3k8m2"
from math import *
```

This imports many names from the module into the current namespace.

For example:

```python id="n7p4v6"
from math import *

print(sqrt(25))
print(pi)
```

Although this works, it is generally **not recommended**.

The problem is that it can make it difficult to know where a particular name came from and can cause naming conflicts.

For example:

```python id="q5z1c8"
from module_a import *
from module_b import *
```

If both modules contain a variable or function with the same name, one may overwrite the other.

It is usually better to write:

```python id="m2r7x9"
import math

print(math.sqrt(25))
```

or:

```python id="a8v3k6"
from math import sqrt

print(sqrt(25))
```

---

# 6. Standard Library

Python comes with a large collection of modules called the **Python Standard Library**.

These modules are included with Python, so you normally do not need to install them separately.

The standard library provides tools for many common programming tasks, including:

* Mathematics
* Random numbers
* Dates and times
* File handling
* Operating system operations
* JSON
* Regular expressions
* Statistics
* Internet and networking
* Data structures

---

## `math`

The `math` module provides mathematical functions and constants.

```python id="v4n8s2"
import math

print(math.sqrt(16))
print(math.pow(2, 3))
print(math.pi)
```

Output:

```text id="j6x1c9"
4.0
8.0
3.141592653589793
```

---

## `random`

The `random` module provides functions for generating random values.

```python id="r8m3q5"
import random

number = random.randint(1, 10)

print(number)
```

The output can be any integer between `1` and `10`.

You can also randomly choose an item from a list:

```python id="k2v7p4"
import random

fruits = ["apple", "banana", "orange"]

fruit = random.choice(fruits)

print(fruit)
```

---

## `datetime`

The `datetime` module is used to work with dates and times.

```python id="c6y9m1"
from datetime import date

today = date.today()

print(today)
```

Example output:

```text id="n4w8s3"
2026-08-14
```

The exact output depends on the current date.

---

## `os`

The `os` module provides functionality for interacting with the operating system.

```python id="p3x7k5"
import os

print(os.getcwd())
```

This prints the current working directory.

You can also list files and directories:

```python id="q9m2v6"
import os

print(os.listdir())
```

---

## `json`

The `json` module is used to work with JSON data.

```python id="w5r8c2"
import json

data = {
    "name": "Ali",
    "age": 20
}

json_data = json.dumps(data)

print(json_data)
```

Output:

```text id="t7n3m9"
{"name": "Ali", "age": 20}
```

---

## `statistics`

The `statistics` module provides functions for working with statistical data.

```python id="x4p6q8"
import statistics

numbers = [10, 20, 30, 40, 50]

print(statistics.mean(numbers))
```

Output:

```text id="m1v5z7"
30
```

---

# 7. Creating Your Own Module

You can create your own modules to organize your code.

Suppose you create:

```text id="b8k2p5"
calculator.py
```

Inside the file:

```python id="r3m7x1"
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b
```

Now create another file:

```text id="v6q9c4"
main.py
```

Import the module:

```python id="y2n8s5"
import calculator

print(calculator.add(10, 5))
print(calculator.multiply(10, 5))
```

Output:

```text id="h4p7m2"
15
50
```

This allows you to separate related functionality into its own file.

# 8. Packages

A **package** is a collection of related Python modules organized inside a directory.

For example:

```text id="q8m3v7"
my_project/
│
├── main.py
│
└── calculator/
    ├── __init__.py
    ├── basic.py
    └── advanced.py
```

Here:

```text id="z5r1k9"
calculator/
```

is the package.

Inside it:

```text id="f7c2x4"
basic.py
advanced.py
```

are modules.

---

## Importing from a Package

Suppose `basic.py` contains:

```python id="m9v4q6"
def add(a, b):
    return a + b
```

You can import it from `main.py`:

```python id="p2x7k1"
from calculator.basic import add

print(add(10, 20))
```

Output:

```text id="w8c5n3"
30
```

You can also import the module itself:

```python id="r6m1v9"
from calculator import basic

print(basic.add(10, 20))
```

---

# 9. `__init__.py`

The `__init__.py` file is associated with Python packages.

For example:

```text id="a4k8p2"
calculator/
├── __init__.py
├── basic.py
└── advanced.py
```

The file can be empty:

```python id="z3x6m1"
```

or it can contain package initialization code.

Modern Python also supports **namespace packages**, so `__init__.py` is not strictly required for every package. However, it is still commonly used when creating regular Python packages.

---

# 10. Installing External Packages

Python's standard library provides many useful features, but sometimes you need functionality that is not included with Python.

In these situations, you can install **third-party packages**.

Third-party packages are created and maintained by developers outside the Python standard library.

Examples include packages for:

* Data analysis
* Web development
* Machine learning
* Scientific computing
* Image processing
* API development
* Automation

A common tool for installing Python packages is **`pip`**.

---

## Using `pip`

The basic command is:

```bash
pip install package_name
```

For example:

```bash
pip install requests
```

After installation, you can use the package in Python:

```python id="h8v2q5"
import requests

response = requests.get("https://example.com")

print(response.status_code)
```

The package must be installed in the Python environment where your program is running.

---

## Using `python -m pip`

A more reliable form is often:

```bash
python -m pip install requests
```

On systems where `python` refers to Python 2 or is unavailable, you may use:

```bash
python3 -m pip install requests
```

This makes it clearer which Python interpreter is being used to run `pip`.

---

## Installing a Specific Version

You can install a specific package version:

```bash
python -m pip install requests==2.32.4
```

You can also specify a minimum version:

```bash
python -m pip install "requests>=2.30"
```

Version requirements are especially important in larger projects where compatibility matters.

---

# 11. Upgrading a Package

To upgrade an installed package:

```bash
python -m pip install --upgrade requests
```

This tells `pip` to install a newer available version that satisfies the package requirements.

---

# 12. Uninstalling a Package

You can remove an installed package using:

```bash
python -m pip uninstall requests
```

`pip` will normally ask for confirmation before removing the package.

---

# 13. Viewing Installed Packages

You can see the packages installed in your current Python environment with:

```bash
python -m pip list
```

Example:

```text id="q4m8x2"
Package    Version
---------- -------
pip        25.x
requests   2.x
```

You can also inspect information about a particular package:

```bash
python -m pip show requests
```

---

# 14. Requirements Files

When working on a project, you may need to record the external packages that your program depends on.

A common approach is to create a file called:

```text id="n7c3v5"
requirements.txt
```

For example:

```text id="j2m8q4"
requests==2.32.4
```

Another developer can then install the project's dependencies with:

```bash
python -m pip install -r requirements.txt
```

This is useful when sharing projects with other developers or deploying applications.

# 15. Virtual Environments

It is recommended to use a **virtual environment** for many Python projects.

A virtual environment creates an isolated Python environment for a project.

For example:

```bash id="v5x1k8"
python -m venv .venv
```

This creates a virtual environment named `.venv`.

You then activate it according to your operating system.

### Windows

```bash id="r9m4p2"
.venv\Scripts\activate
```

### Linux/macOS

```bash id="c6k8w1"
source .venv/bin/activate
```

Once activated, packages installed with `pip` are associated with that environment rather than being installed globally.

For example:

```bash id="x3q7m5"
python -m pip install requests
```

Virtual environments help prevent different projects from interfering with each other's dependencies.

---

# 16. Importing an External Package

After installing a package, you normally import it using Python's `import` statement.

For example, after installing `requests`:

```python id="p8v2m6"
import requests
```

You can then use the functionality provided by the package.

The name used with `import` is not always identical to the name used with `pip install`, so you should check the package's documentation when necessary.

---

# 17. Standard Library vs External Packages

There is an important difference between Python's standard library and third-party packages.

### Standard Library

Included with Python:

```python id="k4m9x1"
import math
import random
import json
import os
```

Normally, you do not need to install these packages separately.

### External Packages

Usually installed using `pip`:

```bash id="w7c3p5"
python -m pip install requests
```

Then imported:

```python id="n2v8q6"
import requests
```

A simple comparison:

| Type                | Description                               | Installation                     |
| ------------------- | ----------------------------------------- | -------------------------------- |
| Standard library    | Included with Python                      | Usually no installation required |
| Third-party package | Created outside Python's standard library | Usually installed with `pip`     |
| Your own module     | Python file created by you                | No package installation required |
| Package             | Collection of related modules             | May be your own or third-party   |

# Practical Example

Suppose you want to create a small project that generates random numbers and calculates their square roots.

You can use Python's standard library:

```python id="b6r2m8"
import random

number = random.randint(1, 100)
```

* `math` calculates the square root.

* No external package installation is required.

# Common Mistakes

## Forgetting to Install an External Package

If you write:

```python id="v3m7q1"
import requests
```

without having the package installed in the current environment, Python may raise:

```text id="c8x2n5"
ModuleNotFoundError
```

You can usually install the package with:

```bash id="p4k9w6"
python -m pip install requests
```

---

## Installing a Package but Using the Wrong Python Environment

You may install a package successfully but still get `ModuleNotFoundError`.

This can happen when `pip` installs the package into a different Python environment from the one running your program.

Using:

```bash id="m7q2v8"
python -m pip install package_name
```

helps associate `pip` with the Python interpreter specified by `python`.

Virtual environments can also help prevent this problem.

---

## Naming Your File After a Standard Module

Avoid naming your own Python files after standard library modules.

For example, don't create:

```text id="x5n8c3"
random.py
```

and then write:

```python id="q1m6v9"
import random
```

Your local file may interfere with Python's attempt to import the standard `random` module.

Use descriptive names such as:

```text id="a7k2p4"
random_numbers.py
```

instead.

---

# Quick Summary

| Concept             | Purpose                                            |
| ------------------- | -------------------------------------------------- |
| Module              | A Python file containing reusable code             |
| Package             | A collection of related Python modules             |
| `import`            | Imports a module                                   |
| `from ... import`   | Imports specific items from a module               |
| `as`                | Creates an alias for an imported name              |
| Standard library    | Modules included with Python                       |
| Third-party package | External functionality created by other developers |
| `pip`               | Tool commonly used to install Python packages      |
| `requirements.txt`  | Records project dependencies                       |
| Virtual environment | Provides an isolated Python environment            |

### Importing a Module

```python id="u8m3q5"
import math

print(math.sqrt(25))
```
