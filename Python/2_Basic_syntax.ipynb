# 🐍 Python Basic Syntax

Python syntax refers to the rules and structure used to write Python programs. Python is designed with a clean and readable syntax, which makes code easier to understand and maintain.

In this section, we focus only on the **basic structure of Python code**. Mathematical operations, data types, and other concepts are covered separately.

## 1. Statements

A **statement** is an instruction written in Python.

```python
print("Hello, Python!")
```

Python executes the statement from top to bottom.

Multiple statements can be written on separate lines:

```python
print("Hello")
print("Welcome")
print("Python")
```

---

## 2. Indentation

**Indentation** means the spaces at the beginning of a line.

Python uses indentation to define a **block of code**.

```python
if condition:
    print("This is inside the block")
```

The indented line belongs to the `if` block.

Python conventionally uses **4 spaces** for indentation.

> **Important:** Indentation is part of Python's syntax. It is not only used for formatting.

---

## 3. Code Blocks

A **code block** is a group of statements that belong to a particular structure.

Python uses indentation instead of `{}` to define blocks.

```python
if condition:
    statement_1
    statement_2
```

The statements with the same indentation level belong to the same block.

---

## 4. The Colon `:`

A colon is used to indicate that an **indented block of code follows**.

It is commonly used with structures such as:

```python
if condition:
    statement
```

Later, you will also see it with:

```python
for item in items:
    statement
```

and:

```python
def function():
    statement
```

The colon tells Python that the next indented lines belong to that structure.

---

## 5. Comments

Comments are notes written inside the code for humans to read. Python ignores comments when executing the program.

A single-line comment begins with `#`.

```python
# This is a comment
print("Hello")
```

Comments can also be placed after a statement:

```python
print("Hello")  # Display a message
```

---

## 6. Case Sensitivity

Python is **case-sensitive**.

This means uppercase and lowercase letters are treated differently.

```python
name
Name
NAME
```

These are considered different names in Python.

For example:

```python
print("Hello")
```

is valid, while:

```python
Print("Hello")
```

is not the same because `print` and `Print` are different.

---

## 7. Naming

Python uses names for variables, functions, classes, and other objects.

Names can contain:

* Letters
* Numbers
* Underscores `_`

A name **cannot begin with a number**.

### Valid names

```python
name
student_name
student1
_total
```

### Invalid names

```python
1student
student-name
student name
```

Python also has reserved keywords that cannot be used as ordinary names.

Examples include:

```text
if
else
for
while
def
class
return
import
```

---

## 8. Naming Convention

Python commonly uses **snake_case** for names.

```python
student_name
first_name
total_students
```

Words are written in lowercase and separated using underscores.

This is a **convention**, not a syntax requirement, but following it makes Python code easier to read.

---

## 9. Line Structure

Normally, each statement is written on its own line.

```python
print("First")
print("Second")
print("Third")
```

Python determines the end of a statement from the end of the line.

Unlike some other programming languages, Python does not normally require a semicolon `;` after every statement.

---

## 10. Line Continuation

A long statement can be split across multiple lines when appropriate.

Parentheses can be used to continue an expression across lines:

```python
result = (
    first_value
    + second_value
    + third_value
)
```

This can make long code easier to read.

---

## 11. Quotes

Quotes are used to represent text in Python.

Python supports both single and double quotation marks:

```python
"Hello"
'Hello'
```

The opening and closing quotes should match.

```python
"Hello"
```

```python
'Hello'
```

Triple quotes can also be used for multi-line text:

```python
"""
This is
multiple lines
of text.
"""
```

---

## 12. Python Syntax Structure

A simple Python structure often looks like this:

```python
# Comment

if condition:
    statement
    statement
```

The basic pattern is:

**Keyword/structure → Colon `:` → Indented block**

For example:

```python
if condition:
    print("Inside the block")
```

---

# ⭐ Key Syntax Rules

| Rule               | Description                           |
| ------------------ | ------------------------------------- |
| **Indentation**    | Defines blocks of code                |
| **`:`**            | Introduces an indented block          |
| **`#`**            | Starts a comment                      |
| **Case-sensitive** | Uppercase and lowercase are different |
| **New line**       | Normally separates statements         |
| **Quotes**         | Used to represent text                |
| **Naming rules**   | Define how identifiers can be written |
| **Snake_case**     | Common naming convention in Python    |

> ### 🧠 Remember
>
> Python's most important syntax feature is **indentation**. Instead of using curly brackets `{}` to organize blocks of code, Python uses indentation to clearly show the structure of a program.
