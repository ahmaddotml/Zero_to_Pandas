# Python Collections — Methods & Functions

Python's main collection data types are:

* `list`
* `tuple`
* `set`
* `dict`

Each collection has its own methods that are useful for adding, removing, searching, and manipulating data.

---

# 1. Lists

A **list** is an ordered, mutable collection that allows duplicate values.

```python
fruits = ["apple", "banana", "mango"]
```

## Common List Methods

### `append()`

Adds one item to the end of the list.

```python
fruits = ["apple", "banana"]

fruits.append("mango")

print(fruits)
```

Output:

```text
['apple', 'banana', 'mango']
```

---

### `extend()`

Adds multiple items from another iterable to the end of the list.

```python
fruits = ["apple", "banana"]

fruits.extend(["mango", "orange"])

print(fruits)
```

Output:

```text
['apple', 'banana', 'mango', 'orange']
```

---

### `insert()`

Adds an item at a specific index.

```python
fruits = ["apple", "mango"]

fruits.insert(1, "banana")

print(fruits)
```

Output:

```text
['apple', 'banana', 'mango']
```

---

### `remove()`

Removes the first occurrence of a specified value.

```python
fruits = ["apple", "banana", "mango"]

fruits.remove("banana")

print(fruits)
```

Output:

```text
['apple', 'mango']
```

---

### `pop()`

Removes and returns an item.

If no index is provided, it removes the last item.

```python
fruits = ["apple", "banana", "mango"]

item = fruits.pop()

print(item)
print(fruits)
```

Output:

```text
mango
['apple', 'banana']
```

You can also specify an index:

```python
fruits = ["apple", "banana", "mango"]

item = fruits.pop(1)

print(item)
print(fruits)
```

---

### `clear()`

Removes all items from the list.

```python
fruits = ["apple", "banana", "mango"]

fruits.clear()

print(fruits)
```

Output:

```text
[]
```

---

### `index()`

Returns the index of the first occurrence of a value.

```python
fruits = ["apple", "banana", "mango"]

print(fruits.index("banana"))
```

Output:

```text
1
```

---

### `count()`

Counts how many times a value appears.

```python
numbers = [1, 2, 2, 3, 2]

print(numbers.count(2))
```

Output:

```text
3
```

---

### `sort()`

Sorts the list in ascending order.

```python
numbers = [5, 2, 8, 1, 3]

numbers.sort()

print(numbers)
```

Output:

```text
[1, 2, 3, 5, 8]
```

For descending order:

```python
numbers.sort(reverse=True)

print(numbers)
```

---

### `reverse()`

Reverses the order of items in the list.

```python
numbers = [1, 2, 3, 4, 5]

numbers.reverse()

print(numbers)
```

Output:

```text
[5, 4, 3, 2, 1]
```

---

### `copy()`

Creates a shallow copy of the list.

```python
fruits = ["apple", "banana", "mango"]

new_fruits = fruits.copy()

print(new_fruits)
```

---

## Useful Built-in Functions with Lists

### `len()`

Returns the number of items.

```python
numbers = [10, 20, 30, 40]

print(len(numbers))
```

Output:

```text
4
```

### `max()`

Returns the largest value.

```python
numbers = [10, 20, 5, 40]

print(max(numbers))
```

### `min()`

Returns the smallest value.

```python
numbers = [10, 20, 5, 40]

print(min(numbers))
```

### `sum()`

Returns the total of numeric items.

```python
numbers = [10, 20, 30]

print(sum(numbers))
```

Output:

```text
60
```

---

# 2. Tuples

A **tuple** is an ordered and immutable collection.

```python
colors = ("red", "green", "blue")
```

Because tuples are immutable, they have fewer methods than lists.

## Tuple Methods

### `count()`

Counts how many times a value occurs.

```python
numbers = (1, 2, 2, 3, 2)

print(numbers.count(2))
```

Output:

```text
3
```

---

### `index()`

Returns the index of the first occurrence of a value.

```python
colors = ("red", "green", "blue")

print(colors.index("green"))
```

Output:

```text
1
```

---

## Useful Built-in Functions with Tuples

### `len()`

```python
colors = ("red", "green", "blue")

print(len(colors))
```

### `max()`

```python
numbers = (10, 20, 5, 40)

print(max(numbers))
```

### `min()`

```python
numbers = (10, 20, 5, 40)

print(min(numbers))
```

### `sum()`

```python
numbers = (10, 20, 30)

print(sum(numbers))
```

### `sorted()`

Returns a **new list** containing the sorted tuple values.

```python
numbers = (5, 2, 8, 1)

result = sorted(numbers)

print(result)
```

Output:

```text
[1, 2, 5, 8]
```

> **Important:** `sorted()` does not modify the original tuple.

# 3. Sets

A **set** is an unordered collection of unique values.

```python
numbers = {1, 2, 3, 4}
```

Sets are especially useful when you want to remove duplicates or perform mathematical set operations.

## Common Set Methods

### `add()`

Adds one item to the set.

```python
numbers = {1, 2, 3}

numbers.add(4)

print(numbers)
```

---

### `update()`

Adds multiple items to a set.

```python
numbers = {1, 2}

numbers.update([3, 4, 5])

print(numbers)
```

---

### `remove()`

Removes a specified item.

```python
numbers = {1, 2, 3}

numbers.remove(2)

print(numbers)
```

> If the item does not exist, `remove()` raises a `KeyError`.

---

### `discard()`

Removes an item if it exists.

```python
numbers = {1, 2, 3}

numbers.discard(2)

print(numbers)
```

Unlike `remove()`, `discard()` does not raise an error if the item is missing.

---

### `pop()`

Removes and returns an arbitrary item from the set.

```python
numbers = {1, 2, 3}

item = numbers.pop()

print(item)
print(numbers)
```

> Because sets are unordered, you should not rely on which item `pop()` removes.

---

### `clear()`

Removes all items.

```python
numbers = {1, 2, 3}

numbers.clear()

print(numbers)
```

Output:

```text
set()
```

---

## Set Operations

### `union()`

Combines two sets and removes duplicates.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
```

Output:

```text
{1, 2, 3, 4, 5}
```

You can also use `|`:

```python
print(a | b)
```

---

### `intersection()`

Returns items that exist in both sets.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.intersection(b))
```

Output:

```text
{3}
```

The `&` operator can also be used:

```python
print(a & b)
```

---

### `difference()`

Returns items that exist in the first set but not the second.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.difference(b))
```

Output:

```text
{1, 2}
```

The `-` operator can also be used:

```python
print(a - b)
```

---

### `symmetric_difference()`

Returns items that are in either set, but **not in both**.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.symmetric_difference(b))
```

Output:

```text
{1, 2, 4, 5}
```

The `^` operator can also be used:

```python
print(a ^ b)
```

---

### `issubset()`

Checks whether all items of one set exist in another set.

```python
a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))
```

Output:

```text
True
```

---

### `issuperset()`

Checks whether a set contains all items of another set.

```python
a = {1, 2, 3, 4}
b = {1, 2}

print(a.issuperset(b))
```

Output:

```text
True
```

---

### `isdisjoint()`

Checks whether two sets have no common items.

```python
a = {1, 2}
b = {3, 4}

print(a.isdisjoint(b))
```

Output:

```text
True
```

---

# 4. Dictionaries

A **dictionary** stores data as **key-value pairs**.

```python
student = {
    "name": "Ali",
    "age": 20,
    "grade": "A"
}
```

## Common Dictionary Methods

### `get()`

Returns the value associated with a key.

```python
student = {
    "name": "Ali",
    "age": 20
}

print(student.get("name"))
```

Output:

```text
Ali
```

A default value can also be provided:

```python
print(student.get("city", "Not found"))
```

Output:

```text
Not found
```

---

### `keys()`

Returns all dictionary keys.

```python
student = {
    "name": "Ali",
    "age": 20,
    "grade": "A"
}

print(student.keys())
```

---

### `values()`

Returns all dictionary values.

```python
print(student.values())
```

---

### `items()`

Returns all key-value pairs.

```python
print(student.items())
```

---

### `update()`

Adds new key-value pairs or updates existing ones.

```python
student = {
    "name": "Ali",
    "age": 20
}

student.update({
    "age": 21,
    "city": "Peshawar"
})

print(student)
```

---

### `pop()`

Removes a specified key and returns its value.

```python
student = {
    "name": "Ali",
    "age": 20
}

age = student.pop("age")

print(age)
print(student)
```

---

### `popitem()`

Removes and returns the **last inserted key-value pair**.

```python
student = {
    "name": "Ali",
    "age": 20
}

item = student.popitem()

print(item)
print(student)
```

---

### `setdefault()`

Returns the value of a key. If the key does not exist, it adds the key with a default value.

```python
student = {
    "name": "Ali"
}

student.setdefault("age", 20)

print(student)
```

Output:

```text
{'name': 'Ali', 'age': 20}
```

---

### `clear()`

Removes all key-value pairs.

```python
student = {
    "name": "Ali",
    "age": 20
}

student.clear()

print(student)
```

Output:

```text
{}
```

---

### `copy()`

Creates a shallow copy of the dictionary.

```python
student = {
    "name": "Ali",
    "age": 20
}

new_student = student.copy()

print(new_student)
```

---

## Useful Built-in Functions with Dictionaries

### `len()`

Returns the number of key-value pairs.

```python
student = {
    "name": "Ali",
    "age": 20,
    "grade": "A"
}

print(len(student))
```

Output:

```text
3
```

### `sorted()`

Returns a sorted list of dictionary keys.

```python
student = {
    "z": 1,
    "a": 2,
    "m": 3
}

print(sorted(student))
```

Output:

```text
['a', 'm', 'z']
```

# Common Built-in Functions

These functions work with many Python collections:

```python
len()       # Number of items
max()       # Largest item
min()       # Smallest item
sum()       # Sum of numeric items
sorted()    # Returns sorted items
```

Example:

```python
numbers = [10, 5, 30, 20]

print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
print(sorted(numbers))
```

Output:

```text
4
30
5
65
[5, 10, 20, 30]
```

> **Key idea:** Methods belong to a specific collection type, while functions such as `len()`, `max()`, `min()`, `sum()`, and `sorted()` are built-in Python functions that can work with multiple collection types.
