# File Handling

**File handling** allows a Python program to create, read, write, and modify files stored on a computer.

Files are useful for storing data permanently. Unlike variables, which normally exist only while a program is running, files can preserve information after the program finishes.

Python provides built-in tools for working with files, so you do not need an external package for basic file operations.

Common file-handling tasks include:

* Reading files
* Writing files
* Appending data to files
* Creating files
* Checking whether files exist
* Working with directories
* Working with file paths
* Renaming and deleting files

The main function used for opening files is:

```python id="m7x2q4"
open()
```

---

# 1. Opening a File

Python uses the `open()` function to open a file.

### Syntax

```python id="x4n8v2"
open(filename, mode)
```

For example:

```python id="p6k3m9"
file = open("data.txt", "r")
```

Here:

* `"data.txt"` is the file name.
* `"r"` means **read mode**.
* `file` is the file object returned by `open()`.

However, when working with files, it is usually better to use the `with` statement because it automatically closes the file.

```python id="c8m1x5"
with open("data.txt", "r") as file:
    content = file.read()

print(content)
```

This is the recommended pattern for most basic file operations.

---

# 2. File Modes

The second argument of `open()` specifies how the file should be opened.

Common modes include:

| Mode  | Purpose            |
| ----- | ------------------ |
| `"r"` | Read the file      |
| `"w"` | Write to the file  |
| `"a"` | Append to the file |
| `"x"` | Create a new file  |
| `"b"` | Binary mode        |
| `"t"` | Text mode          |
| `"+"` | Read and write     |

The most commonly used modes are:

```text id="z5q2m8"
r → read
w → write
a → append
x → create
```

---

# 3. Reading Files

Reading a file means retrieving the data stored inside it.

Suppose you have a file called:

```text id="q8v3m1"
data.txt
```

with the following content:

```text id="w4x7p2"
Hello, Python!
File handling is useful.
```

You can read it using:

```python id="k6m9c4"
with open("data.txt", "r") as file:
    content = file.read()

print(content)
```

Output:

```text id="n2v8x5"
Hello, Python!
File handling is useful.
```

The `with` statement automatically closes the file when the block finishes.

---

# 4. Reading the Entire File with `read()`

The `read()` method reads the contents of a file.

```python id="r3m7q9"
with open("data.txt", "r") as file:
    content = file.read()

print(content)
```

You can also specify how many characters to read:

```python id="v5x2k8"
with open("data.txt", "r") as file:
    content = file.read(5)

print(content)
```

If the file contains:

```text id="c9p4m1"
Hello, Python!
```

the output will be:

```text id="y7n3q6"
Hello
```

`read(5)` reads up to five characters.

---

# 5. Reading One Line with `readline()`

The `readline()` method reads one line at a time.

Suppose `data.txt` contains:

```text id="w8k2v5"
Apple
Banana
Orange
```

You can read the first line:

```python id="p4m9x1"
with open("data.txt", "r") as file:
    line = file.readline()

print(line)
```

Output:

```text id="f6q3n8"
Apple
```

You can call `readline()` again to read the next line:

```python id="z1v7c4"
with open("data.txt", "r") as file:
    print(file.readline())
    print(file.readline())
```

Output:

```text id="r5m8p2"
Apple

Banana
```

---

# 6. Reading All Lines with `readlines()`

The `readlines()` method reads all lines and returns them as a list.

```python id="x3q7m9"
with open("data.txt", "r") as file:
    lines = file.readlines()

print(lines)
```

If the file contains:

```text id="n8c2v5"
Apple
Banana
Orange
```

the result will look like:

```text id="a4m7x1"
['Apple\n', 'Banana\n', 'Orange\n']
```

The `\n` represents a newline character.

You can loop through the lines:

```python id="k9p3w6"
with open("data.txt", "r") as file:
    lines = file.readlines()

for line in lines:
    print(line.strip())
```

Output:

```text id="v2x8q4"
Apple
Banana
Orange
```

---

# 7. Reading a File Line by Line

Instead of using `readlines()`, you can directly iterate over the file.

```python id="m5q1z8"
with open("data.txt", "r") as file:
    for line in file:
        print(line.strip())
```

This is often a good approach for large files because Python does not need to load the entire file into memory at once.

The `strip()` method removes surrounding whitespace, including the newline character at the end of each line.

---

# 8. Writing Files

Writing means adding or replacing content in a file.

To write to a file, use `"w"` mode.

```python id="c7n2m5"
with open("data.txt", "w") as file:
    file.write("Hello, Python!")
```

If the file does not exist, Python creates it.

If the file already exists, `"w"` mode **replaces its existing contents**.

For example:

```python id="p9x4v6"
with open("data.txt", "w") as file:
    file.write("New content")
```

The previous content of the file is overwritten.

> **Important:** Be careful when using `"w"` mode because it can erase existing file contents.

---

# 9. Writing Multiple Lines

You can use `write()` multiple times:

```python id="r8m3q1"
with open("data.txt", "w") as file:
    file.write("Apple\n")
    file.write("Banana\n")
    file.write("Orange\n")
```

The file will contain:

```text id="x2k7m9"
Apple
Banana
Orange
```

The `\n` character creates a new line.

---

## Using `writelines()`

The `writelines()` method can write multiple strings.

```python id="v6p1c8"
lines = [
    "Apple\n",
    "Banana\n",
    "Orange\n"
]

with open("data.txt", "w") as file:
    file.writelines(lines)
```

The resulting file contains:

```text id="m4q9x2"
Apple
Banana
Orange
```

Note that `writelines()` does not automatically add newline characters, so you generally need to include `\n` yourself.

---

# 10. Appending to a File

The `"a"` mode is used to **append new content to the end of an existing file**.

Unlike `"w"` mode, `"a"` does not replace the existing contents.

Suppose `data.txt` contains:

```text id="z8n3m6"
Apple
Banana
```

You can append another item:

```python id="q4x7p1"
with open("data.txt", "a") as file:
    file.write("\nOrange")
```

The file becomes:

```text id="c5m9v2"
Apple
Banana
Orange
```

Appending is useful for:

* Log files
* Saving new records
* Adding new entries
* Storing user activity

---

# 11. Creating a New File with `x`

The `"x"` mode is used to create a new file.

```python id="m2p8q5"
with open("new_file.txt", "x") as file:
    file.write("This is a new file.")
```

If the file does not exist, it is created.

If the file already exists, Python raises a `FileExistsError`.

```python id="v7c3n9"
with open("new_file.txt", "x") as file:
    file.write("Hello")
```

Using `"x"` is useful when you specifically want to make sure you do not accidentally overwrite an existing file.

---

# 12. Reading and Writing with `r+`

The `"r+"` mode allows you to **read and write** to an existing file.

```python id="k4x8m1"
with open("data.txt", "r+") as file:
    content = file.read()
    print(content)

    file.write("\nNew content")
```

The file must already exist when using `"r+"`.

There are other combined modes such as `"w+"` and `"a+"`, but they behave differently with respect to existing content and the current file position.

For beginners, it is usually clearer to use separate read/write operations unless you specifically need combined access.

---

# 13. File Encoding

Text files are stored using a character encoding.

When working with text files, it is good practice to explicitly specify the encoding when you know which encoding your file uses.

For example:

```python id="s7m2q9"
with open("data.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

When writing:

```python id="x5c8v3"
with open("data.txt", "w", encoding="utf-8") as file:
    file.write("Hello, Python!")
```

`UTF-8` is a common encoding that supports a wide range of characters.

---

# 14. Closing Files

When you use:

```python id="q3m7x1"
file = open("data.txt", "r")
```

you should close the file after you finish:

```python id="v8p2k5"
file.close()
```

For example:

```python id="m4x9c6"
file = open("data.txt", "r")

content = file.read()

file.close()

print(content)
```

However, using `with` is preferred:

```python id="r1q7n3"
with open("data.txt", "r") as file:
    content = file.read()

print(content)
```

The `with` statement automatically closes the file when the block finishes, even if an exception occurs.

---

# 15. Working with Paths

A **path** tells Python where a file or directory is located.

For example:

```text id="x6m2v8"
data.txt
```

is a relative path.

An absolute path might look like:

```text id="p9q4c1"
/home/user/documents/data.txt
```

On Windows, an absolute path might look like:

```text id="k7n3m5"
C:\Users\User\Documents\data.txt
```

Paths can become complicated when working across different operating systems, so Python provides the `pathlib` module.

---

# 16. The `pathlib` Module

`pathlib` is the modern Python approach for working with file and directory paths.

Import `Path`:

```python id="v3x8q2"
from pathlib import Path
```

Create a path:

```python id="m6p1k9"
file_path = Path("data.txt")

print(file_path)
```

Output:

```text id="n4c7x5"
data.txt
```

You can then use the path with `open()`:

```python id="q8m2v6"
from pathlib import Path

file_path = Path("data.txt")

with file_path.open("r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

---

# 17. Checking Whether a File Exists

`Path.exists()` checks whether a path exists.

```python id="x5n9m3"
from pathlib import Path

file_path = Path("data.txt")

if file_path.exists():
    print("File exists.")
else:
    print("File does not exist.")
```

Output might be:

```text id="r2c7v8"
File exists.
```

This can be useful before reading, modifying, or deleting a file.

---

# 18. Checking Whether a Path Is a File

You can use `is_file()`:

```python id="p6m4x1"
from pathlib import Path

path = Path("data.txt")

if path.is_file():
    print("This is a file.")
```

---

# 19. Checking Whether a Path Is a Directory

Use `is_dir()` to check whether a path represents a directory.

```python id="k8v2q5"
from pathlib import Path

path = Path("documents")

if path.is_dir():
    print("This is a directory.")
```

---

# 20. Creating Directories

You can create a directory using `mkdir()`.

```python id="w3m7c9"
from pathlib import Path

folder = Path("documents")

folder.mkdir()
```

If the directory already exists, Python will normally raise a `FileExistsError`.

You can avoid that with:

```python id="x9p4n2"
from pathlib import Path

folder = Path("documents")

folder.mkdir(exist_ok=True)
```

Now Python will not raise an error if the directory already exists.

---

# 21. Creating Nested Directories

Suppose you want to create:

```text id="q2m8v5"
project/
└── data/
    └── files/
```

You can use:

```python id="c7x1p4"
from pathlib import Path

folder = Path("project/data/files")

folder.mkdir(parents=True, exist_ok=True)
```

The `parents=True` option allows Python to create the parent directories if they do not already exist.

---

# 22. Joining Paths

You can combine paths using `/` with `pathlib`.

```python id="n5v9m2"
from pathlib import Path

folder = Path("documents")
file_path = folder / "data.txt"

print(file_path)
```

Output:

```text id="w4c8q1"
documents/data.txt
```

This is preferable to manually joining strings such as:

```python id="x7p3m6"
file_path = "documents/" + "data.txt"
```

because `pathlib` handles platform-specific path behavior more appropriately.

---

# 23. Getting File Information

`pathlib` provides useful properties and methods.

For example:

```python id="r9m2x5"
from pathlib import Path

file_path = Path("data.txt")

print(file_path.name)
print(file_path.stem)
print(file_path.suffix)
```

If the file is:

```text id="k6v1q8"
data.txt
```

the output is:

```text id="p3n7c2"
data.txt
data
.txt
```

Here:

* `.name` → complete file name.
* `.stem` → file name without the extension.
* `.suffix` → file extension.

---

# 24. Listing Files in a Directory

You can use `iterdir()` to see the contents of a directory.

```python id="m8q4x1"
from pathlib import Path

folder = Path("documents")

for item in folder.iterdir():
    print(item)
```

This prints the files and directories inside `documents`.

You can also use `glob()` to search for specific file types.

```python id="v2c7n5"
from pathlib import Path

folder = Path("documents")

for file in folder.glob("*.txt"):
    print(file)
```

This finds `.txt` files in the directory.

---

# 25. Deleting Files

You can delete a file using `Path.unlink()`.

```python id="q5m9x3"
from pathlib import Path

file_path = Path("old_data.txt")

if file_path.exists():
    file_path.unlink()
```

Be careful when deleting files because the operation can permanently remove data.

---

# 26. Renaming Files

You can rename a file using `rename()`.

```python id="c8v2m6"
from pathlib import Path

old_name = Path("old.txt")
new_name = Path("new.txt")

old_name.rename(new_name)
```

The file will be renamed from:

```text id="x4p7n1"
old.txt
```

to:

```text id="m9q3v5"
new.txt
```

---

# 27. Moving Files

`Path.rename()` can also be used to move a file to another location.

```python id="r6x1k8"
from pathlib import Path

source = Path("data.txt")
destination = Path("documents/data.txt")

source.rename(destination)
```

For more advanced file operations, Python's `shutil` module also provides functions for copying and moving files and directories.

For example:

```python id="p2m8q4"
import shutil

shutil.copy("data.txt", "backup.txt")
```

This creates a copy of the file.

---

# 28. Relative vs Absolute Paths

There are two common types of paths.

## Relative Path

A relative path describes a location relative to the program's current working directory.

```python id="v7n3c9"
Path("data.txt")
```

or:

```python id="m4x8q2"
Path("documents/data.txt")
```

## Absolute Path

An absolute path describes the complete location of a file.

For example:

```text id="k5p1w7"
/home/user/documents/data.txt
```

On Windows:

```text id="z8m3q6"
C:\Users\User\Documents\data.txt
```

Relative paths are often more convenient for project files because they can work across different machines when the project structure remains the same.

---

# 29. Current Working Directory

Python provides information about the current working directory through `Path.cwd()`.

```python id="q9x4m2"
from pathlib import Path

current_directory = Path.cwd()

print(current_directory)
```

The output depends on where the Python program is running.

You can also change the current working directory using `os.chdir()`, although it is generally better to work with explicit paths when possible.

# Practical Example

Let's create a small program that creates a file, writes data to it, and then reads the data.

```python id="c3m7v1"
from pathlib import Path

file_path = Path("students.txt")

# Write data
with file_path.open("w", encoding="utf-8") as file:
    file.write("Ali\n")
    file.write("Ahmed\n")
    file.write("Sara\n")

# Read data
with file_path.open("r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

Output:

```text id="n8q2x5"
Ali
Ahmed
Sara
```

The program:

1. Creates a path using `Path`.
2. Opens the file in write mode.
3. Writes three names.
4. Closes the file automatically.
5. Opens the file again in read mode.
6. Reads the contents.
7. Prints the contents.

# Practical Example with File Existence

You can combine `pathlib` and file handling to safely work with files.

```python id="v5m9c2"
from pathlib import Path

file_path = Path("data.txt")

if file_path.exists():
    with file_path.open("r", encoding="utf-8") as file:
        content = file.read()

    print(content)
else:
    print("File does not exist.")
```

This prevents your program from attempting to read a file that is not present.

---

# Common Mistakes

## Forgetting to Close a File

Instead of manually managing:

```python id="x2q7m4"
file = open("data.txt", "r")

content = file.read()

file.close()
```

prefer:

```python id="p8m3v6"
with open("data.txt", "r", encoding="utf-8") as file:
    content = file.read()
```

The `with` statement automatically handles closing the file.

---

## Accidentally Overwriting a File

Be careful with:

```python id="m6v1x9"
open("data.txt", "w")
```

The `"w"` mode replaces the existing contents.

If you want to add data without deleting existing content, use:

```python id="q4n8c2"
open("data.txt", "a")
```

---

## Using Incorrect File Paths

This can cause:

```text id="r7p2m5"
FileNotFoundError
```

For example:

```python id="k9x3v1"
with open("missing.txt", "r") as file:
    content = file.read()
```

If `missing.txt` does not exist at the specified location, Python raises an exception.

You can check first:

```python id="w5m8q4"
from pathlib import Path

path = Path("missing.txt")

if path.exists():
    print("File exists.")
else:
    print("File not found.")
```
