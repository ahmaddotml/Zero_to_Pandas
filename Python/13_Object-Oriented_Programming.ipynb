# Object-Oriented Programming (OOP)

**OOP** is a way of organizing Python programs using **classes and objects**.

> **Class = Blueprint | Object = Real thing**

Main concepts:

* Classes
* Objects
* Attributes
* Methods
* Inheritance
* `super()`
* Method overriding

---

# 1. Class

A **class** is a blueprint for creating objects.

```python
class Student:
    pass
```

---

# 2. Object

An **object** is created from a class.

```python
class Student:
    pass

student1 = Student()
student2 = Student()
```

Here:

```text
Student  → Class
student1 → Object
student2 → Object
```

---

# 3. `__init__()` and Attributes

`__init__()` runs automatically when an object is created.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Create an object:

```python
student = Student("Ali", 20)
```

Access attributes:

```python
print(student.name)
print(student.age)
```

Output:

```text
Ali
20
```

### `self`

`self` refers to the **current object**.

```python
self.name = name
```

Each object can have different data:

```python
student1 = Student("Ali", 20)
student2 = Student("Ahmed", 22)
```

---

# 4. Methods

A **method** is a function inside a class.

```python
class Dog:

    def bark(self):
        print("Woof!")
```

Use the method:

```python
dog = Dog()
dog.bark()
```

Output:

```text
Woof!
```

---

# 5. Attributes + Methods

Objects can contain both **data and actions**.

```python
class Dog:

    def __init__(self, name):
        self.name = name

    def bark(self):
        print(f"{self.name} says Woof!")

    def eat(self):
        print(f"{self.name} is eating.")
```

Use it:

```python
dog = Dog("Buddy")

dog.bark()
dog.eat()
```

Output:

```text
Buddy says Woof!
Buddy is eating.
```

Remember:

```text
Attributes → Data
Methods    → Actions
```

# 6. Inheritance

**Inheritance** allows one class to reuse another class's features.

```text
       Animal
       /     \
     Dog     Cat
```

The original class is the **parent**.

The class that inherits is the **child**.

Example:

```python
class Animal:

    def eat(self):
        print("Animal is eating.")


class Dog(Animal):

    def bark(self):
        print("Dog is barking.")
```

Now `Dog` can use both methods:

```python
dog = Dog()

dog.eat()
dog.bark()
```

Output:

```text
Animal is eating.
Dog is barking.
```

# 7. Inheritance with Attributes

```python
class Animal:

    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")


class Dog(Animal):

    def bark(self):
        print(f"{self.name} says Woof!")
```

Create an object:

```python
dog = Dog("Buddy")

dog.eat()
dog.bark()
```

`Dog` inherited:

```text
name
eat()
```

and added:

```text
bark()
```

# 8. `super()`

`super()` is used to call the parent class.

```python
class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
```

Create:

```python
dog = Dog("Buddy", "German Shepherd")

print(dog.name)
print(dog.breed)
```

Output:

```text
Buddy
German Shepherd
```

---

# 9. Method Overriding

A child class can replace a parent's method.

```python
class Animal:

    def sound(self):
        print("Animal sound")


class Dog(Animal):

    def sound(self):
        print("Woof!")
```

Now:

```python
animal = Animal()
dog = Dog()

animal.sound()
dog.sound()
```

Output:

```text
Animal sound
Woof!
```

This is called **method overriding**.

---

# 10. Complete Example

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}.")
        print(f"I am {self.age} years old.")


class Student(Person):

    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def study(self):
        print(f"{self.name} is studying {self.subject}.")
```

Create an object:

```python
student = Student("Ali", 20, "Python")

student.introduce()
student.study()
```

Output:

```text
My name is Ali.
I am 20 years old.
Ali is studying Python.
```

# 📌 OOP Cheat Sheet

| Concept         | Meaning                     |
| --------------- | --------------------------- |
| **Class**       | Blueprint                   |
| **Object**      | Instance of a class         |
| **Attribute**   | Object's data               |
| **Method**      | Object's action             |
| `__init__()`    | Runs when object is created |
| `self`          | Current object              |
| **Inheritance** | Reuse another class         |
| **Parent**      | Class being inherited       |
| **Child**       | Class that inherits         |
| `super()`       | Access parent class         |
| **Overriding**  | Replace a parent method     |
