# What is Concatenation?

**Concatenation** means joining DataFrames together.

For example, suppose we have sales data from January:

```text
Product       Sales
Laptop        50000
Mouse         2000
Keyboard      3000
```

And sales data from February:

```text
Product       Sales
Monitor       25000
Printer       18000
Headphones    4000
```

We can concatenate them to create one DataFrame:

```text
Product       Sales
Laptop        50000
Mouse         2000
Keyboard      3000
Monitor       25000
Printer       18000
Headphones    4000
```

In Pandas, we use:

```python
pd.concat()
```

---

# Creating Sample DataFrames

First, let's import Pandas and create two DataFrames.

## Learn: Create Two DataFrames

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"],
    "Marks": [85, 90, 78]
})

df2 = pd.DataFrame({
    "Name": ["Ayesha", "Usman", "Fatima"],
    "Marks": [88, 95, 82]
})

df1
```

```python
df2
```

Both DataFrames have the same columns:

* `"Name"`
* `"Marks"`

This makes them suitable for combining rows.

---

# concat()

The `pd.concat()` function is used to concatenate DataFrames.

The basic syntax is:

```python
pd.concat([df1, df2])
```

The DataFrames are placed inside a list.

## Learn: Concatenate Two DataFrames

```python
result = pd.concat([df1, df2])

result
```

### What happened?

* Pandas combined `df1` and `df2`.
* The rows from `df2` were added below the rows from `df1`.
* By default, the original indexes were kept.

---

# Combining Rows

By default, `pd.concat()` combines DataFrames vertically.

This means rows are added below existing rows.

## Learn: Combine Student Rows

```python
students_1 = pd.DataFrame({
    "Name": ["Ali", "Ahmed"],
    "Marks": [85, 90]
})

students_2 = pd.DataFrame({
    "Name": ["Sara", "Ayesha"],
    "Marks": [78, 88]
})

combined_students = pd.concat(
    [students_1, students_2]
)

combined_students
```

The result contains all rows from both DataFrames.

---

# Understanding the Index

Look carefully at the result:

```text
   Name      Marks
0  Ali       85
1  Ahmed     90
0  Sara      78
1  Ayesha    88
```

The indexes `0` and `1` are repeated.

This happens because Pandas keeps the original indexes by default.

---

# Index Handling

When concatenating DataFrames, we can control how indexes are handled.

The original indexes may be useful in some situations.

For example:

```python
result = pd.concat(
    [df1, df2]
)

result
```

This keeps the original indexes.

---

# ignore_index

If we want Pandas to create a new continuous index, we can use:

```python
ignore_index=True
```

## Learn: Reset the Index While Concatenating

```python
result = pd.concat(
    [df1, df2],
    ignore_index=True
)

result
```

The result will have a new index:

```text
   Name      Marks
0  Ali       85
1  Ahmed     90
2  Sara      78
3  Ayesha    88
4  Usman     95
5  Fatima    82
```

### What happened?

* Pandas combined both DataFrames.
* The old indexes were ignored.
* Pandas created a new continuous index.

This is often useful when combining datasets with the same structure.

# Combining Columns

We can also concatenate DataFrames horizontally.

To do this, use:

```python
axis=1
```

## Learn: Combine Columns

```python
names = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"]
})

marks = pd.DataFrame({
    "Marks": [85, 90, 78]
})

result = pd.concat(
    [names, marks],
    axis=1
)

result
```

### What happened?

* `axis=1` tells Pandas to combine DataFrames by columns.
* The `"Marks"` column was placed beside the `"Name"` column.
* Matching indexes were combined together.

---

# Understanding axis

The `axis` parameter controls the direction of concatenation.

| Value    | Meaning                      |
| -------- | ---------------------------- |
| `axis=0` | Combine rows vertically      |
| `axis=1` | Combine columns horizontally |

By default:

```python
axis=0
```

So this:

```python
pd.concat([df1, df2])
```

is the same as:

```python
pd.concat(
    [df1, df2],
    axis=0
)
```

# Combining DataFrames with Different Columns

DataFrames do not always have exactly the same columns.

For example:

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Ahmed"],
    "Marks": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Sara", "Ayesha"],
    "City": ["Peshawar", "Lahore"]
})
```

Now concatenate them:

```python
result = pd.concat(
    [df1, df2],
    ignore_index=True
)

result
```

The result will contain all columns:

```text
   Name      Marks      City
0  Ali       85         NaN
1  Ahmed     90         NaN
2  Sara      NaN        Peshawar
3  Ayesha    NaN        Lahore
```

### What happened?

* Pandas combined all available columns.
* Missing values were filled with `NaN`.
* Rows from both DataFrames were included.

---

# Using join in concat()

The `join` parameter controls how columns are handled when DataFrames have different columns.

The default is:

```python
join="outer"
```

This keeps all columns.

We can also use:

```python
join="inner"
```

This keeps only columns common to all DataFrames.

## Learn: Use join="inner"

```python
result = pd.concat(
    [df1, df2],
    join="inner",
    ignore_index=True
)

result
```

Since `"Name"` is the only common column, only that column is kept.

---

# Concatenating Multiple DataFrames

We can concatenate more than two DataFrames.

## Learn: Combine Three DataFrames

```python
january = pd.DataFrame({
    "Product": ["Laptop", "Mouse"],
    "Sales": [50000, 2000]
})

february = pd.DataFrame({
    "Product": ["Keyboard", "Monitor"],
    "Sales": [3000, 25000]
})

march = pd.DataFrame({
    "Product": ["Printer", "Headphones"],
    "Sales": [18000, 4000]
})

combined_sales = pd.concat(
    [january, february, march],
    ignore_index=True
)

combined_sales
```

### What happened?

* Pandas combined January data.
* Pandas added February data below it.
* Pandas added March data below that.
* A new continuous index was created.

---

# Concatenating Many DataFrames

We can place many DataFrames inside the list.

For example:

```python
result = pd.concat(
    [df1, df2, df3, df4],
    ignore_index=True
)
```

This is useful when multiple datasets have the same structure.

---

# Adding a Source Column Before Concatenation

Sometimes we want to know where each row came from.

For example:

```python
january["Month"] = "January"
february["Month"] = "February"
march["Month"] = "March"
```

Now concatenate them:

```python
result = pd.concat(
    [january, february, march],
    ignore_index=True
)

result
```

The `"Month"` column helps us identify the source of each row.
