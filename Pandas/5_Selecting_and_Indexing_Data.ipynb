# Selecting Rows

Rows can be selected using their:

* Index position
* Index label

There are two important Pandas methods:

* `.loc[]` for label-based selection
* `.iloc[]` for position-based selection

Before learning them in detail, remember:

```text
loc   → Label
iloc  → Integer position
```

---

# `.loc[]`

The `.loc[]` indexer is used for **label-based selection**.

Syntax:

```python
df.loc[row_label]
```

By default, our DataFrame has these row labels:

```text
0, 1, 2, 3, 4
```

---

## Learn: Select One Row Using `.loc[]`

```python
df.loc[0]
```

This selects the row with the label `0`.

---

## Learn: Select Multiple Rows Using `.loc[]`

```python
df.loc[[0, 2, 4]]
```

This selects rows with labels:

* `0`
* `2`
* `4`

---

## Learn: Select a Range of Rows Using `.loc[]`

```python
df.loc[1:3]
```

This selects rows with labels:

```text
1, 2, 3
```

### Important

When using `.loc[]` with a slice, the ending label is **included**.

```python
df.loc[1:3]
```

includes row `3`.

# `.iloc[]`

The `.iloc[]` indexer is used for **integer position-based selection**.

Syntax:

```python
df.iloc[row_position]
```

The positions are:

```text
0, 1, 2, 3, 4
```

---

## Learn: Select the First Row

```python
df.iloc[0]
```

This selects the first row.

---

## Learn: Select Multiple Rows

```python
df.iloc[[0, 2, 4]]
```

This selects rows at positions:

* `0`
* `2`
* `4`

---

## Learn: Select a Range of Rows

```python
df.iloc[1:4]
```

This selects positions:

```text
1, 2, 3
```

### Important

The ending position is **not included** with `.iloc[]`.

```python
df.iloc[1:4]
```

includes positions `1`, `2`, and `3`, but not `4`.

---

# `.loc[]` vs `.iloc[]`

| Feature                   | `.loc[]`        | `.iloc[]`         |
| ------------------------- | --------------- | ----------------- |
| Selects using             | Labels          | Integer positions |
| Works with custom indexes | Yes             | No                |
| Slice ending              | Included        | Not included      |
| Example                   | `df.loc["Ali"]` | `df.iloc[0]`      |

Remember:

```text
.loc  → Label
.iloc → Integer position
```

---

# Label-Based Selection

Labels are the names used to identify rows or columns.

For example, we can change the DataFrame index to student names.

```python
df = df.set_index("Name")

df
```

Now the row labels are:

```text
Ali
Ahmed
Sara
Ayesha
Usman
```

We can select a student using `.loc[]`.

---

## Learn: Select a Student by Label

```python
df.loc["Ali"]
```

This selects Ali's complete row.

---

## Learn: Select Multiple Students by Label

```python
df.loc[["Ali", "Sara", "Usman"]]
```

---

## Learn: Select a Range of Labels

```python
df.loc["Ahmed":"Ayesha"]
```

This selects:

* Ahmed
* Sara
* Ayesha

The ending label is included.

---

# Index-Based Selection

After setting student names as indexes, `.iloc[]` still uses integer positions.

```python
df.iloc[0]
```

This selects the first row, which is Ali.

```python
df.iloc[2]
```

This selects the third row, which is Sara.

The position does not change just because the row labels changed.

---

## Learn: Compare `.loc[]` and `.iloc[]`

```python
# Select using the row label
df.loc["Sara"]
```

```python
# Select using the row position
df.iloc[2]
```

Both select Sara in our current DataFrame.

However, they work differently:

* `.loc["Sara"]` uses the label.
* `.iloc[2]` uses the position.

---

# Selecting Specific Rows and Columns

We can select rows and columns together.

With `.loc[]`, the syntax is:

```python
df.loc[rows, columns]
```

With `.iloc[]`, the syntax is:

```python
df.iloc[row_positions, column_positions]
```

---

## Learn: Select One Value Using `.loc[]`

```python
df.loc["Ali", "Marks"]
```

This selects Ali's marks.

---

## Learn: Select Multiple Rows and Columns Using `.loc[]`

```python
df.loc[
    ["Ali", "Sara"],
    ["Age", "Marks"]
]
```

This selects:

* Rows: Ali and Sara
* Columns: Age and Marks

---

## Learn: Select One Value Using `.iloc[]`

```python
df.iloc[0, 1]
```

This selects:

* First row → position `0`
* Second column → position `1`

---

## Learn: Select Specific Rows and Columns Using `.iloc[]`

```python
df.iloc[
    [0, 2, 4],
    [0, 2]
]
```

This selects:

* Rows at positions `0`, `2`, and `4`
* Columns at positions `0` and `2`

---

## Understanding Column Positions

After setting `"Name"` as the index, our columns are:

```text
Position 0 → Age
Position 1 → Marks
Position 2 → City
```

So:

```python
df.iloc[0, 1]
```

selects:

```text
Ali's Marks
```

---

# Changing Indexes

By default, Pandas uses numeric indexes:

```text
0, 1, 2, 3, 4
```

We can change an existing column into the DataFrame index.

Pandas uses:

```python
df.set_index()
```

---

## Learn: Set `Name` as the Index

First, create the DataFrame again:

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 92, 88],
    "City": ["Peshawar", "Lahore", "Karachi", "Islamabad", "Quetta"]
}

df = pd.DataFrame(data)

df
```

Now change the index:

```python
df = df.set_index("Name")

df
```

Now `"Name"` is no longer a normal column.

It becomes the row index.

---

## Learn: Change the Index Using `inplace=True`

We can also modify the existing DataFrame directly.

```python
df.set_index("Name", inplace=True)

df
```

> Use `inplace=True` carefully. It changes the existing DataFrame instead of creating a new one.

---

# Resetting Indexes

Sometimes we want to change the custom index back to the default numeric index.

We can use:

```python
df.reset_index()
```

---

## Learn: Reset the Index

```python
df.reset_index()
```

This returns a new DataFrame where:

* The custom index becomes a normal column.
* Pandas creates a new numeric index.

To save the result:

```python
df = df.reset_index()

df
```

---

## Learn: Reset Index and Remove the Old Index

Sometimes we do not want to keep the old index as a column.

Use:

```python
df.reset_index(drop=True)
```

Example:

```python
df = df.reset_index(drop=True)

df
```

Now Pandas creates a new numeric index, and the old custom index is removed.
