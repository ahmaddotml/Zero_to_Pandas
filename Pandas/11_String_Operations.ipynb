# Replacing Text with replace()

We can replace specific text values using `replace()`.

## Learn: Replace a City Name

```python
df["City"] = df["City"].replace(
    "PESHAWAR",
    "Peshawar"
)

df
```

### What happened?

* Pandas searched for `"PESHAWAR"` in the `"City"` column.
* Every matching value was replaced with `"Peshawar"`.

---

# Replacing Multiple Text Values

We can use a dictionary to replace multiple values.

```python
df["City"] = df["City"].replace({
    "LAHORE": "Lahore",
    "KARACHI": "Karachi",
    "ISLAMABAD": "Islamabad"
})

df
```

### What happened?

* Pandas searched for each incorrect value.
* The dictionary contains old values and new values.
* Matching values were replaced.

---

# Searching Text with contains()

The `contains()` method checks whether text contains a specific value.

## Learn: Find Names Containing "Ali"

```python
result = df[
    df["Name"].str.contains("Ali")
]

result
```

### What happened?

* Pandas checked every value in the `"Name"` column.
* It searched for the text `"Ali"`.
* Only rows containing `"Ali"` were returned.

---

# Splitting Strings

The `split()` method divides text into multiple parts.

## Learn: Split Full Names

Suppose the `"Name"` column contains first and last names.

```python
df["Name"].str.split()
```

### What happened?

* Pandas looked at each name.
* Spaces were used as separators.
* Each name was divided into separate parts.

For example:

```text
Ali Khan → ["Ali", "Khan"]
Ahmed Raza → ["Ahmed", "Raza"]
```

---

# Splitting Strings into Separate Columns

We can split a column and store the parts in new columns.

## Learn: Create First and Last Name Columns

```python
df[["First_Name", "Last_Name"]] = (
    df["Name"]
    .str.split(
        " ",
        n=1,
        expand=True
    )
)

df
```

### What happened?

* The `"Name"` column was split at the first space.
* `n=1` means split only once.
* `expand=True` created separate columns.
* The first part was stored in `"First_Name"`.
* The second part was stored in `"Last_Name"`.

# Extracting Text

We can extract specific parts of text using string operations.

This is useful when text contains structured information.

---

# Extracting Text from Email Addresses

Suppose we want to extract the username from an email address.

For example:

```text
ali@gmail.com → ali
ahmed@yahoo.com → ahmed
```

We can use `split()`.

## Learn: Extract Email Usernames

```python
df["Email_Username"] = (
    df["Email"]
    .str.split("@")
    .str[0]
)

df
```

### What happened?

* Each email was split using `"@"`.
* `.str[0]` selected the first part.
* The username was stored in `"Email_Username"`.

---

# Extracting Email Domains

We can also extract the domain.

```python
df["Email_Domain"] = (
    df["Email"]
    .str.split("@")
    .str[1]
)

df
```

### What happened?

* Each email was split using `"@"`.
* `.str[1]` selected the second part.
* The domain was stored in `"Email_Domain"`.

---

# Extracting Text Using extract()

The `extract()` method can extract text using patterns.

## Learn: Extract Numbers from Text

```python
students = pd.DataFrame({
    "Student_ID": [
        "STD-101",
        "STD-102",
        "STD-103"
    ]
})

students
```

Now extract the numbers.

```python
students["ID_Number"] = students["Student_ID"].str.extract(
    r"(\d+)"
)

students
```

### What happened?

* `\d+` represents one or more digits.
* `extract()` found the number inside each value.
* The extracted number was stored in `"ID_Number"`.

For example:

```text
STD-101 → 101
STD-102 → 102
STD-103 → 103
```

---

# Extracting Text Before a Symbol

We can also extract text before a symbol.

```python
df["Username"] = df["Email"].str.extract(
    r"([^@]+)"
)

df
```

### What happened?

* `[^@]+` means one or more characters that are not `"@"`.
* Pandas extracted the text before `"@"`.
* The result was stored in `"Username"`.
