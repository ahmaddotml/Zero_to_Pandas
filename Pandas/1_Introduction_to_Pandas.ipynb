## What is Pandas?

**Pandas** is a popular Python library used for working with **structured data**.

It helps us:

* Store data
* Organize data
* Read data
* Analyze data
* Clean data
* Filter data
* Modify data

Pandas mainly provides two important data structures:

1. **Series** — one-dimensional data
2. **DataFrame** — two-dimensional table-like data

You can think of a Pandas DataFrame as something similar to an Excel spreadsheet.

# Why Pandas is Used

Python can work with data using lists, dictionaries, and other built-in data structures.

However, when we work with larger or more structured data, Pandas makes the work much easier.

Pandas is commonly used for:

* Data analysis
* Data cleaning
* Data manipulation
* Reading CSV files
* Reading Excel files
* Working with tables
* Handling missing values
* Filtering data
* Preparing data for Machine Learning

For example, imagine you have student data:

| Name  | Age | Marks |
| ----- | --: | ----: |
| Ali   |  20 |    85 |
| Ahmed |  21 |    90 |
| Sara  |  19 |    88 |

Pandas allows us to organize and work with this type of data easily.

---

## Learn: A Simple Data Table

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df
```

### What happened?

* We created a Python dictionary called `data`.
* `"Name"`, `"Age"`, and `"Marks"` are column names.
* Each list contains values for that column.
* `pd.DataFrame()` converts the data into a table.
* `df` stores our DataFrame.
* Writing `df` displays the table.

# Pandas vs Python Lists

A **Python list** is useful for storing a collection of values.

Example:

```python
students = ["Ali", "Ahmed", "Sara"]

students
```

This is simple and useful, but lists are limited when working with table-like data.

For example, if we want to store names, ages, and marks, we may need multiple lists:

```python
names = ["Ali", "Ahmed", "Sara"]

ages = [20, 21, 19]

marks = [85, 90, 88]
```

Managing many separate lists can become difficult.

Pandas organizes related information into rows and columns.

---

## Learn: Using Python Lists

```python
names = ["Ali", "Ahmed", "Sara"]

ages = [20, 21, 19]

marks = [85, 90, 88]

print(names)
print(ages)
print(marks)
```

---

## Learn: The Same Data Using Pandas

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df
```

### What is the difference?

With lists, the data is stored separately.

With Pandas:

* Data is organized in columns.
* Related information stays together.
* Data looks like a table.
* Filtering and analysis become easier.

# Pandas vs NumPy

Both **Pandas** and **NumPy** are popular Python libraries for working with data.

However, they are designed for different purposes.

### NumPy

NumPy is mainly used for:

* Numerical data
* Arrays
* Mathematical operations
* Scientific computing
* Fast numerical calculations

### Pandas

Pandas is mainly used for:

* Tabular data
* Rows and columns
* Data analysis
* Data cleaning
* Labeled data

Pandas is also built on top of NumPy.

---

## Learn: NumPy Array

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

numbers
```

A NumPy array is excellent for numerical operations.

---

## Learn: Pandas Series

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers
```

A Pandas Series stores one-dimensional data and provides useful labels and features for data analysis.

---

## Pandas vs NumPy Summary

| Feature                 | Python Lists         | NumPy               | Pandas        |
| ----------------------- | -------------------- | ------------------- | ------------- |
| Main purpose            | General data storage | Numerical computing | Data analysis |
| Table-like data         | Limited              | Limited             | Excellent     |
| Rows and columns        | No                   | Not directly        | Yes           |
| Labels                  | No                   | Limited             | Yes           |
| Mathematical operations | Basic                | Excellent           | Excellent     |
| Data cleaning           | Manual               | Limited             | Excellent     |

# Creating Your First Pandas Object

The simplest Pandas object we can create is a **Series**.

A Series is a one-dimensional collection of data.

For example:

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers
```

The values are automatically given index labels:

```text
0    10
1    20
2    30
3    40
4    50
```

The numbers on the left are called the **index**.

The numbers on the right are the actual **values**.

---

## Learn: Create Your First Series

```python
import pandas as pd

fruits = pd.Series(["Apple", "Banana", "Mango", "Orange"])

fruits
```

### What happened?

* `pd.Series()` creates a Pandas Series.
* We passed a Python list to the Series.
* Pandas automatically created index values.
* The Series stores the data in one dimension.
