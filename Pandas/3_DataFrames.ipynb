## What is a DataFrame?

A **DataFrame** is a two-dimensional data structure provided by Pandas.

It organizes data into:

* **Rows**
* **Columns**

You can think of a DataFrame as a table or an Excel spreadsheet.

For example:

| Name  | Age | Marks |
| ----- | --: | ----: |
| Ali   |  20 |    85 |
| Ahmed |  21 |    90 |
| Sara  |  19 |    88 |

In this table:

* `Name`, `Age`, and `Marks` are **columns**.
* Each student's information is a **row**.
* The complete table is a **DataFrame**.

---

## Learn: Create Your First DataFrame

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df
```

### What happened?

* `data` is a Python dictionary.
* `"Name"`, `"Age"`, and `"Marks"` become columns.
* Each list provides the values for its column.
* `pd.DataFrame()` creates the DataFrame.
* `df` stores our DataFrame.

# DataFrame from a Dictionary

Creating a DataFrame from a dictionary is one of the most common methods.

Each dictionary:

* **Key** becomes a column name.
* **Value** becomes the data for that column.

---

## Learn: DataFrame from a Dictionary

```python
import pandas as pd

student_data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

students = pd.DataFrame(student_data)

students
```

### What happened?

The dictionary keys:

```text
Name
Age
Marks
```

became DataFrame columns.

The values inside each list became the values of those columns.

---

## Execute Yourself

Create a DataFrame containing information about four students.

```python
import pandas as pd

student_data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 88, 92]
}

df = pd.DataFrame(student_data)

df
```

# DataFrame from Lists

We can also create a DataFrame using a list of lists.

Each inner list usually represents one **row**.

---

## Learn: DataFrame from Lists

```python
import pandas as pd

data = [
    ["Ali", 20, 85],
    ["Ahmed", 21, 90],
    ["Sara", 19, 88]
]

df = pd.DataFrame(
    data,
    columns=["Name", "Age", "Marks"]
)

df
```

### What happened?

Each inner list represents one row:

```text
["Ali", 20, 85]
["Ahmed", 21, 90]
["Sara", 19, 88]
```

The `columns` parameter gives names to the columns.

---

## Understanding the Code

```python
pd.DataFrame(
    data,
    columns=["Name", "Age", "Marks"]
)
```

* `data` contains the rows.
* `columns` defines the column names.
* Pandas creates a table from the information.

---

## Execute Yourself

Add two more students to the DataFrame.

```python
import pandas as pd

data = [
    ["Ali", 20, 85],
    ["Ahmed", 21, 90],
    ["Sara", 19, 88],
    ["Ayesha", 22, 92]
]

df = pd.DataFrame(
    data,
    columns=["Name", "Age", "Marks"]
)

df
```

# DataFrame from Series

A DataFrame can also be created using multiple Pandas Series.

Each Series can become a column in the DataFrame.

---

## Learn: Create Series

```python
import pandas as pd

names = pd.Series(["Ali", "Ahmed", "Sara"])

ages = pd.Series([20, 21, 19])

marks = pd.Series([85, 90, 88])
```

Now, we can combine these Series into a DataFrame.

```python
df = pd.DataFrame({
    "Name": names,
    "Age": ages,
    "Marks": marks
})

df
```

---

### What happened?

* `names` becomes the `Name` column.
* `ages` becomes the `Age` column.
* `marks` becomes the `Marks` column.
* Pandas combines them into one DataFrame.

---

## Execute Yourself

Create three Series and combine them into a DataFrame.

```python
import pandas as pd

names = pd.Series(["Ali", "Ahmed", "Sara", "Ayesha"])

cities = pd.Series([
    "Peshawar",
    "Lahore",
    "Karachi",
    "Islamabad"
])

ages = pd.Series([20, 21, 19, 22])

df = pd.DataFrame({
    "Name": names,
    "City": cities,
    "Age": ages
})

df
```

# Custom Row Indexes

By default, Pandas creates row indexes:

```text
0, 1, 2, 3, ...
```

We can provide our own row indexes using the `index` parameter.

---

## Learn: Create Custom Row Indexes

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(
    data,
    index=["Student 1", "Student 2", "Student 3"]
)

df
```

### What happened?

Instead of:

```text
0
1
2
```

Pandas uses:

```text
Student 1
Student 2
Student 3
```

These are called **custom row indexes**.

---

## Execute Yourself

Create a DataFrame with custom row indexes.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "Marks": [85, 90, 88, 92]
}

df = pd.DataFrame(
    data,
    index=["A", "B", "C", "D"]
)

df
```

# Understanding Rows and Columns

A DataFrame has two main parts:

## Columns

Columns store one type of information.

For example:

```text
Name
Age
Marks
```

Each column contains related data.

## Rows

Each row represents one complete record.

For example:

```text
Ali | 20 | 85
```

This row contains all the information about one student.

---

## Learn: Rows and Columns

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df
```

In this DataFrame:

* There are **3 rows**.
* There are **3 columns**.
* Each row represents one student.
* Each column represents one type of information.

# `shape`

The `shape` attribute tells us the size of a DataFrame.

Syntax:

```python
df.shape
```

It returns:

```text
(rows, columns)
```

---

## Learn: Using `shape`

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df.shape
```

The output will be:

```text
(3, 3)
```

This means:

* `3` rows
* `3` columns

# `dtypes`

The `dtypes` attribute shows the data type of every column.

Syntax:

```python
df.dtypes
```

---

## Learn: Check Column Data Types

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85.5, 90.0, 88.5]
}

df = pd.DataFrame(data)

df.dtypes
```

The result will show the data type for each column.

For example:

```text
Name      object
Age        int64
Marks    float64
```

### What happened?

* `Name` contains text values.
* `Age` contains integer values.
* `Marks` contains decimal values.

# `info()`

The `info()` method gives us a quick summary of the DataFrame.

Syntax:

```python
df.info()
```

It can show information such as:

* Number of rows
* Number of columns
* Column names
* Non-null values
* Data types
* Memory usage

---

## Learn: Using `info()`

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df.info()
```

### What happened?

Pandas displays useful technical information about the DataFrame.

This becomes especially useful when working with large datasets.

---

# `describe()`

The `describe()` method gives a statistical summary of numeric columns.

Syntax:

```python
df.describe()
```

It can display information such as:

* `count`
* `mean`
* `std`
* `min`
* `25%`
* `50%`
* `75%`
* `max`

---

## Learn: Using `describe()`

```python
import pandas as pd

data = {
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

df.describe()
```

### What do these values mean?

* `count` — number of values
* `mean` — average value
* `std` — standard deviation
* `min` — smallest value
* `25%` — first quartile
* `50%` — median
* `75%` — third quartile
* `max` — largest value

For now, focus on understanding that `describe()` provides a quick statistical summary.
