# Step 36: Complete main.py

Now combine everything into one complete program.

```python
import pandas as pd


# ==========================================
# 1. LOAD DATA
# ==========================================

df = pd.read_csv("data/students.csv")


# ==========================================
# 2. EXPLORE DATA
# ==========================================

print("\n" + "=" * 50)
print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

print("\n--- FIRST 5 ROWS ---")
print(df.head())

print("\n--- DATASET SHAPE ---")
print(df.shape)

print("\n--- COLUMNS ---")
print(df.columns)

print("\n--- DATA TYPES ---")
print(df.dtypes)


# ==========================================
# 3. CLEAN DATA
# ==========================================

print("\n--- MISSING VALUES BEFORE CLEANING ---")
print(df.isna().sum())

# Fill missing marks with average
df["Marks"] = df["Marks"].fillna(
    df["Marks"].mean()
)

# Remove duplicate rows
df = df.drop_duplicates()

# Clean text columns
text_columns = [
    "Name",
    "Gender",
    "Class",
    "City",
    "Subject"
]

for column in text_columns:
    df[column] = df[column].str.strip()

# Standardize names
df["Name"] = df["Name"].str.title()

# Ensure numerical data types
df["Marks"] = pd.to_numeric(
    df["Marks"],
    errors="coerce"
)

df["Attendance"] = pd.to_numeric(
    df["Attendance"],
    errors="coerce"
)

print("\n--- MISSING VALUES AFTER CLEANING ---")
print(df.isna().sum())

print("\n--- DUPLICATE ROWS AFTER CLEANING ---")
print(df.duplicated().sum())


# ==========================================
# 4. CREATE FUNCTIONS
# ==========================================

def get_grade(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 50:
        return "D"
    else:
        return "F"


def get_performance(mark):
    if mark >= 85:
        return "Excellent"
    elif mark >= 70:
        return "Good"
    elif mark >= 50:
        return "Average"
    else:
        return "Needs Improvement"


# ==========================================
# 5. CREATE NEW COLUMNS
# ==========================================

df["Grade"] = df["Marks"].apply(
    get_grade
)

df["Result"] = df["Marks"].apply(
    lambda mark: "Pass" if mark >= 50 else "Fail"
)

df["Performance"] = df["Marks"].apply(
    get_performance
)


# ==========================================
# 6. BASIC STATISTICS
# ==========================================

print("\n--- BASIC STATISTICS ---")

print("Total Students:", len(df))
print("Average Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

print("Average Attendance:", df["Attendance"].mean())
print("Highest Attendance:", df["Attendance"].max())
print("Lowest Attendance:", df["Attendance"].min())


# ==========================================
# 7. FILTER DATA
# ==========================================

high_marks = df[
    df["Marks"] >= 80
]

failed_students = df[
    df["Result"] == "Fail"
]

low_attendance = df[
    df["Attendance"] < 80
]


# ==========================================
# 8. GROUP AND ANALYZE
# ==========================================

city_analysis = df.groupby(
    "City"
).agg(
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max"),
    Lowest_Marks=("Marks", "min"),
    Student_Count=("Student_ID", "count"),
    Average_Attendance=("Attendance", "mean")
)

class_analysis = df.groupby(
    "Class"
).agg(
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max"),
    Lowest_Marks=("Marks", "min"),
    Student_Count=("Student_ID", "count")
)

gender_analysis = df.groupby(
    "Gender"
).agg(
    Average_Marks=("Marks", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Student_Count=("Student_ID", "count")
)

subject_analysis = df.groupby(
    "Subject"
).agg(
    Average_Marks=("Marks", "mean"),
    Student_Count=("Student_ID", "count")
)


# ==========================================
# 9. SORT AND RANK
# ==========================================

df["Rank"] = df["Marks"].rank(
    ascending=False,
    method="min"
)

df = df.sort_values(
    by="Marks",
    ascending=False
)

top_students = df.nlargest(
    5,
    "Marks"
)

bottom_students = df.nsmallest(
    5,
    "Marks"
)


# ==========================================
# 10. FIND PATTERNS
# ==========================================

performance_counts = df[
    "Performance"
].value_counts()

correlation = df[
    ["Marks", "Attendance"]
].corr()


# ==========================================
# 11. DISPLAY RESULTS
# ==========================================

print("\n--- CITY ANALYSIS ---")
print(city_analysis)

print("\n--- CLASS ANALYSIS ---")
print(class_analysis)

print("\n--- GENDER ANALYSIS ---")
print(gender_analysis)

print("\n--- SUBJECT ANALYSIS ---")
print(subject_analysis)

print("\n--- PERFORMANCE COUNTS ---")
print(performance_counts)

print("\n--- TOP 5 STUDENTS ---")
print(
    top_students[
        ["Name", "Marks", "Grade", "Rank"]
    ]
)

print("\n--- BOTTOM 5 STUDENTS ---")
print(
    bottom_students[
        ["Name", "Marks", "Grade", "Rank"]
    ]
)

print("\n--- MARKS AND ATTENDANCE CORRELATION ---")
print(correlation)


# ==========================================
# 12. SAVE CLEANED DATA
# ==========================================

df.to_csv(
    "data/cleaned_students.csv",
    index=False
```
