# What is Missing Data?

Missing data means that some values are not available in a DataFrame.

For example, a student's age, marks, or city may be missing.

```text
Name      Age     Marks     City
Ali       20      85        Peshawar
Ahmed     NaN     92        Lahore
Sara      19      NaN       Karachi
Ayesha    22      88        NaN
```

In Pandas, missing numerical values are commonly represented as `NaN`.

`NaN` means:

```text
Not a Number
```

It is used by Pandas to represent a missing value.

# Creating a DataFrame with Missing Values

Let's create a sample student DataFrame containing some missing values.

## Learn: Create a DataFrame

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Age": [20, np.nan, 19, 22, np.nan],
    "Marks": [85, 92, np.nan, 88, 95],
    "City": ["Peshawar", "Lahore", np.nan, "Islamabad", "Karachi"]
}

df = pd.DataFrame(data)

df
```

### What happened?

* We imported Pandas as `pd`.
* We imported NumPy as `np`.
* `np.nan` represents a missing value.
* Some values are missing from the DataFrame.
* Pandas displays numerical missing values as `NaN`.

# Finding Missing Values with isna()

The `isna()` method checks every value in a DataFrame for missing data.

If a value is missing, it returns `True`.

If a value is available, it returns `False`.

## Learn: Use isna()

```python
df.isna()
```

### What happened?

* Pandas checked every value in the DataFrame.
* `True` means the value is missing.
* `False` means the value is available.

---

We can also check missing values in a specific column.

```python
df["Age"].isna()
```

### What happened?

# Counting Missing Values

We can count missing values using `isna()` or `isnull()` together with `sum()`.

## Learn: Count Missing Values in Each Column

```python
df.isna().sum()
```

### What happened?

* `isna()` identified the missing values.
* Missing values became `True`.
* `sum()` counted the `True` values.
* The result shows the number of missing values in each column.

---

### What happened?

* The result is the total number of missing values.

# Removing Missing Values with dropna()

The `dropna()` method removes rows that contain missing values.

## Learn: Remove Rows with Missing Values

```python
clean_df = df.dropna()

clean_df
```

### What happened?

* Pandas checked every row.
* Rows containing missing values were removed.
* Only complete rows remained.
* The original DataFrame was not changed.
* The result was stored in `clean_df`.

---

# Removing Rows with Missing Values from Specific Columns

We can remove rows only when specific columns contain missing values.

## Learn: Remove Rows with Missing Marks

```python
clean_df = df.dropna(
    subset=["Marks"]
)

clean_df
```

### What happened?

* Pandas checked the `"Marks"` column.
* Rows with missing marks were removed.
* Rows with missing values in other columns were kept.

# Filling Missing Values with fillna()

Instead of removing missing values, we can fill them with another value.

The `fillna()` method is used for this purpose.

## Learn: Fill All Missing Values

```python
filled_df = df.fillna("Unknown")

filled_df
```

# Filling Missing Values with the Mean

A common way to fill missing numerical values is with the average value.

* Missing values in `"Marks"` were replaced with the average.
* Available marks remained unchanged.

# Forward Fill

Forward fill uses the previous available value to fill a missing value.

It is also called:

```text
Forward Fill
```

The missing value receives the value from the row above it.

## Learn: Forward Fill

First, create a new DataFrame.

```python
students = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Marks": [85, np.nan, np.nan, 88, 95]
})

students
```

Now use forward fill.

```python
filled_students = students.ffill()

filled_students
```

### What happened?

* Pandas found missing values.
* Each missing value was filled using the previous available value.
* The value from the row above moved forward.

For example:

```text
85
NaN
NaN
```

After forward fill:

```text
85
85
85
```

---

# Forward Fill Using fillna()

Forward fill can also be used with `fillna()`.

```python
filled_students = students.fillna(
    method="ffill"
)

filled_students
```

# Backward Fill

Backward fill uses the next available value to fill a missing value.

The missing value receives the value from the row below it.

It is also called:

```text
Backward Fill
```

## Learn: Backward Fill

```python
students = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Marks": [85, np.nan, np.nan, 88, 95]
})

students
```

Now use backward fill.

```python
filled_students = students.bfill()

filled_students
```

### What happened?

* Pandas found missing values.
* Each missing value was filled using the next available value.
* The value from the row below moved backward.

For example:

```text
85
NaN
NaN
88
```

After backward fill:

```text
85
88
88
88
```

---

# Backward Fill Using fillna()

Backward fill can also be used with `fillna()`.

```python
filled_students = students.fillna(
    method="bfill"
)

filled_students
```
