# Why Combine Datasets?

Large datasets are often divided into separate tables to keep information organized.

For example, we may have a customer dataset:

```text
Customer_ID    Name       City
101            Ali        Peshawar
102            Ahmed      Lahore
103            Sara       Karachi
```

And an order dataset:

```text
Order_ID       Customer_ID    Product      Amount
1              101            Laptop       50000
2              102            Mouse        1500
3              101            Keyboard     3000
```

Both datasets contain `"Customer_ID"`.

We can use this common column to combine the datasets.

The result can become:

```text
Order_ID    Customer_ID    Name      City        Product      Amount
1           101            Ali       Peshawar    Laptop       50000
2           102            Ahmed     Lahore      Mouse        1500
3           101            Ali       Peshawar    Keyboard     3000
```

---

# Understanding Keys

A **key** is a column used to connect two datasets.

For example:

```text
Customers DataFrame              Orders DataFrame

Customer_ID                      Customer_ID
101                              101
102                              102
103                              101
```

Here, `"Customer_ID"` is the common key.

We use the key to tell Pandas how records from one DataFrame should match records from another DataFrame.

Common examples of keys include:

* `Customer_ID`
* `Student_ID`
* `Product_ID`
* `Order_ID`
* `Employee_ID`

---

# Creating Sample Datasets

Let's create a customer DataFrame and an order DataFrame.

## Learn: Create Customer Data

```python
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "City": ["Peshawar", "Lahore", "Karachi", "Islamabad"]
})

customers
```

---

## Learn: Create Order Data

```python
orders = pd.DataFrame({
    "Order_ID": [1, 2, 3, 4, 5],
    "Customer_ID": [101, 102, 101, 103, 105],
    "Product": [
        "Laptop",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Printer"
    ],
    "Amount": [50000, 1500, 3000, 25000, 18000]
})

orders
```

### What happened?

* `customers` contains customer information.
* `orders` contains order information.
* Both DataFrames contain `"Customer_ID"`.
* We can use `"Customer_ID"` as the key to combine them.

---

# merge()

The `merge()` method is used to combine DataFrames based on common columns.

The basic syntax is:

```python
pd.merge(
    left_dataframe,
    right_dataframe,
    on="Common_Column"
)
```

For our example:

```python
pd.merge(
    customers,
    orders,
    on="Customer_ID"
)
```

By default, `merge()` performs an **inner join**.

---

# Inner Join

An inner join returns only rows where the key exists in both DataFrames.

## Learn: Perform an Inner Join

```python
inner_result = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="inner"
)

inner_result
```

### What happened?

* Pandas compared `"Customer_ID"` in both DataFrames.
* Only matching customer IDs were included.
* Customer `104` was removed because they have no matching order.
* Order with customer `105` was removed because there is no matching customer.

---

# Inner Join Using the Default merge()

Because inner join is the default behavior, we can also write:

```python
inner_result = pd.merge(
    customers,
    orders,
    on="Customer_ID"
)

inner_result
```

# Left Join

A left join keeps all rows from the left DataFrame.

Matching information from the right DataFrame is added where available.

If there is no match, Pandas uses missing values.

## Learn: Perform a Left Join

```python
left_result = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="left"
)

left_result
```

### What happened?

* All customers from `customers` were kept.
* Matching order information was added.
* Customer `104` has no order.
* Therefore, order-related columns contain `NaN` for customer `104`.

---

# Right Join

A right join keeps all rows from the right DataFrame.

Matching information from the left DataFrame is added where available.

# Outer Join

An outer join keeps all rows from both DataFrames.

Where matching data exists, the rows are combined.

Where no match exists, Pandas fills missing values with `NaN`.

## Learn: Perform an Outer Join

```python
outer_result = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="outer"
)

outer_result
```

### What happened?

The result contains:

* All customers from `customers`.
* All orders from `orders`.
* Matching data combined together.
* `NaN` where matching information does not exist.

---

# Understanding Join Types

Here is a simple summary:

| Join Type | Result                                        |
| --------- | --------------------------------------------- |
| `inner`   | Keeps only matching rows from both DataFrames |
| `left`    | Keeps all rows from the left DataFrame        |
| `right`   | Keeps all rows from the right DataFrame       |
| `outer`   | Keeps all rows from both DataFrames           |

---

# Merge on One Column

The most common type of merge uses one common column.

For example:

```python
result = pd.merge(
    customers,
    orders,
    on="Customer_ID"
)

result
```

The `"Customer_ID"` column is used as the key.

---

# Using Different Column Names as Keys

Sometimes the matching columns have different names.

For example:

```python
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Ali", "Ahmed", "Sara"]
})

orders = pd.DataFrame({
    "Customer": [101, 102, 101],
    "Product": ["Laptop", "Mouse", "Keyboard"]
})
```

Here, the key columns have different names:

* `"Customer_ID"` in `customers`
* `"Customer"` in `orders`

We can use `left_on` and `right_on`.

```python
result = pd.merge(
    customers,
    orders,
    left_on="Customer_ID",
    right_on="Customer"
)

result
```

---

# Merge on Multiple Columns

We can merge DataFrames using more than one key.

For example, suppose we have sales data.

## Learn: Create Two Sales DataFrames

```python
sales_1 = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "City": ["Peshawar", "Lahore", "Karachi"],
    "Sales": [50000, 2000, 3000]
})

sales_2 = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard"],
    "City": ["Peshawar", "Lahore", "Karachi"],
    "Quantity": [5, 20, 10]
})

sales_1
```

```python
sales_2
```

Both DataFrames contain:

* `"Product"`
* `"City"`

We can merge using both columns.

```python
result = pd.merge(
    sales_1,
    sales_2,
    on=["Product", "City"]
)

result
```

### What happened?

* Pandas matched `"Product"` values.
* Pandas also matched `"City"` values.
* Rows were combined only when both keys matched.

# join()

The `join()` method is another way to combine DataFrames.

It commonly combines data based on indexes.

## Learn: Create Two DataFrames

```python
students = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"],
    "Marks": [85, 90, 78]
})

cities = pd.DataFrame({
    "City": ["Peshawar", "Lahore", "Karachi"]
})
```

Now join them:

```python
result = students.join(cities)

result
```

### What happened?

* `join()` matched rows using their indexes.
* Both DataFrames have matching default indexes.
* The `"City"` column was added to `students`.

---

# join() with a Common Column

We can also use `join()` with a key.

Suppose we have:

```python
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Ali", "Ahmed", "Sara"]
})

orders = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Product": ["Laptop", "Mouse", "Keyboard"]
})
```

Set the key as an index in the second DataFrame:

```python
orders_indexed = orders.set_index(
    "Customer_ID"
)

orders_indexed
```

Now join:

```python
result = customers.join(
    orders_indexed,
    on="Customer_ID"
)

result
```

### What happened?

* `"Customer_ID"` in `customers` was used as the key.
* Pandas matched it with the index of `orders_indexed`.
* Matching product information was added.

---

# merge() vs join()

Both methods combine DataFrames, but they are commonly used differently.

| Method    | Common Use                               |
| --------- | ---------------------------------------- |
| `merge()` | Combine DataFrames using columns or keys |
| `join()`  | Combine DataFrames mainly using indexes  |

For most beginner data analysis tasks involving common columns, `merge()` is easier to understand and use.
