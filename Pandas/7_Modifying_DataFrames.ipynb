
### What happened?

* We imported Pandas as `pd`.
* We created a dictionary containing student data.
* `"Name"`, `"Age"`, and `"Marks"` are column names.
* `pd.DataFrame()` converted the dictionary into a DataFrame.
* `df` stores our DataFrame.

---

## Execute Yourself

Create your own DataFrame containing three students.

```python
# Write your code below

import pandas as pd
```

---

# Adding a Column

A new column can be added by assigning values to a new column name.

## Learn: Add a New Column

```python
df["City"] = [
    "Peshawar",
    "Lahore",
    "Karachi",
    "Islamabad"
]

df
```

### What happened?

* `"City"` is a new column.
* We assigned a list of values to the column.
* Each value was added to its corresponding row.
* The number of values should match the number of rows.

---

## Execute Yourself

Add a column called `"Course"` to the DataFrame.

Give each student a course name.

```python
# Write your code below
```

---

# Updating a Column

We can update an existing column by assigning new values to it.

## Learn: Update a Column

```python
df["Marks"] = [90, 95, 80, 92]

df
```

### What happened?

* `"Marks"` already existed in the DataFrame.
* The old marks were replaced with new values.
* The column name remained the same.
* Only the values were updated.

---

## Execute Yourself

Update the values in the `"Age"` column.

```python
# Write your code below
```

---

# Creating Columns from Calculations

We can create a new column by performing calculations on existing columns.

## Learn: Create a Column from a Calculation

```python
df["Double_Marks"] = df["Marks"] * 2

df
```

### What happened?

* Pandas used the values from the `"Marks"` column.
* Every mark was multiplied by `2`.
* The results were stored in a new column called `"Double_Marks"`.

---

## Another Example

Let's calculate the remaining marks from a total of `100`.

```python
df["Remaining_Marks"] = 100 - df["Marks"]

df
```

### What happened?

* Pandas performed the calculation for every row.
* Each student's marks were subtracted from `100`.
* The result was stored in `"Remaining_Marks"`.

---

## Execute Yourself

Create a new column called `"Percentage"`.

Assume the total marks are `100`.

```python
# Write your code below
```

---

# Adding a Column with insert()

The `insert()` method allows us to add a column at a specific position.

The basic syntax is:

```python
df.insert(position, "Column_Name", values)
```

## Learn: Using insert()

```python
df.insert(
    1,
    "Student_ID",
    [101, 102, 103, 104]
)

df
```

### What happened?

* `1` is the position where the column will be inserted.
* `"Student_ID"` is the new column name.
* The list contains values for the new column.
* The column was added at the specified position.

---

## Execute Yourself

Insert a column called `"Class"` at position `2`.

```python
# Write your code below
```

---

# Renaming Columns

We can change column names using the `rename()` method.

## Learn: Rename a Column

```python
df.rename(
    columns={"Name": "Student_Name"},
    inplace=True
)

df
```

### What happened?

* `"Name"` was the old column name.
* `"Student_Name"` is the new column name.
* `columns={}` tells Pandas which columns to rename.
* `inplace=True` permanently changes the DataFrame.

---

## Renaming Multiple Columns

We can rename multiple columns at the same time.

```python
df.rename(
    columns={
        "Age": "Student_Age",
        "Marks": "Student_Marks"
    },
    inplace=True
)

df
```

### What happened?

* `"Age"` was renamed to `"Student_Age"`.
* `"Marks"` was renamed to `"Student_Marks"`.
* A dictionary can contain multiple old and new column names.

---

## Execute Yourself

Rename the following columns:

* `"City"` to `"Student_City"`
* `"Course"` to `"Student_Course"`

```python
# Write your code below
```

---

# Changing Individual Values

We can change a specific value in a DataFrame using `loc`.

## Learn: Change a Value

```python
df.loc[0, "Student_Age"] = 25

df
```

### What happened?

* `0` represents the row index.
* `"Student_Age"` represents the column.
* The value at that specific row and column was changed to `25`.

---

# Changing Values Using a Condition

We can also change values based on a condition.

## Learn: Change Values with a Condition

```python
df.loc[
    df["Student_Marks"] < 80,
    "Student_Marks"
] = 80

df
```

### What happened?

* Pandas checked the `"Student_Marks"` column.
* It found values less than `80`.
* Those values were changed to `80`.

---

## Execute Yourself

Change the marks of the student at index `1`.

```python
# Write your code below
```

---

# Changing Values with replace()

The `replace()` method is used to replace one value with another.

## Learn: Using replace()

First, let's add a simple `"Result"` column.

```python
df["Result"] = [
    "Pass",
    "Pass",
    "Fail",
    "Pass"
]

df
```

Now replace `"Fail"` with `"Needs Improvement"`.

```python
df["Result"] = df["Result"].replace(
    "Fail",
    "Needs Improvement"
)

df
```

### What happened?

* Pandas searched for `"Fail"` in the `"Result"` column.
* Every matching value was replaced.
* `"Pass"` values remained unchanged.

---

# Replacing Multiple Values

We can use a dictionary to replace multiple values.

## Learn: Replace Multiple Values

```python
df["Result"] = df["Result"].replace({
    "Pass": "Successful",
    "Needs Improvement": "Unsuccessful"
})

df
```

### What happened?

* `"Pass"` was replaced with `"Successful"`.
* `"Needs Improvement"` was replaced with `"Unsuccessful"`.
* A dictionary allows us to replace multiple values.

---

## Execute Yourself

Replace:

* `"Successful"` with `"Passed"`
* `"Unsuccessful"` with `"Failed"`

```python
# Write your code below
```

---

# Removing Columns with drop()

The `drop()` method can be used to remove columns from a DataFrame.

## Learn: Remove a Column

```python
df = df.drop(
    "Remaining_Marks",
    axis=1
)

df
```

### What happened?

* `"Remaining_Marks"` was removed.
* `axis=1` means we are working with columns.
* The updated DataFrame was assigned back to `df`.

---

# Removing Multiple Columns

We can remove multiple columns by providing their names in a list.

## Learn: Remove Multiple Columns

```python
df = df.drop(
    ["Double_Marks", "Result"],
    axis=1
)

df
```

### What happened?

* Both columns were removed.
* The column names were placed inside a list.
* `axis=1` tells Pandas to remove columns.

---

## Execute Yourself

Remove the `"Student_City"` column.

```python
# Write your code below
```

---

# Removing Rows with drop()

The `drop()` method can also remove rows.

To remove a row, we use its index.

## Learn: Remove a Row

```python
df = df.drop(3)

df
```

### What happened?

* Row index `3` was removed.
* `drop()` removes the specified row.
* Rows use `axis=0` by default.

---

# Removing Multiple Rows

We can remove multiple rows by placing their indexes inside a list.

## Learn: Remove Multiple Rows

```python
students = pd.DataFrame({
    "Name": [
        "Ali",
        "Ahmed",
        "Sara",
        "Ayesha",
        "Usman"
    ],
    "Marks": [85, 92, 78, 88, 95]
})

students
```

```python
students = students.drop([1, 3])

students
```

### What happened?

* Rows with indexes `1` and `3` were removed.
* The remaining rows stayed in the DataFrame.

---

## Execute Yourself

Remove the row with index `0`.

```python
# Write your code below
```

---

# Practice

## Practice 1: Add Grades

Create a DataFrame containing:

* 5 students
* Their ages
* Their marks
* Their cities

Then add a `"Grade"` column.

Use:

* `A` for marks `90` and above
* `B` for marks from `80` to `89`
* `C` for marks below `80`

```python
# Write your code below
```

---

## Practice 2: Rename Columns

Create or use a student DataFrame.

Rename:

* `"Name"` to `"Student_Name"`
* `"Age"` to `"Student_Age"`
* `"Marks"` to `"Student_Marks"`

```python
# Write your code below
```

---

## Practice 3: Remove Unnecessary Columns

Create a DataFrame with the following columns:

* Name
* Age
* Marks
* City
* Temporary_ID

Remove the `"Temporary_ID"` column using `drop()`.

```python
# Write your code below
```

---

# Challenge

## Challenge 1: Complete Student Data Modification

Create a DataFrame containing:

* 5 student names
* Their ages
* Their marks
* Their cities

Then complete all of the following:

1. Add a `"Grade"` column.
2. Add a `"Total_Marks"` column with the value `100`.
3. Create a `"Percentage"` column from the marks.
4. Rename `"Name"` to `"Student_Name"`.
5. Change the marks of one student.
6. Replace one grade value with another value.
7. Remove the `"City"` column.
8. Remove one student row.

```python
# Write your code below
```

---

## Challenge 2: Use Multiple Modification Methods

Create a DataFrame containing student information.

Then use all of the following methods:

* Add a column using `df["Column"]`
* Add a column using `insert()`
* Rename a column using `rename()`
* Change a value using `loc`
* Replace values using `replace()`
* Remove a column using `drop()`
* Remove a row using `drop()`

```python
# Write your code below
```

---

# Final Practice

Complete all of the following in one cell:

1. Import Pandas as `pd`.
2. Create a DataFrame containing five students.
3. Include `"Name"`, `"Age"`, and `"Marks"` columns.
4. Add a `"City"` column.
5. Add a `"Grade"` column.
6. Create a `"Double_Marks"` column from `"Marks"`.
7. Rename `"Name"` to `"Student_Name"`.
8. Change one student's marks.
9. Use `replace()` to change a grade.
10. Remove the `"City"` column.
11. Remove one student row.
12. Display the final DataFrame.

```python
# 1. Import Pandas


# 2. Create a DataFrame


# 3. Add a City column


# 4. Add a Grade column


# 5. Create a Double_Marks column


# 6. Rename the Name column


# 7. Change one student's marks


# 8. Replace a grade value


# 9. Remove the City column


# 10. Remove one student row


# 11. Display the final DataFrame
```

---

# Summary

In this notebook, you learned:

* How to add a column
* How to update an existing column
* How to create columns from calculations
* How to add columns using `insert()`
* How to rename columns using `rename()`
* How to change individual values using `loc`
* How to change values using conditions
* How to replace values using `replace()`
* How to remove columns using `drop()`
* How to remove rows using `drop()`

You now understand the basic ways to modify the structure and values of a Pandas DataFrame.

These skills are important when cleaning, preparing, and transforming data for analysis.

---
