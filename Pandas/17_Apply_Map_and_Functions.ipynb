# apply()

The `apply()` method applies a function to data.

A common use is applying a function to every value in a Series.

The basic syntax is:

```python
df["Column"].apply(function)
```

For example:

```python
df["Marks"].apply(lambda x: x + 5)
```

This adds `5` to every value in the `"Marks"` column.

---

# Applying a Function to a Column

## Learn: Add Bonus Marks

```python
df["Bonus_Marks"] = df["Marks"].apply(
    lambda x: x + 5
)

df
```

### What happened?

* Pandas selected the `"Marks"` column.
* `apply()` processed every mark.
* The lambda function added `5`.
* The results were stored in `"Bonus_Marks"`.

---

# apply() with a Named Function

Instead of using a lambda function, we can create a normal Python function.

```python
def add_bonus(mark):
    return mark + 5
```

Now apply it:

```python
df["Bonus_Marks"] = df["Marks"].apply(add_bonus)

df
```

This produces the same result.

# Lambda Functions

A lambda function is a small function written in one line.

The basic syntax is:

```python
lambda input: result
```

For example:

```python
lambda x: x * 2
```

This means:

* Take a value called `x`.
* Multiply it by `2`.
* Return the result.

We can use lambda functions with `apply()`.

---

# Using Lambda with apply()

## Learn: Double the Marks

```python
df["Double_Marks"] = df["Marks"].apply(
    lambda x: x * 2
)

df
```

---

# Another Example

Convert marks into percentages.

Suppose the total marks are `100`.

```python
df["Percentage"] = df["Marks"].apply(
    lambda x: (x / 100) * 100
)

df
```

In this example, the values remain the same because the total is `100`, but the same idea can be used when calculating percentages from different totals.

---

# Conditional Lambda Functions

We can use conditions inside lambda functions.

For example:

```python
lambda x: "Pass" if x >= 50 else "Fail"
```

This means:

* If `x` is greater than or equal to `50`, return `"Pass"`.
* Otherwise, return `"Fail"`.

## Learn: Create a Result Column

```python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)

df
```

# Custom Functions

A custom function is a function that we create ourselves.

Custom functions are useful when the transformation contains more than one condition.

## Learn: Create a Grade Function

```python
def get_grade(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 50:
        return "D"
    else:
        return "F"
```

Now apply the function:

```python
df["Grade"] = df["Marks"].apply(get_grade)

df
```

### What happened?

* `get_grade()` received one mark at a time.
* The function checked the conditions.
* A grade was returned.
* `apply()` performed this process for every value.

---

# Applying Custom Functions to Multiple Columns

`apply()` can also work with an entire DataFrame.

We can use `axis=1` to apply a function to each row.

For example, create a function:

```python
def student_category(row):
    if row["Marks"] >= 80 and row["City"] == "Peshawar":
        return "High Performer - Peshawar"
    elif row["Marks"] >= 80:
        return "High Performer"
    else:
        return "Regular Student"
```

Apply it to rows:

```python
df["Category"] = df.apply(
    student_category,
    axis=1
)

df
```

### What happened?

* `axis=1` tells Pandas to process one row at a time.
* The function receives the complete row.
* The function can access multiple columns.
* A category is created for every student.

---

# Understanding axis in apply()

When using `apply()` on a DataFrame:

| Value    | Meaning                           |
| -------- | --------------------------------- |
| `axis=0` | Apply the function to each column |
| `axis=1` | Apply the function to each row    |

For example:

```python
df.apply(function, axis=1)
```

applies the function to every row.

# map()

The `map()` method is used to transform values in a Series.

The basic syntax is:

```python
df["Column"].map(mapping)
```

We can use:

* A dictionary
* A function
* A lambda function

---

# Using map() with a Dictionary

Suppose we have city codes:

```python
city_codes = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "City_Code": ["PEW", "LHR", "KHI", "PEW"]
})

city_codes
```

Create a mapping dictionary:

```python
city_map = {
    "PEW": "Peshawar",
    "LHR": "Lahore",
    "KHI": "Karachi"
}
```

Now use `map()`:

```python
city_codes["City"] = city_codes["City_Code"].map(
    city_map
)

city_codes
```

### What happened?

* Pandas looked at each `"City_Code"`.
* `map()` searched for that value in `city_map`.
* The matching full city name was returned.
* The result was stored in the `"City"` column.

---

# map() with a Lambda Function

We can also use `map()` with a lambda function.

## Learn: Add Text to Every Name

```python
df["Student_Name"] = df["Name"].map(
    lambda x: "Student: " + x
)

df
```

This applies the transformation to every value in the `"Name"` column.

---

# map() with a Function

We can also pass a custom function.

```python
def add_title(name):
    return "Student: " + name
```

Then:

```python
df["Student_Name"] = df["Name"].map(add_title)

df
```

---

# apply() vs map()

Both methods can apply functions, but they are commonly used differently.

| Method    | Common Use                               |
| --------- | ---------------------------------------- |
| `map()`   | Transform values in one Series           |
| `apply()` | Apply functions to a Series or DataFrame |

For simple value-to-value transformations on one column, `map()` is often convenient.

For row-based operations involving multiple columns, `apply()` with `axis=1` is useful.

---

# applymap() and Element-Wise Operations

Older Pandas code often uses `applymap()` to apply a function to every individual element in a DataFrame.

For example:

```python
numbers = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [4, 5, 6]
})

numbers.applymap(lambda x: x * 2)
```

This processes each individual value.

> **Note:** In modern Pandas, `DataFrame.map()` is used for element-wise operations on an entire DataFrame. You may still see `applymap()` in older tutorials and code.

---

# DataFrame.map() for Element-Wise Operations

## Learn: Double Every Value

```python
numbers = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [4, 5, 6]
})

result = numbers.map(
    lambda x: x * 2
)

result
```

### What happened?

* Every value in the DataFrame was processed.
* Each value was multiplied by `2`.
* A new transformed DataFrame was returned.

---

# Cleaning Text with Element-Wise Operations

Suppose we have text with extra spaces:

```python
names = pd.DataFrame({
    "First_Name": [
        " Ali ",
        " Ahmed ",
        " Sara "
    ],
    "Last_Name": [
        " Khan ",
        " Shah ",
        " Ali "
    ]
})

names
```

We can remove spaces from every element:

```python
clean_names = names.map(
    lambda x: x.strip()
)

clean_names
```

# Conditional Transformations

Conditional transformations change values based on conditions.

For example:

```python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)
```

We can also create more categories using a custom function.

---

# Creating Custom Categories

## Learn: Create Performance Categories

```python
def get_category(mark):
    if mark >= 85:
        return "Excellent"
    elif mark >= 70:
        return "Good"
    elif mark >= 50:
        return "Average"
    else:
        return "Needs Improvement"
```

Apply the function:

```python
df["Performance"] = df["Marks"].apply(
    get_category
)

df
```

---

# Using a Conditional Function with Multiple Columns

We can create categories using more than one column.

```python
def classify_student(row):
    if row["Marks"] >= 80 and row["City"] == "Peshawar":
        return "Excellent - Peshawar"
    elif row["Marks"] >= 80:
        return "Excellent"
    elif row["Marks"] >= 50:
        return "Pass"
    else:
        return "Fail"
```

Apply the function:

```python
df["Student_Category"] = df.apply(
    classify_student,
    axis=1
)

df
```
