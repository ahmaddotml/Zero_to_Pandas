# Why Data Cleaning Matters

Real-world data is often messy.

A dataset may contain:

* Duplicate rows
* Missing or incorrect values
* Incorrect data types
* Dates stored as text
* Extra spaces
* Inconsistent text
* Incorrect column names

Data cleaning helps us prepare data before analysis.

For example, this data may be messy:

```text
 Name        Age       Marks       City
 Ali         "20"      "85"        peshawar
 Ahmed       "21"      "92"        Lahore 
 Sara        "19"      "78"        KARACHI
 Ali         "20"      "85"        peshawar
```

Problems include:

* `"Age"` and `"Marks"` may be stored as text.
* `"City"` has inconsistent capitalization.
* Some values contain unnecessary spaces.
* The first and last rows are duplicates.

---

# Creating a Messy DataFrame

Let's create a sample student DataFrame containing several common data problems.

## Learn: Create a Messy DataFrame

```python
import pandas as pd

data = {
    " Student Name ": [
        " Ali ",
        "Ahmed",
        " Sara",
        "Ayesha ",
        "Ali"
    ],
    " Age ": ["20", "21", "19", "22", "20"],
    " Marks ": ["85", "92", "78", "88", "85"],
    " City ": [
        "peshawar",
        "Lahore ",
        " KARACHI",
        "Islamabad",
        "peshawar"
    ],
    "Join Date": [
        "2025-01-10",
        "2025-02-15",
        "2025-03-20",
        "2025-04-25",
        "2025-01-10"
    ]
}

df = pd.DataFrame(data)

df
```

### What happened?

* We created a DataFrame containing student information.
* Some column names contain unnecessary spaces.
* Some text values contain extra spaces.
* Ages and marks are stored as text.
* City names have inconsistent capitalization.
* The DataFrame contains duplicate information.

# Duplicate Data

Duplicate data means that the same row appears more than once in a dataset.

Duplicates can cause problems during data analysis.

For example:

```text
Ali      20      85
Ali      20      85
```

The same information appears twice.

Pandas provides methods to find and remove duplicate data.

---

# Finding Duplicates with duplicated()

The `duplicated()` method checks whether a row is a duplicate.

If a row is duplicated, it returns `True`.

If a row is unique, it returns `False`.

## Learn: Use duplicated()

```python
df.duplicated()
```

### What happened?

* Pandas checked every row.
* The first occurrence of a row is usually `False`.
* Repeated rows are marked as `True`.
* `True` means the row is a duplicate.

---

# Displaying Duplicate Rows

We can display only the duplicate rows.

```python
df[df.duplicated()]
```

### What happened?

* `df.duplicated()` finds duplicate rows.
* It returns `True` for duplicated rows.
* The DataFrame displays only rows marked as duplicates.

---

# Counting Duplicate Rows

We can count duplicate rows using `sum()`.

```python
df.duplicated().sum()
```

### What happened?

* `duplicated()` finds duplicate rows.
* Duplicate rows are represented as `True`.
* `sum()` counts the `True` values.
* The result is the number of duplicate rows.

# Removing Duplicates with drop_duplicates()

The `drop_duplicates()` method removes duplicate rows.

## Learn: Remove Duplicate Rows

```python
clean_df = df.drop_duplicates()

clean_df
```

### What happened?

* Pandas checked for duplicate rows.
* Duplicate rows were removed.
* The first occurrence was kept.
* The result was stored in `clean_df`.

---

# Removing Duplicates Permanently

We can also update the original DataFrame.

```python
df.drop_duplicates(inplace=True)

df
```

### What happened?

* Duplicate rows were removed.
* `inplace=True` permanently updated `df`.

---

# Removing Duplicates Based on Specific Columns

Sometimes we want to check duplicates based on only one or more columns.

## Learn: Remove Duplicate Names

```python
clean_df = df.drop_duplicates(
    subset=[" Student Name "]
)

clean_df
```

### What happened?

* Pandas checked only the `" Student Name "` column.
* Repeated student names were considered duplicates.
* Only the first occurrence was kept.

Sometimes values have the wrong data type.

For example:

```text
Age = "20"
Marks = "85"
```

These values look like numbers but may actually be stored as text.

We can check data types using:

```python
df.dtypes
```

---

## Learn: Check Data Types

```python
df.dtypes
```

### What happened?

* Pandas displayed the data type of every column.
* Text columns usually have the `object` data type.
* Numerical columns may have `int64` or `float64`.
* Dates may have the `datetime64` data type.

# Converting Data Types with astype()

The `astype()` method is used to convert data from one data type to another.

## Learn: Convert Age to Integer

First, let's create a simple DataFrame.

```python
students = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"],
    "Age": ["20", "21", "19"],
    "Marks": ["85", "92", "78"]
})

students
```

Now convert the `"Age"` column.

```python
students["Age"] = students["Age"].astype(int)

students
```

### What happened?

* The `"Age"` values were originally text.
* `astype(int)` converted them into integers.
* The column can now be used more easily for numerical calculations.

---

# Converting Multiple Columns

We can convert multiple columns separately.

```python
students["Age"] = students["Age"].astype(int)

students["Marks"] = students["Marks"].astype(int)

students
```

---

# Converting Data to Float

We can also convert values into decimal numbers.

```python
students["Marks"] = students["Marks"].astype(float)

students
```

# Converting Dates with to_datetime()

Dates are sometimes stored as text.

For example:

```text
"2025-01-10"
"2025-02-15"
"2025-03-20"
```

We can convert text dates into datetime values using `to_datetime()`.

## Learn: Convert a Date Column

```python
df["Join Date"] = pd.to_datetime(
    df["Join Date"]
)

df
```

### What happened?

* Pandas read the values in `"Join Date"`.
* `pd.to_datetime()` converted the text into datetime values.
* The column can now be used for date operations.

---

# Checking the Date Data Type

```python
df.dtypes
```

The `"Join Date"` column should now have a datetime data type.

---

# Converting Different Date Formats

Pandas can also convert many different date formats.

```python
dates = pd.DataFrame({
    "Date": [
        "10/01/2025",
        "15/02/2025",
        "20/03/2025"
    ]
})

dates["Date"] = pd.to_datetime(
    dates["Date"],
    dayfirst=True
)

dates
```

### What happened?

* The dates used the `day/month/year` format.
* `dayfirst=True` tells Pandas that the day comes first.
* Pandas converted the values into datetime objects.

# Cleaning Column Names

Column names may contain:

* Extra spaces
* Unnecessary capitalization
* Inconsistent formatting

For example:

```text
" Student Name "
" AGE "
"Marks "
```

We can clean column names before working with the DataFrame.

---

# Removing Spaces from Column Names

We can use `.str.strip()` to remove spaces from the beginning and end of every column name.

## Learn: Clean Column Names

```python
df.columns = df.columns.str.strip()

df
```

### What happened?

* Pandas selected all column names.
* `.str.strip()` removed spaces from the beginning and end.
* The column names became cleaner.

For example:

```text
" Student Name " → "Student Name"
" Age " → "Age"
" Marks " → "Marks"
```

---

# Converting Column Names to Lowercase

We can make all column names lowercase.

```python
df.columns = df.columns.str.lower()

df
```

For example:

```text
Student Name → student name
Age → age
Marks → marks
```

---

# Replacing Spaces in Column Names

We can replace spaces with underscores.

```python
df.columns = df.columns.str.replace(
    " ",
    "_"
)

df
```

For example:

```text
student name → student_name
join date → join_date
```

---

# Complete Column Name Cleaning

We can combine multiple cleaning steps.

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df
```

### What happened?

* `.str.strip()` removed unnecessary spaces.
* `.str.lower()` converted names to lowercase.
* `.str.replace()` replaced spaces with underscores.

# Replacing Incorrect Values

Sometimes a dataset contains incorrect or inconsistent values.

For example:

```text
Peshawar
Peshawar
Peshawar City
PESHAWAR
```

We can replace incorrect values with correct values.

## Learn: Replace an Incorrect Value

```python
df["city"] = df["city"].replace(
    "Peshawar City",
    "Peshawar"
)

df
```

### What happened?

* Pandas searched for `"Peshawar City"`.
* Every matching value was replaced with `"Peshawar"`.
* Other city values remained unchanged.

---

# Replacing Multiple Incorrect Values

We can use a dictionary to replace multiple values.

```python
df["city"] = df["city"].replace({
    "peshawar": "Peshawar",
    "PESHAWAR": "Peshawar",
    "lahore": "Lahore",
    "KARACHI": "Karachi"
})

df
```

### What happened?

* Pandas searched for each incorrect value.
* The dictionary defined the correct replacement.
* Matching values were replaced.

---

# Cleaning Incorrect Numerical Values

We can replace an incorrect value.

```python
students["Marks"] = students["Marks"].replace(
    -10,
    0
)

students
```
