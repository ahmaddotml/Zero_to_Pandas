# What is Data Visualization?

**Data visualization** means representing data using charts and graphs.

Instead of looking only at numbers:

```text
Month      Sales
January    50000
February   65000
March      72000
April      60000
```

We can create a chart to understand the data more easily.

Visualizations help us:

* Find trends
* Compare values
* Find patterns
* Identify high and low values
* Understand data quickly
* Communicate results clearly

# Pandas Plotting

Pandas provides a built-in plotting method called:

```python
plot()
```

The basic syntax is:

```python
df.plot()
```

Pandas plotting uses Matplotlib in the background.

This means that Pandas makes basic visualization easy, while Matplotlib gives us more control over charts.

---

# Import Pandas

## Learn

```python
import pandas as pd
```

For this notebook, we will also import Matplotlib so that we can display and customize charts.

```python
import matplotlib.pyplot as plt
```

---

# Create a Sample Sales DataFrame

We will use the following dataset throughout this notebook.

## Learn: Create the DataFrame

```python
data = {
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],
    "Sales": [
        50000,
        65000,
        72000,
        60000,
        80000,
        90000
    ],
    "Profit": [
        8000,
        12000,
        15000,
        10000,
        18000,
        22000
    ],
    "Orders": [
        120,
        150,
        180,
        140,
        200,
        230
    ]
}

df = pd.DataFrame(data)

df
```

---

# plot()

The `plot()` method is the main method used for plotting data in Pandas.

For example:

```python
df.plot()
```

When multiple numerical columns exist, Pandas can create lines for those columns.

We can also specify a chart type using:

```python
kind=""
```

For example:

```python
df.plot(kind="bar")
```

Common chart types include:

* `"line"`
* `"bar"`
* `"barh"`
* `"hist"`
* `"box"`
* `"scatter"`

---

# Line Charts

A **line chart** is useful for showing changes and trends over time.

For example:

* Monthly sales
* Daily temperature
* Yearly profit
* Website visitors over time

---

# Create a Basic Line Chart

## Learn

```python
df.plot(
    x="Month",
    y="Sales"
)

plt.show()
```

This creates a line chart where:

* `"Month"` is displayed on the x-axis.
* `"Sales"` is displayed on the y-axis.

---

# Add a Title and Labels

We can improve the chart by adding:

* A title
* An x-axis label
* A y-axis label

```python
df.plot(
    x="Month",
    y="Sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

---

# Figure Size

We can control the size of the chart using:

```python
figsize=(width, height)
```

For example:

```python
df.plot(
    x="Month",
    y="Sales",
    figsize=(10, 5)
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

---

# Plot Multiple Lines

We can display more than one numerical column.

## Learn: Sales and Profit

```python
df.plot(
    x="Month",
    y=["Sales", "Profit"],
    figsize=(10, 5)
)

plt.title("Sales and Profit by Month")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.show()
```

A legend is automatically created to help identify each line.

---

# Legends

A **legend** explains what each line, bar, or visual element represents.

We can display a legend using:

```python
plt.legend()
```

Example:

```python
df.plot(
    x="Month",
    y=["Sales", "Profit"]
)

plt.title("Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

A **bar chart** is useful for comparing values between categories.

---

# Create a Sample Product DataFrame

```python
products = pd.DataFrame({
    "Product": [
        "Laptop",
        "Mouse",
        "Keyboard",
        "Monitor",
        "Headphones"
    ],
    "Sales": [
        120000,
        25000,
        30000,
        80000,
        45000
    ]
})

products
```

---

# Create a Bar Chart

## Learn

```python
products.plot(
    x="Product",
    y="Sales",
    kind="bar"
)

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

The important part is:

```python
kind="bar"
```

This tells Pandas to create a vertical bar chart.

---

# Set Figure Size for a Bar Chart

```python
products.plot(
    x="Product",
    y="Sales",
    kind="bar",
    figsize=(10, 5)
)

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

---

```python
products.plot(
    x="Product",
    y=["Sales", "Profit"],
    kind="bar",
    figsize=(10, 5)
)

plt.title("Sales and Profit by Product")
plt.xlabel("Product")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

# Histograms

A **histogram** shows the distribution of numerical data.

It helps us understand:

* Where values are concentrated
* The range of values
* How frequently values occur
* Whether there are very high or low values

For example, we can analyze student marks.

---

# Create a Sample Student DataFrame

```python
students = pd.DataFrame({
    "Name": [
        "Ali",
        "Ahmed",
        "Sara",
        "Ayesha",
        "Usman",
        "Fatima",
        "Bilal",
        "Hina",
        "Hamza",
        "Maryam"
    ],
    "Marks": [
        85,
        92,
        76,
        68,
        55,
        81,
        47,
        73,
        95,
        64
    ]
})

students
```

---

# Create a Histogram

## Learn

```python
students["Marks"].plot(
    kind="hist"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

---

# Change the Number of Bins

A histogram groups values into intervals called **bins**.

```python
students["Marks"].plot(
    kind="hist",
    bins=5,
    figsize=(8, 5)
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

# Scatter Plots

A **scatter plot** is useful for finding relationships between two numerical variables.

For example:

* Study hours and marks
* Advertising cost and sales
* Temperature and ice cream sales
* Experience and salary

---

# Create a Sample DataFrame

```python
performance = pd.DataFrame({
    "Study_Hours": [
        1,
        2,
        3,
        4,
        5,
        6,
        7,
        8
    ],
    "Marks": [
        45,
        52,
        58,
        65,
        72,
        80,
        88,
        95
    ]
})

performance
```

---

# Create a Scatter Plot

## Learn

```python
performance.plot(
    x="Study_Hours",
    y="Marks",
    kind="scatter",
    figsize=(8, 5)
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

This chart can help us see whether marks generally change as study hours increase.

---

# Finding Patterns with Scatter Plots

Suppose points generally move upward from left to right.

This may indicate a **positive relationship** between the variables.

For example:

```text
More Study Hours
        ↓
Higher Marks
```

A scatter plot helps us visually investigate patterns, but we should avoid assuming that one variable directly causes changes in another variable based only on the chart.

# Complete Visualization Example

Let's combine everything.

```python
import pandas as pd
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June"
    ],
    "Sales": [
        50000,
        65000,
        72000,
        60000,
        80000,
        90000
    ],
    "Profit": [
        8000,
        12000,
        15000,
        10000,
        18000,
        22000
    ]
})

sales.plot(
    x="Month",
    y=["Sales", "Profit"],
    figsize=(10, 5)
)

plt.title("Monthly Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()

plt.show()
```

# Choosing the Right Chart

Different charts are useful for different purposes.

| Chart                | Best For                                                |
| -------------------- | ------------------------------------------------------- |
| Line chart           | Trends and changes over time                            |
| Bar chart            | Comparing categories                                    |
| Horizontal bar chart | Comparing categories with long names                    |
| Histogram            | Understanding numerical distributions                   |
| Box plot             | Comparing distributions and finding possible outliers   |
| Scatter plot         | Exploring relationships between two numerical variables |

Before creating a chart, ask:

> **What question am I trying to answer with this visualization?**
