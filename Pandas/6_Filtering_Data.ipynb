## What is Filtering?

**Filtering** means selecting only the data that matches a specific condition.

For example, imagine we want to find only students with marks greater than `80`.

Our DataFrame contains all students:

| Name   | Age | Marks | City      |
| ------ | --: | ----: | --------- |
| Ali    |  20 |    85 | Peshawar  |
| Ahmed  |  21 |    90 | Lahore    |
| Sara   |  19 |    78 | Karachi   |
| Ayesha |  22 |    92 | Islamabad |
| Usman  |  20 |    88 | Peshawar  |
| Hina   |  21 |    75 | Lahore    |

After filtering marks greater than `80`, we get only:

* Ali
* Ahmed
* Ayesha
* Usman

# Boolean Conditions

A **Boolean condition** produces either:

```text
True
```

or:

```text
False
```

For example:

```python
10 > 5
```

The result is:

```text
True
```

Another example:

```python
10 < 5
```

The result is:

```text
False
```

In Pandas, we can apply a condition to an entire column.

---

## Learn: Create a Boolean Condition

```python
df["Marks"] > 80
```

The result will look similar to:

```text
0     True
1     True
2    False
3     True
4     True
5    False
Name: Marks, dtype: bool
```

### What happened?

Pandas checks every value in the `Marks` column.

For each row:

* If marks are greater than `80` → `True`
* Otherwise → `False`

This Boolean Series can then be used to filter the DataFrame.

---

# Filtering Rows

To filter rows, place the Boolean condition inside square brackets.

Syntax:

```python
df[condition]
```

For example:

```python
df[df["Marks"] > 80]
```

---

## Learn: Find Students with Marks Greater Than 80

```python
df[df["Marks"] > 80]
```

This returns only the students whose marks are greater than `80`.

---

### What happened?

First, Pandas creates the condition:

```python
df["Marks"] > 80
```

Then we use that condition:

```python
df[df["Marks"] > 80]
```

Rows with `True` are selected.

Rows with `False` are not selected.

# Multiple Conditions

Sometimes one condition is not enough.

For example, we may want students who:

* Have marks greater than `80`
* And are older than `20`

We can combine conditions using:

* `&` for **AND**
* `|` for **OR**
* `~` for **NOT**

---

# `&` — AND

The `&` operator means **both conditions must be True**.

Example:

```python
(df["Marks"] > 80) & (df["Age"] > 20)
```

When filtering, each condition should be placed inside parentheses.

---

## Learn: Use `&`

```python
df[
    (df["Marks"] > 80) &
    (df["Age"] > 20)
]
```

This selects students who:

* Have marks greater than `80`
* AND are older than `20`

Both conditions must be True.

# `|` — OR

The `|` operator means **at least one condition must be True**.

Example:

```python
(df["Marks"] > 90) | (df["Age"] < 20)
```

---

## Learn: Use `|`

```python
df[
    (df["Marks"] >= 90) |
    (df["City"] == "Karachi")
]
```

This selects students who:

* Have marks greater than or equal to `90`
* OR are from Karachi

A row is selected if either condition is True.

# `~` — NOT

The `~` operator means **NOT**.

It reverses a Boolean condition.

For example:

```python
~(df["City"] == "Peshawar")
```

This means:

```text
City is NOT Peshawar
```

---

## Learn: Use `~`

```python
df[
    ~(df["City"] == "Peshawar")
]
```

This selects all students who are not from Peshawar.

---

## Another Example

```python
df[
    ~(df["Marks"] > 80)
]
```

This selects students who do not have marks greater than `80`.

# Filtering Missing Values

Real-world datasets may contain missing values.

Pandas represents missing values in different ways, including:

```text
NaN
None
```

We can identify missing values using:

```python
isna()
```

We can identify non-missing values using:

```python
notna()
```

---

## Create Data with Missing Values

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "Marks": [85, None, 78, 92],
    "City": ["Peshawar", "Lahore", None, "Islamabad"]
}

df_missing = pd.DataFrame(data)

df_missing
```

---

## Learn: Filter Missing Marks

```python
df_missing[
    df_missing["Marks"].isna()
]
```

This selects rows where the `Marks` value is missing.

---

## Learn: Filter Missing Cities

```python
df_missing[
    df_missing["City"].isna()
]
```

---

## Learn: Filter Non-Missing Values

```python
df_missing[
    df_missing["Marks"].notna()
]
```

This selects rows where marks are available.

---

### What happened?

* `.isna()` checks for missing values.
* `.notna()` checks for values that are not missing.
* The result is a Boolean condition.

# Filtering Text

We can filter text columns using conditions and string methods.

For example, we can search for text using:

```python
.str.contains()
```

---

## Learn: Find Names Containing `"a"`

```python
df[
    df["Name"].str.contains("a")
]
```

This finds names containing the letter `"a"`.

By default, text matching is case-sensitive.

---

## Case-Insensitive Text Filtering

To ignore uppercase and lowercase differences:

```python
df[
    df["Name"].str.contains("a", case=False)
]
```

This finds names containing `"a"` or `"A"`.

---

## Learn: Filter Cities Starting with `"P"`

```python
df[
    df["City"].str.startswith("P")
]
```

This selects cities starting with the letter `"P"`.

---

## Learn: Filter Cities Ending with `"e"`

```python
df[
    df["City"].str.endswith("e")
]
```

# Combining Multiple Conditions

We can combine different filtering methods.

For example:

```python
df[
    (df["Marks"] > 80) &
    (df["City"].isin(["Peshawar", "Lahore"]))
]
```

This selects students who:

* Have marks greater than `80`
* AND are from Peshawar or Lahore

---

## Learn: Combined Conditions

```python
df[
    (df["Marks"] >= 85) &
    (df["Age"].between(20, 21))
]
```

This selects students who:

* Have marks greater than or equal to `85`
* AND have ages between `20` and `21`
