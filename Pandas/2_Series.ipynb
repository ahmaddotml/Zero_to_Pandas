## What is a Series?

A **Series** is a one-dimensional data structure provided by Pandas.

It can store a collection of values such as:

* Numbers
* Strings
* Boolean values

Each value in a Series has an associated **index**.

You can think of a Series as a single column of a table.

Example:

```text
0    10
1    20
2    30
3    40
4    50
```

The values on the left are called the **indexes**, and the values on the right are the **data**.

---

## Learn: Create Your First Series

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers
```

### What happened?

* `pd.Series()` creates a Pandas Series.
* We passed a Python list to the Series.
* Pandas automatically created indexes starting from `0`.
* The numbers on the right are the actual values.

# Creating a Series

We create a Pandas Series using:

```python
pd.Series(data)
```

The `data` can come from different sources, including:

* A list
* A dictionary
* Another Series
* A NumPy array

In this notebook, we will focus on lists and dictionaries.

---

## Series from a List

A Python list is one of the easiest ways to create a Series.

```python
import pandas as pd

fruits = ["Apple", "Banana", "Mango", "Orange"]

fruit_series = pd.Series(fruits)

fruit_series
```

### What happened?

Pandas converted the Python list into a Series.

The result will look similar to:

```text
0     Apple
1    Banana
2     Mango
3    Orange
dtype: object
```

Pandas automatically creates indexes starting from `0`.

---

## Learn: Series from a List

```python
import pandas as pd

marks = [85, 90, 78, 92, 88]

student_marks = pd.Series(marks)

student_marks
```

### What happened?

* `marks` is a Python list.
* `pd.Series(marks)` converts the list into a Series.
* Pandas automatically adds indexes.
* The Series stores the marks in one dimension.

# Series from a Dictionary

We can also create a Series from a Python dictionary.

A dictionary contains:

* Keys
* Values

When creating a Series from a dictionary:

* Dictionary keys become the Series indexes.
* Dictionary values become the Series values.

---

## Learn: Series from a Dictionary

```python
import pandas as pd

marks = {
    "Ali": 85,
    "Ahmed": 90,
    "Sara": 88,
    "Ayesha": 92
}

student_marks = pd.Series(marks)

student_marks
```

### What happened?

* `"Ali"`, `"Ahmed"`, `"Sara"`, and `"Ayesha"` become indexes.
* Their marks become the values.
* Pandas creates the Series directly from the dictionary.

The result looks similar to:

```text
Ali       85
Ahmed     90
Sara      88
Ayesha    92
dtype: int64
```

# Series with Custom Index

By default, Pandas creates numeric indexes:

```text
0, 1, 2, 3, ...
```

However, we can create our own indexes using the `index` parameter.

Syntax:

```python
pd.Series(data, index=custom_indexes)
```

---

## Learn: Create a Series with Custom Indexes

```python
import pandas as pd

marks = [85, 90, 78, 92]

students = ["Ali", "Ahmed", "Sara", "Ayesha"]

student_marks = pd.Series(marks, index=students)

student_marks
```

### What happened?

* The `marks` list contains the values.
* The `students` list contains the custom indexes.
* `index=students` tells Pandas to use student names as indexes.

The result looks similar to:

```text
Ali       85
Ahmed     90
Sara      78
Ayesha    92
dtype: int64
```

---

## Execute Yourself

Create a Series containing five subjects and their marks.

```python
import pandas as pd

marks = [85, 90, 80, 88, 92]

subjects = [
    "English",
    "Mathematics",
    "Physics",
    "Chemistry",
    "Computer"
]

subject_marks = pd.Series(marks, index=subjects)

subject_marks
```

We can access values from a Series using indexes.

# Accessing Values with Custom Indexes

When a Series has custom indexes, we can use those indexes to access values.

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 78],
    index=["Ali", "Ahmed", "Sara"]
)

marks["Ali"]
```

# `head()` and `tail()`

When working with large data, displaying the complete Series may not be useful.

Pandas provides:

* `head()` to display values from the beginning.
* `tail()` to display values from the end.

By default, both methods display `5` values.

---

## Learn: `head()`

```python
import pandas as pd

numbers = pd.Series(range(1, 21))

numbers.head()
```

This displays the first `5` values.

---

## Display a Specific Number of Values

We can provide a number inside `head()`.

```python
numbers.head(3)
```

This displays the first `3` values.

---

## Learn: `tail()`

```python
import pandas as pd

numbers = pd.Series(range(1, 21))

numbers.tail()
```

This displays the last `5` values.

---

## Display a Specific Number of Values

```python
numbers.tail(3)
```

This displays the last `3` values.

# Series Data Types

A Series also has a data type.

We can check the data type using:

```python
series.dtype
```

Common data types include:

* `int64` — integers
* `float64` — decimal numbers
* `object` — usually text data
* `bool` — True or False values

---

## Learn: Integer Data Type

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40])

numbers.dtype
```

The result will normally show an integer data type.

---

## Learn: Float Data Type

```python
import pandas as pd

prices = pd.Series([10.5, 20.0, 30.75])

prices.dtype
```

The result will normally show a floating-point data type.

---

## Learn: Text Data Type

```python
import pandas as pd

names = pd.Series(["Ali", "Ahmed", "Sara"])

names.dtype
```

Text values are commonly represented by an object or string-related dtype depending on the Pandas version and how the Series is created.

---

## Learn: Boolean Data Type

```python
import pandas as pd

results = pd.Series([True, False, True, True])

results.dtype
```

# Basic Series Operations

Pandas allows us to perform operations on Series.

For example:

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40])

numbers + 5
```

The result adds `5` to every value.

## Learn: Addition

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers + 10
```

## Learn: Subtraction

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers - 5
```

---

## Learn: Multiplication

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers * 2
```

---

## Learn: Division

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

numbers / 2
```

### What happened?

Pandas applies the operation to every value in the Series.

For example:

```python
numbers * 2
```

```text
10 → 20
20 → 40
30 → 60
40 → 80
50 → 100
```
