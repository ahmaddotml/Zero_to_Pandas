# Complete List of `pd.read_*` Functions

Pandas provides a dedicated `pd.read_*` function for loading data from many common file formats and data sources.

> **Tip:** In a Jupyter Notebook (`.ipynb`), these functions are commonly used to load external data into a **DataFrame**.

---

## 📚 Complete `pd.read_*` Function Reference

| Function                  | Data Format / Source     | Common Use Case                                                                |
| ------------------------- | ------------------------ | ------------------------------------------------------------------------------ |
| **`pd.read_csv()`**       | **CSV / Delimited Text** | Plain-text tabular data such as `.csv`, `.txt`, and `.tsv` files               |
| **`pd.read_json()`**      | **JSON**                 | Web APIs, semi-structured data, and nested configurations                      |
| **`pd.read_excel()`**     | **Microsoft Excel**      | Excel spreadsheets such as `.xlsx`, `.xls`, and `.xlsm`                        |
| **`pd.read_html()`**      | **HTML Tables**          | Extracting `<table>` elements from websites or HTML files                      |
| **`pd.read_sql()`**       | **SQL Database**         | Reading data from relational databases such as PostgreSQL, MySQL, and SQLite   |
| **`pd.read_parquet()`**   | **Apache Parquet**       | Fast, columnar storage commonly used in analytics and big-data workflows       |
| **`pd.read_feather()`**   | **Apache Arrow Feather** | Fast binary storage, especially useful for local data exchange and caching     |
| **`pd.read_hdf()`**       | **HDF5**                 | Large, hierarchical scientific and numerical datasets                          |
| **`pd.read_pickle()`**    | **Python Pickle**        | Loading Python/Pandas objects previously serialized to disk                    |
| **`pd.read_clipboard()`** | **System Clipboard**     | Loading tabular data copied from spreadsheets, websites, or other applications |
| **`pd.read_xml()`**       | **XML**                  | Reading structured data stored in `.xml` files                                 |
| **`pd.read_stata()`**     | **Stata**                | Loading Stata statistical datasets such as `.dta`                              |
| **`pd.read_sas()`**       | **SAS**                  | Loading SAS datasets such as `.sas7bdat`                                       |
| **`pd.read_spss()`**      | **SPSS**                 | Loading IBM SPSS statistical files such as `.sav`                              |
| **`pd.read_orc()`**       | **Apache ORC**           | Reading columnar data optimized for big-data and Hadoop workloads              |
| **`pd.read_fwf()`**       | **Fixed-Width Text**     | Reading text files where columns are defined by fixed character widths         |

---

## ⭐ Most Commonly Used Readers

For everyday data analysis, these are the readers you will use most often:


# 🔍 Basic DataFrame Overview

After loading a dataset into a Pandas DataFrame, these are the most useful commands for getting a **quick overview of the data**.

```python
import pandas as pd

df = pd.read_csv("data.csv")
```

---

## 1. `df.head()`

Shows the **first 5 rows** of the DataFrame.

```python
df.head()
```

You can specify the number of rows:

```python
df.head(10)
```

**Use:** Quickly see what the data looks like.

---

## 2. `df.tail()`

Shows the **last 5 rows** of the DataFrame.

```python
df.tail()
```

You can specify the number of rows:

```python
df.tail(10)
```

**Use:** Check the end of the dataset.

---

## 3. `df.shape`

Returns the **number of rows and columns**.

```python
df.shape
```

Example:

```text
(1000, 8)
```

This means:

* `1000` → rows
* `8` → columns

> **Note:** `shape` is an attribute, so don't use `()`.

---

## 4. `df.info()`

Provides a **summary of the DataFrame**.

```python
df.info()
```

It shows:

* Number of rows
* Column names
* Non-null values
* Data types
* Memory usage

**Use:** Understand the overall structure of your dataset.

---

## 5. `df.describe()`

Provides **basic statistical information** about numerical columns.

```python
df.describe()
```

It includes:

* `count` → number of values
* `mean` → average
* `std` → standard deviation
* `min` → minimum
* `25%` → first quartile
* `50%` → median
* `75%` → third quartile
* `max` → maximum

**Use:** Get a quick statistical overview of numerical data.

---

## 6. `df.columns`

Shows the **names of all columns**.

```python
df.columns
```

You can convert them into a normal Python list:

```python
df.columns.tolist()
```

**Use:** Quickly see what variables/features are available.

---

## 7. `df.dtypes`

Shows the **data type of each column**.

```python
df.dtypes
```

Example:

```text
Name       object
Age         int64
Salary    float64
City       object
```

**Use:** Check whether columns contain numbers, text, dates, etc.

---

## 8. `df.isnull().sum()`

Counts the **missing values in each column**.

```python
df.isnull().sum()
```

Example:

```text
Name       0
Age       15
Salary     8
City       0
```

**Use:** Quickly identify missing data.

---
