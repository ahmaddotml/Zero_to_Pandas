# What is a Pivot Table?

A **pivot table** summarizes data by reorganizing it into rows and columns.

For example, suppose we have this sales data:

```text
Region      Product      Sales
North       Laptop       50000
North       Mouse        2000
South       Laptop       45000
South       Mouse        2500
```

We can create a pivot table:

```text
Product     Laptop      Mouse
Region
North       50000       2000
South       45000       2500
```

This makes the data easier to compare.

We can quickly answer questions such as:

* Which region has the highest laptop sales?
* Which product has the highest sales?
* How do products perform in different regions?

---

# Creating a Sample Sales DataFrame

Let's create a DataFrame that we will use throughout this notebook.

## Learn: Create a Sales DataFrame

```python
import pandas as pd

data = {
    "Region": [
        "North",
        "North",
        "South",
        "South",
        "East",
        "East",
        "West",
        "West"
    ],
    "Product": [
        "Laptop",
        "Mouse",
        "Laptop",
        "Mouse",
        "Laptop",
        "Keyboard",
        "Mouse",
        "Keyboard"
    ],
    "Sales": [
        50000,
        2000,
        45000,
        2500,
        40000,
        3500,
        1800,
        4000
    ]
}

df = pd.DataFrame(data)

df
```

### What happened?

* We imported Pandas as `pd`.
* We created a sales DataFrame.
* Each row contains a region.
* Each row contains a product.
* Each row contains a sales amount.
* We will use this data to create pivot tables.

# Understanding Rows, Columns, and Values

A pivot table usually contains three important parts:

* **Rows** — categories displayed vertically.
* **Columns** — categories displayed horizontally.
* **Values** — numerical data used for calculations.

For example:

```text
Product     Laptop      Mouse
Region
North       50000       2000
South       45000       2500
```

Here:

* `"Region"` is the row.
* `"Product"` is the column.
* `"Sales"` contains the values.

In Pandas, these are commonly written as:

```python
index="Region"
columns="Product"
values="Sales"
```

---

# pivot()

The `pivot()` method reshapes data into rows and columns.

The basic syntax is:

```python
df.pivot(
    index="Row_Column",
    columns="Column_Column",
    values="Value_Column"
)
```

For example:

```python
df.pivot(
    index="Region",
    columns="Product",
    values="Sales"
)
```

---

# Creating a Pivot Table with pivot()

## Learn: Pivot Sales by Region and Product

```python
pivot_result = df.pivot(
    index="Region",
    columns="Product",
    values="Sales"
)

pivot_result
```

### What happened?

* Pandas used `"Region"` as the rows.
* Pandas used `"Product"` as the columns.
* Pandas placed `"Sales"` values inside the table.
* Each region and product combination was displayed.

---

# Important Limitation of pivot()

`pivot()` works when every row-column combination has only one value.

For example:

```text
Region      Product      Sales
North       Laptop       50000
North       Laptop       30000
```

There are two sales values for the same:

* Region: `North`
* Product: `Laptop`

In this situation, `pivot()` cannot decide which value to use.

For repeated combinations, we should use `pivot_table()`.

# pivot_table()

The `pivot_table()` method is more flexible than `pivot()`.

It can handle repeated values by performing an aggregation.

The basic syntax is:

```python
df.pivot_table(
    index="Row_Column",
    columns="Column_Column",
    values="Value_Column",
    aggfunc="mean"
)
```

By default, `pivot_table()` uses `mean` as the aggregation function.

---

# Creating a Sample DataFrame with Repeated Values

```python
sales_data = {
    "Region": [
        "North",
        "North",
        "North",
        "South",
        "South",
        "East",
        "East",
        "West"
    ],
    "Product": [
        "Laptop",
        "Laptop",
        "Mouse",
        "Laptop",
        "Mouse",
        "Laptop",
        "Keyboard",
        "Mouse"
    ],
    "Sales": [
        50000,
        30000,
        2000,
        45000,
        2500,
        40000,
        3500,
        1800
    ]
}

sales_df = pd.DataFrame(sales_data)

sales_df
```

Notice that `"North"` and `"Laptop"` appear more than once.

We can use `pivot_table()` to handle this situation.

---

# Basic pivot_table()

## Learn: Create a Pivot Table

```python
pivot_table_result = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales"
)

pivot_table_result
```

Since the default aggregation is `mean`, Pandas calculates the average when repeated combinations exist.

---

# Aggregation with pivot_table()

We can choose different aggregation functions.

Common functions include:

* `"sum"`
* `"mean"`
* `"min"`
* `"max"`
* `"count"`

---

# Using sum()

## Learn: Calculate Total Sales

```python
total_sales = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum"
)

total_sales
```

This calculates total sales for each region and product.

---

# Using mean()

```python
average_sales = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="mean"
)

average_sales
```

This calculates average sales.

---

# Using min()

```python
minimum_sales = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="min"
)

minimum_sales
```

This finds the smallest sales value for each combination.

---

# Using max()

```python
maximum_sales = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="max"
)

maximum_sales
```

This finds the largest sales value for each combination.

---

# Using count()

```python
sales_count = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="count"
)

sales_count
```

This counts sales records for each combination.

---

# Using Multiple Aggregation Functions

We can also perform multiple calculations.

```python
result = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc=["sum", "mean", "max"]
)

result
```

This creates a table containing:

* Total sales
* Average sales
* Maximum sales

# Filling Missing Values in a Pivot Table

Sometimes some combinations do not exist.

For example:

```text
Region      Product
North       Keyboard
```

may have no sales record.

Pandas displays these missing combinations as `NaN`.

We can replace them using `fill_value`.

## Learn: Fill Missing Values

```python
result = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

result
```

Now missing combinations are displayed as `0`.

---

# Adding Totals with margins

We can add totals to rows and columns using:

```python
margins=True
```

## Learn: Add Grand Totals

```python
result = sales_df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

result
```

The `"All"` row and column contain totals.

# crosstab()

The `pd.crosstab()` function is used to compare two or more categories.

It creates a frequency table showing how often combinations occur.

For example, suppose we have:

```text
Student      Gender      Result
Ali          Male        Pass
Ahmed        Male        Pass
Sara         Female      Pass
Hina         Female      Fail
```

We can compare:

* Gender
* Result

---

# Creating a Sample Student DataFrame

```python
students = pd.DataFrame({
    "Name": [
        "Ali",
        "Ahmed",
        "Sara",
        "Hina",
        "Usman",
        "Ayesha",
        "Bilal",
        "Fatima"
    ],
    "Gender": [
        "Male",
        "Male",
        "Female",
        "Female",
        "Male",
        "Female",
        "Male",
        "Female"
    ],
    "Result": [
        "Pass",
        "Pass",
        "Pass",
        "Fail",
        "Fail",
        "Pass",
        "Pass",
        "Fail"
    ]
})

students
```

---

# Basic crosstab()

## Learn: Compare Gender and Result

```python
result_table = pd.crosstab(
    students["Gender"],
    students["Result"]
)

result_table
```

### What happened?

* Pandas used `"Gender"` as the rows.
* Pandas used `"Result"` as the columns.
* Pandas counted how many records belong to each combination.

For example:

```text
          Fail    Pass
Female      2       2
Male        1       3
```

This makes category comparison easier.

---

# Comparing Categories with crosstab()

We can use `crosstab()` for many comparisons.

For example:

* Gender and Result
* City and Department
* Product and Region
* Class and Grade

The general syntax is:

```python
pd.crosstab(
    df["Category_1"],
    df["Category_2"]
)
```

---

# Adding Row and Column Totals

We can add totals using:

```python
margins=True
```

## Learn: Add Totals

```python
result_table = pd.crosstab(
    students["Gender"],
    students["Result"],
    margins=True
)

result_table
```

This adds totals for:

* Every row
* Every column
* The complete dataset

---

# crosstab() with Normalized Values

We can also calculate percentages.

```python
result_table = pd.crosstab(
    students["Gender"],
    students["Result"],
    normalize=True
)

result_table
```

This shows the proportion of each combination instead of only counts.

# pivot() vs pivot_table()

Both methods create pivot-style tables, but there is an important difference.

| Method          | Main Purpose                                      |
| --------------- | ------------------------------------------------- |
| `pivot()`       | Reshapes data when combinations are unique        |
| `pivot_table()` | Summarizes data and handles repeated combinations |

For example:

Use `pivot()` when:

```text
North + Laptop → Only one value
```

Use `pivot_table()` when:

```text
North + Laptop → Multiple values
```

because an aggregation function can be used.

---

# pivot_table() vs crosstab()

| Method          | Main Purpose                   |
| --------------- | ------------------------------ |
| `pivot_table()` | Summarizes numerical values    |
| `crosstab()`    | Compares and counts categories |

Use `pivot_table()` for questions such as:

* What are total sales by region and product?
* What are average marks by class and subject?

Use `crosstab()` for questions such as:

* How many male and female students passed or failed?
* How many employees belong to each department and city?
