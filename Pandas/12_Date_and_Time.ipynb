# Introduction to Dates

Dates are very common in real-world datasets.

For example:

* Student admission dates
* Employee joining dates
* Product order dates
* Sales dates
* Customer registration dates

A date may initially be stored as text.

For example:

```text
2026-01-15
2026-02-20
2026-03-10
```

To properly work with dates, we usually convert them into the `datetime` data type.

---

# Creating a Sample Sales DataFrame

Let's create a DataFrame containing sales information and dates.

# Understanding datetime

`datetime` is used to represent dates and times.

A datetime value can contain information such as:

* Year
* Month
* Day
* Hour
* Minute
* Second

For example:

```text
2026-02-20 10:30:00
```

In Pandas, datetime values allow us to perform date-related operations easily.

---

# Date Differences

We can calculate the difference between two dates.

This is useful for finding:

* Number of days between orders
* Employee working duration
* Customer registration duration
* Delivery time

## Learn: Calculate the Difference Between Two Dates

```python
dates = pd.DataFrame({
    "Start_Date": [
        "2026-01-01",
        "2026-02-10",
        "2026-03-15"
    ],
    "End_Date": [
        "2026-01-10",
        "2026-02-20",
        "2026-03-25"
    ]
})

dates
```

First, convert both columns to datetime.

```python
dates["Start_Date"] = pd.to_datetime(
    dates["Start_Date"]
)

dates["End_Date"] = pd.to_datetime(
    dates["End_Date"]
)
```

Now calculate the difference.

```python
dates["Days_Difference"] = (
    dates["End_Date"] - dates["Start_Date"]
).dt.days

dates
```

### What happened?

* Both date columns were converted to datetime.
* The start date was subtracted from the end date.
* Pandas calculated the time difference.
* `.dt.days` extracted the number of days.

# Filtering by Dates

We can filter a DataFrame to find records from a specific date.

## Learn: Find Records from One Date

```python
result = df[
    df["Date"] == "2026-02-20"
]

result
```

Since the `"Date"` column is datetime, we can also write:

```python
result = df[
    df["Date"] == pd.Timestamp("2026-02-20")
]

result
```

---

# Finding Records After a Specific Date

```python
result = df[
    df["Date"] > "2026-02-01"
]

result
```

This returns records after `2026-02-01`.

---

# Finding Records Before a Specific Date

```python
result = df[
    df["Date"] < "2026-03-01"
]

result
```

This returns records before `2026-03-01`.

---

# Finding Records Between Two Dates

```python
result = df[
    (df["Date"] >= "2026-02-01") &
    (df["Date"] <= "2026-03-31")
]

result
```

### What happened?

* Pandas checked every date.
* The first condition selected dates from `2026-02-01`.
* The second condition selected dates until `2026-03-31`.
* Both conditions together returned records between the two dates.
