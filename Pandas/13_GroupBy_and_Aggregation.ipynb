# What is groupby()?

The `groupby()` method divides data into groups based on common values.

For example, suppose we have student data:

```text
Name       Class      Marks
Ali        10A        85
Ahmed      10A        90
Sara       10B        78
Ayesha     10B        88
Usman      10A        95
```

We can group students by `"Class"`.

After grouping, Pandas can calculate information for each class.

For example:

```text
Class      Average Marks
10A        90.0
10B        83.0
```

The basic structure is:

```python
df.groupby("Column")["Value_Column"].calculation()
```

For example:

```python
df.groupby("Class")["Marks"].mean()
```

This means:

* Group the data by `"Class"`.
* Select the `"Marks"` column.
* Calculate the average marks for each class.

# Basic groupby()

The basic syntax of `groupby()` is:

```python
df.groupby("Column")
```

For example:

```python
df.groupby("Class")
```

This creates groups based on the `"Class"` column.

Usually, we combine `groupby()` with a calculation such as:

* `sum()`
* `mean()`
* `min()`
* `max()`
* `count()`

---

# Grouping by One Column

We can group data using one column.

## Learn: Calculate Average Marks by Class

```python
average_marks = df.groupby("Class")["Marks"].mean()

average_marks
```

### What happened?

* Pandas grouped all students by `"Class"`.
* Students from `"10A"` were placed in one group.
* Students from `"10B"` were placed in another group.
* Pandas selected the `"Marks"` column.
* `mean()` calculated the average marks for each class.

---

# Grouping by One Column with sum()

We can calculate the total marks for each class.

```python
total_marks = df.groupby("Class")["Marks"].sum()

total_marks
```

### What happened?

* The data was grouped by `"Class"`.
* Pandas added the marks of all students in each class.
* The result shows total marks for every class.

# Using count()

The `count()` method counts non-missing values.

## Learn: Count Students by Class

```python
student_count = df.groupby(
    "Class"
)["Name"].count()

student_count
```

### What happened?

* Pandas grouped students by class.
* `"Name"` values were counted in each group.
* Missing values are not included in `count()`.

---

# Using size()

The `size()` method counts the total number of rows in each group.

## Learn: Find Group Size

```python
group_size = df.groupby(
    "Class"
).size()

group_size
```

### What happened?

* Pandas grouped the rows by `"Class"`.
* `size()` counted every row in each group.
* Unlike `count()`, `size()` counts rows even when some values are missing.

---

# Difference Between count() and size()

Suppose we have:

```text
Name       Class      Marks
Ali        10A        85
Ahmed      10A        NaN
Sara       10B        78
```

Using:

```python
df.groupby("Class")["Marks"].count()
```

The missing `NaN` mark is not counted.

Using:

```python
df.groupby("Class").size()
```

Every row is counted.

So:

* `count()` counts non-missing values.
* `size()` counts all rows.

# Grouping by Multiple Columns

We can group data using more than one column.

For example, we can group by:

* `"City"`
* `"Class"`

## Learn: Group by City and Class

```python
result = df.groupby(
    ["City", "Class"]
)["Marks"].mean()

result
```

### What happened?

* Pandas first grouped data by `"City"`.
* Inside each city, it grouped students by `"Class"`.
* `mean()` calculated average marks for every city and class combination.

---

# Multiple Column Grouping with reset_index()

The result of grouping by multiple columns often uses a hierarchical index.

We can convert it into a regular DataFrame using `reset_index()`.

```python
result = (
    df.groupby(
        ["City", "Class"]
    )["Marks"]
    .mean()
    .reset_index()
)

result
```

# Aggregation

Aggregation means performing calculations on grouped data.

Common aggregation functions include:

* `sum()`
* `mean()`
* `median()`
* `min()`
* `max()`
* `count()`

We can use one aggregation function or multiple aggregation functions.

---

# Using agg()

The `agg()` method allows us to perform multiple calculations at the same time.

## Learn: Multiple Calculations

```python
result = df.groupby(
    "Class"
)["Marks"].agg(
    ["sum", "mean", "min", "max", "count"]
)

result
```

### What happened?

For each class, Pandas calculated:

* Total marks using `sum`.
* Average marks using `mean`.
* Lowest marks using `min`.
* Highest marks using `max`.
* Number of marks using `count`.

---

# Using agg() with median

We can also include `median()`.

```python
result = df.groupby(
    "Class"
)["Marks"].agg(
    ["sum", "mean", "median", "min", "max", "count"]
)

result
```

---

# Giving Custom Names to Aggregated Columns

We can give meaningful names to the results.

```python
result = df.groupby(
    "Class"
).agg(
    Total_Marks=("Marks", "sum"),
    Average_Marks=("Marks", "mean"),
    Median_Marks=("Marks", "median"),
    Lowest_Marks=("Marks", "min"),
    Highest_Marks=("Marks", "max"),
    Student_Count=("Name", "count")
)

result
```

### What happened?

* The data was grouped by `"Class"`.
* Multiple calculations were performed.
* Each result received a custom column name.
* The final result is easier to read.
