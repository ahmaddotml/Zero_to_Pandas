# Creating a Sample DataFrame

Before sorting and ranking data, let's create a sample student DataFrame.

* `pd.DataFrame()` converted the dictionary into a DataFrame.

# Sorting Data with sort_values()

The `sort_values()` method is used to sort a DataFrame based on the values of one or more columns.

The basic syntax is:

```python
df.sort_values("Column_Name")
```

---

# Sorting in Ascending Order

By default, `sort_values()` sorts values in ascending order.

For numbers:

```text
Smallest → Largest
```

For text:

```text
A → Z
```

## Learn: Sort Students by Marks

```python
sorted_df = df.sort_values("Marks")

sorted_df
```

### What happened?

* Pandas looked at the `"Marks"` column.
* The students were sorted from the lowest marks to the highest marks.
* The original `df` was not changed.
* The sorted DataFrame was stored in `sorted_df`.

# Sorting in Descending Order

We can sort data from largest to smallest using `ascending=False`.

The syntax is:

```python
df.sort_values("Column_Name", ascending=False)
```

## Learn: Sort Marks in Descending Order

```python
sorted_df = df.sort_values(
    "Marks",
    ascending=False
)

sorted_df
```

### What happened?

* Pandas looked at the `"Marks"` column.
* `ascending=False` changed the sorting order.
* The highest marks appear first.
* The lowest marks appear last.

# Sorting by Multiple Columns

We can sort a DataFrame using more than one column.

To do this, we provide the column names inside a list.

## Learn: Sort by City and Marks

```python
sorted_df = df.sort_values(
    ["City", "Marks"]
)

sorted_df
```

### What happened?

* Pandas first sorted the students by `"City"`.
* Students with the same city were then sorted by `"Marks"`.
* The columns inside the list determine the sorting priority.

---

# Sorting Multiple Columns with Different Orders

We can use different sorting orders for different columns.

## Learn: Sort City Ascending and Marks Descending

```python
sorted_df = df.sort_values(
    ["City", "Marks"],
    ascending=[True, False]
)

sorted_df
```

### What happened?

* `"City"` was sorted in ascending order.
* `"Marks"` was sorted in descending order.
* `True` means ascending order.
* `False` means descending order.

# Sorting Indexes with sort_index()

The `sort_index()` method is used to sort a DataFrame based on its index.

## Learn: Create an Unsorted Index

```python
students = pd.DataFrame(
    {
        "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
        "Marks": [85, 92, 78, 88]
    },
    index=[3, 1, 4, 2]
)

students
```

The indexes are not in order.

Now sort them.

```python
sorted_students = students.sort_index()

sorted_students
```

### What happened?

* Pandas looked at the row indexes.
* The indexes were sorted from smallest to largest.
* The rows moved according to their indexes.

---

# Sorting Indexes in Descending Order

We can also sort indexes from largest to smallest.

```python
sorted_students = students.sort_index(
    ascending=False
)

sorted_students
```

### What happened?

* `ascending=False` was used.
* The largest index appears first.
* The smallest index appears last.

# Ranking Data

Ranking gives a position to each value.

For example:

```text
Highest Marks → Rank 1
Second Highest → Rank 2
Third Highest → Rank 3
```

We can rank data using the `rank()` method.

---

# Using rank()

## Learn: Rank Students by Marks

```python
df["Rank"] = df["Marks"].rank(
    ascending=False
)

df
```

### What happened?

* Pandas looked at the `"Marks"` column.
* `ascending=False` gave the highest mark rank `1`.
* Lower marks received higher rank numbers.
* The ranks were stored in a new `"Rank"` column.

---

# Ranking in Ascending Order

By default, `rank()` gives the smallest value rank `1`.

## Learn: Rank Marks from Lowest to Highest

```python
df["Ascending_Rank"] = df["Marks"].rank()

df
```

### What happened?

* The smallest mark received rank `1`.
* The next smallest mark received rank `2`.
* The ranks increased with the marks.

# Finding Top Values

We can find the highest values after sorting data in descending order.

## Learn: Find the Top 3 Students

```python
top_students = df.sort_values(
    "Marks",
    ascending=False
).head(3)

top_students
```

### What happened?

* The DataFrame was sorted by `"Marks"` in descending order.
* The highest marks appeared first.
* `head(3)` selected the first three rows.
* These rows contain the top three students.

---

# Finding the Top 5 Students

```python
top_5_students = df.sort_values(
    "Marks",
    ascending=False
).head(5)

top_5_students
```

### What happened?

* Students were sorted from highest to lowest marks.
* `head(5)` selected the first five students.
* The result contains the top five students.

# Finding Bottom Values

We can find the lowest values by sorting data in ascending order.

## Learn: Find the Bottom 3 Students

```python
bottom_students = df.sort_values(
    "Marks"
).head(3)

bottom_students
```

### What happened?

* The DataFrame was sorted from lowest to highest marks.
* The lowest marks appeared first.
* `head(3)` selected the first three rows.
* These rows contain the bottom three students.

---

# Using tail() to Find Bottom Values

We can also sort in descending order and use `tail()`.

```python
bottom_students = df.sort_values(
    "Marks",
    ascending=False
).tail(3)

bottom_students
```

### What happened?

* The DataFrame was sorted from highest to lowest marks.
* The lowest marks moved to the bottom.
* `tail(3)` selected the last three rows.
