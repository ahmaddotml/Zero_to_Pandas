# Reading CSV Files

CSV stands for **Comma-Separated Values**.

A CSV file stores tabular data in a simple text format.

For example:

```text
Name,Age,Marks
Ali,20,85
Ahmed,21,90
Sara,19,88
```

Pandas can read a CSV file using:

```python
pd.read_csv()
```

---

## Learn: Read a CSV File

```python
import pandas as pd

df = pd.read_csv("students.csv")

df
```

### What happened?

* `pd.read_csv()` reads a CSV file.
* `"students.csv"` is the file name.
* The data is loaded into a DataFrame.
* `df` stores the DataFrame.

# Writing CSV Files

After creating or modifying a DataFrame, we can save it as a CSV file.

Pandas uses:

```python
df.to_csv()
```

---

## Learn: Save a DataFrame as CSV

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df.to_csv("students.csv")
```

This creates a file named:

```text
students.csv
```

---

## Understanding the Index

By default, Pandas also saves the DataFrame index.

For example:

```text
,Name,Marks
0,Ali,85
1,Ahmed,90
2,Sara,88
```

Usually, we do not want to save the default index.

We can use:

```python
index=False
```

---

## Learn: Save CSV Without the Index

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara"],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)
```

### What happened?

* `to_csv()` saves the DataFrame.
* `"students.csv"` is the output file name.
* `index=False` prevents Pandas from saving the row indexes.

---

# File Paths

A **file path** tells Python where a file is located.

For example:

```text
students.csv
```

or:

```text
data/students.csv
```

or:

```text
C:\Users\Username\Documents\students.csv
```

Pandas needs the correct path to find and read a file.

---

# Relative Paths

A **relative path** is based on your current working folder.

Suppose your project looks like this:

```text
project/
│
├── 4_Reading_and_Writing_Data.ipynb
│
└── data/
    └── students.csv
```

You can read the file using:

```python
df = pd.read_csv("data/students.csv")
```

The path starts from the folder where the notebook is running.

---

## Learn: Relative Path

```python
import pandas as pd

df = pd.read_csv("data/students.csv")

df
```

---

# Absolute Paths

An **absolute path** contains the complete location of a file.

Example on Windows:

```text
C:\Users\Username\Documents\project\data\students.csv
```

Example:

```python
df = pd.read_csv(
    r"C:\Users\Username\Documents\project\data\students.csv"
)
```

The `r` before the string creates a raw string, which helps Windows paths work correctly.

---

## Learn: Absolute Path

```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\Username\Documents\students.csv"
)

df
```

Replace the path with the actual location of your file.

---

# Relative vs Absolute Paths

| Relative Path                | Absolute Path                          |
| ---------------------------- | -------------------------------------- |
| Based on the current folder  | Contains the complete file location    |
| Shorter                      | Usually longer                         |
| Better for portable projects | Depends on the computer                |
| Good for GitHub projects     | Can break when shared                  |
| Example: `data/file.csv`     | Example: `C:\Users\Name\data\file.csv` |

For GitHub projects, relative paths are usually better because your project can work on different computers.

# Common File-Reading Errors

When loading external files, you may encounter errors.

Understanding these errors will help you solve problems quickly.

---

## Error 1: `FileNotFoundError`

Example:

```text
FileNotFoundError: [Errno 2] No such file or directory
```

This usually means:

* The file does not exist at the given location.
* The file name is incorrect.
* The file extension is incorrect.
* The file path is incorrect.

### Check Your File

If the file is:

```text
students.csv
```

Make sure you are not writing:

```python
pd.read_csv("student.csv")
```

The names are different.

---

## Error 2: Incorrect File Extension

A file might be named:

```text
students.csv
```

but you try to read:

```python
pd.read_excel("students.csv")
```

Use the correct function for the correct file format.

| File Type | Read              | Write           |
| --------- | ----------------- | --------------- |
| CSV       | `pd.read_csv()`   | `df.to_csv()`   |
| Excel     | `pd.read_excel()` | `df.to_excel()` |
| JSON      | `pd.read_json()`  | `df.to_json()`  |

---

## Error 3: Missing Excel Dependency

You may get an error when working with `.xlsx` files if the required Excel package is not installed.

Install:

```python
pip install openpyxl
```

Or with `uv`:

```python
uv add openpyxl
```

Then restart the notebook kernel if necessary.

---

## Error 4: Wrong Working Directory

Sometimes your file exists, but Pandas still cannot find it.

You can check your current working directory using:

```python
import os

print(os.getcwd())
```

Make sure your file is located in the correct folder or provide the correct relative path.

# Basic Reading and Writing Workflow

A common workflow is:

### Step 1: Read the data

```python
df = pd.read_csv("students.csv")
```

### Step 2: Inspect the data

```python
df.info()
```

### Step 3: Modify the data

```python
df["Bonus Marks"] = 5
```

### Step 4: Save the modified data

```python
df.to_csv(
    "modified_students.csv",
    index=False
)
```

---

## Learn: Complete CSV Workflow

```python
import pandas as pd

# Step 1: Read the CSV file
df = pd.read_csv("students.csv")

# Step 2: Display the data
print(df)

# Step 3: Add a new column
df["Bonus Marks"] = 5

# Step 4: Save the modified data
df.to_csv(
    "modified_students.csv",
    index=False
)
```

### What happened?

* Pandas loaded the original dataset.
* We added a new column called `"Bonus Marks"`.
* Every row received the value `5`.
* The modified dataset was saved as a new CSV file.

The original file remains unchanged.
