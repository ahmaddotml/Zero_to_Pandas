# What is Data Analysis?

**Data analysis** is the process of examining data to find useful information.

For example, if we have student data, we may ask:

* How many students are there?
* What is the average marks?
* Who has the highest marks?
* Which students failed?
* Which city has the highest average marks?
* Are there any missing values?
* What patterns can we find?

Pandas helps us answer these questions.

A typical data analysis process looks like this:

```text id="q18a01"
Load Data
    ↓
Explore Data
    ↓
Clean Data
    ↓
Ask Questions
    ↓
Filter Data
    ↓
Group Data
    ↓
Analyze and Aggregate
    ↓
Find Patterns
    ↓
Draw Conclusions
```

---

# Our Mini Project: Student Performance Analysis

For this notebook, we will use a **student performance dataset**.

The same analysis process can also be used with:

* Sales data
* E-commerce data
* Employee data
* Weather data

Our dataset will contain:

* Student ID
* Name
* Gender
* Class
* City
* Subject
* Marks
* Attendance

---

# Step 1: Import Pandas

First, import Pandas.

## Learn

```python id="p18a01"
import pandas as pd
```

---

# Fill Missing Values

We can fill missing marks with the average marks.

```python id="p18a32"
df["Marks"] = df["Marks"].fillna(
    df["Marks"].mean()
)

df
```

### What happened?

* Pandas found the missing value in `"Marks"`.
* The missing value was replaced with the average marks.
* The dataset now has fewer missing-value problems.

---

# Remove Missing Values

Another option is:

```python id="p18a33"
clean_df = df.dropna()
```

This removes rows containing missing values.

However, removing data is not always the best choice. Always think about the dataset before choosing between `dropna()` and `fillna()`.

This provides multiple statistics for every city.

---

# Analyze Multiple Columns

We can also analyze different columns.

```python id="p18a44"

    "Marks": ["mean", "min", "max"],
    "Attendance": ["mean", "min", "max"]
})
```

---

# Step 10: Finding Patterns

Data analysis is not only about calculations.

We should also look for patterns.

For example:

* Which city has the highest average marks?
* Do students with higher attendance usually have higher marks?
* Which class performs better?
* Which students need improvement?

---

# Compare Attendance and Marks

Select both columns:

```python id="p18a45"
df[
    ["Marks", "Attendance"]
]
```

We can calculate the correlation:

```python id="p18a46"

    ["Marks", "Attendance"]
].corr()
```

The result helps us examine the relationship between numerical columns.

---

# Find Students with High Attendance but Low Marks

```python id="p18a47"
df[
    (df["Attendance"] >= 85) &
    (df["Marks"] < 60)
]
```

This may help identify students who attend regularly but need academic support.

This allows us to compare academic performance and attendance across cities.

# Step 11: Create Categories

We can create categories from existing data.

For example, create a performance category.

## Learn

```python id="p18a50"
def get_performance(mark):
    if mark >= 85:
        return "Excellent"
    elif mark >= 70:
        return "Good"
    elif mark >= 50:
        return "Average"
    else:
        return "Needs Improvement"
```

Apply the function:

```python id="p18a51"
df["Performance"] = df["Marks"].apply(
    get_performance
)

df
```

---

# Analyze Performance Categories

Count each category:

```python id="p18a52"
df["Performance"].value_counts()
```

We can also group the categories:

```python id="p18a53"
df.groupby(
    "Performance"
)["Marks"].agg(
    ["count", "mean", "min", "max"]
)
```

df["Rank"] = df["Marks"].rank(
